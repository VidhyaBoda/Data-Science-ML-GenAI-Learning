# Pandas Data Selection-Practice Question's Answer:-

```python
import pandas as pd

Dataset={ 
            "Name":["Amit","Riya","Sumit","Pooja"],
            "Math":[78,89,7,90], 
            "Science":[88,76,95,80] 
        }

df = pd.DataFrame(Dataset)

print(df)
```

# Q-1: Use Above Dataset and solve this problems.

```python
# (i) Select Only Math column.

print(df["Math"])
```

```python
# (ii) Select only the Name and Science columns.

print(df[['Name','Science']])
```

```python
# (iii) Select the first 2 rows using iloc.

df.iloc[2]
```

```python
# (iv) Select rows with index 1 and 2.

df.iloc[[1,2]]
```

```python
# (v) Get all students who scored more than 70 in Math.

df[df["Math"]>70]
```

```python
# (vi) Get students with Math>70 and Science>80.

print(df[(df["Math"]>70) & (df["Science"]>80)])
```

```python
# (vii) Get students with Math>85 or Science>90.

print(df[(df["Math"]>85) | (df["Science"]>90)])
```

```python
# (viii) Select Name and Science of the first 2 Students.

df.loc[[0,1], ["Name","Science"]]
```

# Q-3: 

```python
data = { 
        "Player": ["Virat", "Rohit", "Dhoni", "Sachin", "Hardik"], 
        "Runs": [78, 102, 45, 120, 65], 
        "Wickets": [0, 1, 2, 0, 3], 
        "Team": ["India", "India", "India", "India", "India"] 
        }
df = pd.DataFrame(data)

# print(df)
# Select only the Runs column.

print(df["Runs"])
```

```python
# (1) Select Player and Runs columns.

print(df[["Player","Runs"]])
```

```python
# (2) Select first 3 rows using iloc.

print(df.iloc[0:3])

df
```

```python
# (3) Select last 2 players using loc.

print(df.loc[3:4])
```

```python
# (4) Show players who scored more than 70 runs.

print(df[df["Runs"]>70])
```

```python
# (5) Show players who scored more than 50 runs and took at least 1 wicket.

print(df)

df[(df["Runs"]>50) & (df["Wickets"]>=1)]
```

# Q-4: 



```python
data = { 
        "FlightNo": [101, 102, 103, 104, 105], 
        "Airline": ["IndiGo", "Air India", "SpiceJet", "GoAir", "Vistara"], 
        "Price": [4500, 5200, 3900, 6100, 7000], 
        "Duration": [2.5, 3.0, 2.0, 2.8, 3.5] 
        }

df = pd.DataFrame(data)
print(df)

# 1. Select the Airline column.
print("-----------")
print(df["Airline"])
```

```python
# 2. Select rows of flights 101 and 104 only.

print(df[(df["FlightNo"]==101) | (df["FlightNo"]==104)])
```

```python
# 3. Show flights with price greater than 5000.

df[df["Price"]>5000]
```

```python
# 4. Show flights with duration less than 3 hours and price under 6000.

df1 = df[(df["Duration"]<3)  &  (df["Price"]<6000)]

print(df1)


print("----------")
df2 = df.loc[(df["Duration"]<3)  &  (df["Price"]<6000)]

print(df2)
```

```python
 data = { 
        "Product": ["Laptop", "Mobile", "Shoes", "Watch", "Headphones"], 
        "Price": [60000, 20000, 4000, 7000, 2000], 
        "Stock": [12, 30, 50, 20, 100] 
        }

df = pd.DataFrame(data)

print(df)
# 1. Select the Price column.

print("---------")
df["Price"]
```

```python
# 2. Select Product and Stock columns.

print(df[["Product","Stock"]])
```

```python
# 3. Show products costing more than 10,000.

print(df[df["Price"] >10000])
```

```python
# 4. Show products with price < 10,000 and stock > 20.

df.loc[(df["Price"]<10000) & (df["Stock"]>20)]
```
