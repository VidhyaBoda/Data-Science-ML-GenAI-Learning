# 1.Adding and removing columns.

```python
# Add column:df['new_col']=values

# Remove column:df.drop('colname',axis=1,inplace=True)

import pandas as pd

df=pd.DataFrame({
    "Name":['Rahul','Aman','Priya'],
    'Marks':[85,90,78]
})

# add new column
df['Grade']=['B','A','C']
print(df)

# remove column

df.drop('Marks',axis=1,inplace=True)     #OR df = df[df.drop('Marks',axis=1)] 
# print(df)
df
```

# 3.Sorting

```python
# sort_values(by='column')->kisi column ke hisab se sort.

df.sort_values(by='age',ascending=False,inplace=True)
print(df)
```

# 4.Resetting Index

```python
# data filter/sort krne ke bad index bigad jata hai ->reset krna padta hai.

df.reset_index(drop=True,inplace=True)
print(df)
```

# 5. Setting Index

```python
# kisi column ko index bana sakte ho

df.set_index("Name",inplace=True)
print(df)
```

# Practice Task's Solution:

# Task-1: Add/Remove Columns

```python
# Create a DataFrame:
import pandas as pd

data={ 
        "Name":["Rahul","Aman","Priya"], 
        "Age":[21,19,22], 
        "Marks":[85,92,76]
     }
df = pd.DataFrame(data)
df

# ->Add a new column Grade based on Marks:
# Marks>=90->A
# Marks>=80->B
# Else->C

df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x>=90 else "B" if x >= 80 else "C")

print(df)
# ->Remove the Age column.
```

# Task-2:Derived Column

```python
# Create a DataFrame:

data={ 
        "Product":["Pen","Book","Bag"], 
        "Price":[10,50,500], 
        "Quantity":[5,2,1]
    }

df = pd.DataFrame(data)
df

# ->Add a new column Total=Price*Quantity

df['Total'] = df["Price"] * df["Quantity"]
df


# ->Add another column DiscountPrice=90% of Total.

df["DiscountPrice"] = (df["Total"]/100)*90
df
```

# Task-3: Sorting

```python
# create DataFrame:

data={ "Name":["Ravi","Sneha","Kiran"], 
       "Age":[25,28,22], 
       "Salary":[40000,50000,35000] 
     }

df = pd.DataFrame(data)
df
# ->Sort by Salary in descending order.

df = df.sort_values(by = "Salary", ascending = False)
df
# ->Sort by Age in ascending order.
df.sort_values(by = "Age", ascending = True, inplace = True)
df
```

# Task-4:Restting and setting Index.

```python
# create DataFrame:

data={ 
        "City":["Delhi","Mumbai","Chennai"], 
        "Population":[19000000,20000000,10000000]
    }
df = pd.DataFrame(data)
df

# ->After sorting by Population, reset the index.

df.sort_values(by = "Population",inplace = True)
df

df.reset_index(drop = True, inplace = True)
df

# ->Set City as the index.

df.set_index("City", inplace=True)
df
```
