```python
# property decorator(@property)

class Product:
    def __init__(self,price):
        self.__price=price

    @property
    def price(self):
        return self.__price
    
    @price.setter
    def price(self,value):
        if value>0:
            self.__price=value
        else:
            print("Invalid Price!")

p=Product(500)
print(p.price) #getter call
p.price=100 #setter call
print(p.price)
```

```python
class ATM:

    def __init__(self, balance, pin):
        self.__balance = balance   # private variable
        self.__pin = pin           # private variable

    def check_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        else:
            return "Wrong PIN!"

    def withdraw(self, pin, amount):
        if pin == self.__pin:
            if 0 < amount <= self.__balance:
                self.__balance -= amount   # balance se minus karna hai
                return f"Withdrawn {amount}, Remaining {self.__balance}"
            else:
                return "Insufficient Balance!"
        else:
            return "Wrong PIN!"
        
atm = ATM(5000, 1234)
print(atm.check_balance(1234))     # 5000
print(atm.withdraw(1234, 2000))    # Withdrawn 2000, Remaining 3000
print(atm.check_balance(1234))     # 3000
print(atm.withdraw(1234, 4000))  # Insufficient Balance!
print(atm.check_balance(11112))     # Wrong PIN!
```

# Abstraction in Python – Axe Wali Clarity


---

1
👉 Abstraction ka matlab hai:

Sirf important cheez dikhana, details chhupana.

What to do dikhana, par How to do chhupana.


📌 Example (Life me):

Jab tu bike chalata hai: tu bas accelerator ghumata hai → speed badh jaati hai.

Tu ko pata nahi andar fuel injection, piston, combustion kaise kaam karte hain.
👉 Ye hi abstraction hai: user ko sirf zaroori cheez dikhao, andar ki detail mat dikhao.



---

2. Python Me Abstraction Kaise Karte Hain?

Python me abstraction ko ABC (Abstract Base Class) se implement karte hain.

abc module ka use hota hai.

Abstract class → ek aisi class jisme abstract methods (methods jinka sirf naam hai, body nahi).

Child class ko un methods ko implement karna hi hoga.

```python
from abc import ABC,abstractmethod

#abstract class
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# child classes must implement abstract methods

class Car(Vehicle):
    def start(self):
        return "car is starting"
    
    def stop(self):
        return "car is stopped"
    
class Bike(Vehicle):

    def start(self):
        return "Bike is running"
    
    def stop(self):
        return "bike is stoped"
    
vehicles=[Car(),Bike()]

for v in vehicles:
    print(v.start())
    print(v.stop())
```

Vehicle = abstract class (rule-book).

start aur stop = abstract methods (rule define).

Car aur Bike = un rules ko apne tareeke se implement kar rahe hain.

```python
from abc import ABC,abstractmethod

class ATM(ABC):

    @abstractmethod
    def withdraw(self,amount):
        pass

class SBI(ATM):

    def withdraw(self, amount):
        return f"SBI ATM: Withdraw {amount}"
    
class HDFC(ATM):

    def withdraw(self, amount):
        return f"HDFC ATM: Withdraw {amount}"
    

# user ko bas rule dikhte hai andar ka system alag hota hai

atm=HDFC()
print(atm.withdraw(1000))

atm=SBI()
print(atm.withdraw(20000))

atm=HDFC()
print(atm.withdraw(66))

atm=HDFC()
print(atm.withdraw(106655400))

atm=HDFC()
print(atm.withdraw(5240))
```

Inheritance in Python 


---

1. Meaning 

👉 Inheritance ka matlab hai:
Ek class (child) dusre class (parent) ke properties aur methods ko inherit (use) kar sakti hai.

📌 Life Example:

Beta apne baap ki property inherit karta hai.

Baap ke ghar, gaadi, zameen → beta use kar sakta hai.

Lekin beta apne khud ke naye cheezein bhi add kar sakta hai.



---

2. Types of Inheritance in Python

1. Single Inheritance – ek parent, ek child


2. Multilevel Inheritance – parent → child → grandchild


3. Multiple Inheritance – ek child ke do parent

```python
# class polymorphism(method overriding)

# same method name,lekin child class me uska behiour alag hota hai.

class Animal:
    def sound(self):
        print("Animals Make Sound.")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

class Cat(Animal):
    def sound(self):
        print("Car Makes Sound")

# Polymorphism in action

animals=[Dog(),Cat()]

for a in animals:
    a.sound()


# ek hi method sound() lekin alg objects ke liye result alag
```

```python
# Ans-6

from abc import ABC,abstractmethod

class Payment:

    @abstractmethod
    def make_payment(self,amount):
        pass

class CreditCard:
    def make_payment(self,amount):
        print(f"Paid {amount} from CreditCard. ")

class PayPal:
    def make_payment(self,amount):
        print(f"Paid {amount} from PayPal.")

class UPI:
    def make_payment(self,amount):
        print(f"Paid {amount} from UPI.")

    def transaction(self,amount,currency="INR"):
        print(f"Transaction of {amount} {currency} via UPI")

    def __add__(self,other):
        return f"Combined Payment using {self.__class__.__name__}+{other.__class__.__name__}"
    
def processing_payment(payment_method,amount):
    payment_method.make_payment(amount)


cc=CreditCard()
pp=PayPal()
uu=UPI()

processing_payment(cc,1000)
processing_payment(pp,500)
processing_payment(uu,200)

uu.transaction(100)
uu.transaction(250,"USD")

print(uu+pp)
    
```

```python
# (B) Class Method

# ye class ke data ke sath kaam krta hai.
# isme @classmethod decorator lagana padta hai
# first argument cls hota hai

class Demo:
    count=0

    def __init__(self):
        Demo.count+=1

    @classmethod
    def get_count(cls):
        print("Object Created:",cls.count)

d1=Demo()
d2=Demo()
d3=Demo()
Demo.get_count()
```

3.Super()

super() parent class ke method ko call krne ke liye use hota hai.
ye mainly inheritance me kaam ata hai.


```python
class Parent:
    def show(self):
        print("This is Parent Class.")

class Child(Parent):
    def show(self):
        super().show() #parent ka method call 
        print("this is child class")
c=Child()
c.show()

# super().show() ne pahele parent ka show() call liya fir apna.
```

```python
# 3.__str__-->String representation

# object ko human readable string banata hai(jab print() karte ho)

class Book:
    def __init__(self,title):
        self.title=title

    def __str__(self):
        return f"Book:{self.title}"

b=Book("Python OOP")
print(b)
```

```python
# 4.__len__->length return

# len() function ke liye object ka length batata hai.

class Course:
    def __init__(self,students):
        self.students=students
    def __len__(self):
        return len(self.students)

c=Course(["A","B","C"])
print(len(c))
```

```python
# 5.__add__->operator overloading

# object ko + operator ke sath add karne deta hai

class Bank:
    def __init__(self,balance):
        self.balance=balance

    def __add__(self,other):
        return self.balance+other.balance

b1=Bank(1000)
b2=Bank(500)
print(b1+b2)
        
```

```python
# 6.__eq__->Equality check(==)
# objects ko == ke sath compare krne ke liye

class Pen:
    def __init__(self,color):
        self.color=color
    
    def __eq__(self,other):
        return self.color==other.color
    
p1=Pen("blue")
p2=Pen("blue")
print(p1==p2)
```

```python
# 7.__getitem__->Indexing Support

# Object ko list ki tarah indexing karne ke liye

class Numbers:
    def __init__(self,nums):
        self.nums=nums
    
    def __getitem__(self,index):
        return self.nums[index]
n=Numbers([10,20,30])
print(n[1])
```
