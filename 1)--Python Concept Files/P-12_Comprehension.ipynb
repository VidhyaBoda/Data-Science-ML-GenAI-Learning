# 1️⃣ What is List Comprehension?

Comprehension ek short and elegant way hai to create lists, sets, or dictionaries in one line,
without using multiple loops or append().

List comprehension is a short and powerful way to create lists in Python using a single line of code instead of writing long loops.

A compact syntax for creating a new list by iterating, filtering, and transforming data.

# Syntax

[expression for item in iterable]

# With condition:

[expression for item in iterable if condition]

# 2️⃣ Without List Comprehension (Traditional Way)

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for n in numbers:
    squares.append(n * n)

print(squares)
```

# 3️⃣ Using List Comprehension

```python
numbers = [1, 2, 3, 4, 5]
squares = [n * n for n in numbers]
print(squares)
```

# 4️⃣ List Comprehension with Condition

```python
# Example: Even numbers only

nums = [1, 2, 3, 4, 5, 6]

evens = [n for n in nums if n % 2 == 0]
print(evens)
```

# 5️⃣ If–Else in List Comprehension

# Syntax

[expression_if_true if condition else expression_if_false for item in iterable]


```python
# Example: Even → "Even", Odd → "Odd"
nums = [1, 2, 3, 4, 5]

result = ["Even" if n % 2 == 0 else "Odd" for n in nums]
print(result)
```

# 6️⃣ Nested List Comprehension

```python
# Example: Flatten a list

matrix = [[1, 2], [3, 4], [5, 6]]

flat = [num for row in matrix for num in row]
print(flat)
```

# 7️⃣ List Comprehension with Strings

```python
# Example: Get vowels
word = "python"

vowels = [ch for ch in word if ch in "aeiou"]
print(vowels)
```

# 8️⃣ List Comprehension with range()

```python
nums = [n for n in range(1, 11)]
print(nums)
```

# 9️⃣ List Comprehension vs map() + lambda

```python
# Using map()

squares = list(map(lambda x: x * x, range(5)))

# Using List Comprehension

squares = [x * x for x in range(5)]


print(squares)
# ✔ List comprehension is more Pythonic and readable.
```

# 1️⃣3️⃣ Real-Life Examples

```python
# 1) Filter passed students
marks = [35, 78, 90, 42, 30]

passed = [m for m in marks if m >= 40]
print(passed)
```

```python
# 2) Convert Celsius to Fahrenheit
celsius = [0, 20, 30]

fahrenheit = [(c * 9/5) + 32 for c in celsius]
print(fahrenheit)
```

# 3️⃣ With Condition (if)

```python
even=[x for x in range(10) if x%2==0]
print(even)
```

# 4️⃣ With Else Condition

```python
nums=["Even" if x%2==0 else "Odd" for x in range(10)]
print(nums)
```

# 5️⃣ Nested Loop Comprehension

```python
pairs=[(x,y) for x in [1,2,3] for y in [4,5,6]]
print(pairs)
```

# 6️⃣ Dictionary Comprehension

```python
# squares={}

# for i in range(5):
#     squares[i]=i*i

# print(squares)

squares={i:i*i for i in range(5)}
print(squares)
```

With Condition in Dictionary

```python
even_square={i:i*i for i in range(10) if i%2==0}
print(even_square)
```

# 7️⃣ Set Comprehension

```python
nums={x for x in range(10)}
print(nums)
```

With Condition

```python
odd={x for x in range(10) if x%2!=0}
print(odd)
```

# 8️⃣ Generator Comprehension (Skip topic - in ML)

Same as list comprehension, but uses () instead of []

Does not store all elements in memory (memory-efficient)

```python
gen=(x*x for x in range(10))


# access value one by one
for value in gen:
    print(value)
```

Q-1 Create a list of all numbers from 1 to 10 using list comprehension.

Q-2 Create a list of squares of numbers from 1 to 10.

Q-3 Create a list of even numbers from 1 to 20.

Q-4 Create a list of cubes of odd numbers from 1 to 10.

Q-5 Convert all words in a list to uppercase using comprehension.

Q-6 Create a dictionary of numbers and their cubes (1–5).

```python
# Q-1 Create a list of all numbers from 1 to 10 using list comprehension.

numbers = [x for x in range(1,11)]
print(numbers)
```

```python
# Q-2 Create a list of squares of numbers from 1 to 10.

squares = [x**2 for x in range(1,11)]
print(squares)
```

```python
# Q-3 Create a list of even numbers from 1 to 20.

# Not Valid syntax 
evenNum = [x%2==0 for x in range(1,21)]
print(evenNum)

# Valid Syntax
evenNum = [x for x in range(1,21) if x%2==0]
print(evenNum)
```

```python
# Q-4 Create a list of cubes of odd numbers from 1 to 10.

cubes = [x**3 for x in range(1,11) if x%2 != 0]
print(cubes)
```

```python
# Q-5 Convert all words in a list to uppercase using comprehension.

words = ["Python", "Data Science","Machine learning"]

uppercase = [word.upper() for word in words]
print(uppercase)
```

```python
# Q-6 Create a dictionary of numbers and their cubes (1–5).

cubes = {x:x**3 for x in range(1,6)}
print(cubes)
```

```python
# Q-7 Create a dictionary of numbers from 1–10 with even numbers only.

evenNum = {x:x for x in range(1,11) if x%2==0}
print(evenNum)
```

```python
# Q-8 Create a list of tuples (x, y) for x in [1,2,3] and y in [4,5,6].

pairs = [(x,y) for x in [1,2,3] for y in [4,5,6]]
print(pairs)
```

```python
# Q-9 Use a generator comprehension to print squares of 1–10.

squareGen = (x*x for x in range(10))

for value in squareGen:
    print(value)
```
