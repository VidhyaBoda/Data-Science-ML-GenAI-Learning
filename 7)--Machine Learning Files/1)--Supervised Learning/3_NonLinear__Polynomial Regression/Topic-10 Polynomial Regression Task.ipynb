<h4>For the givencsv, write a python program satisfying following tasks to demonstrate application of machine learning through 
multiple linear regression as follows – </h4>
<h3>Dataset:</h3>
<h4>RealEstate.csv</h4>
<h3>Steps to Complete the Task:</h3>
<h4>
1. Import required libraries.<br>
2. Load RealEstate dataset, create a dataframe and check datatypes of its attributes using appropriate method.<br>
3. Remove ‘No’ column from the dataframe.<br>
4. Check for any null values in features using appropriate method.<br>
5. Create feature variables x and y as given above.<br> 
6. Create training and testing sets of feature variables with 70% of data for training and with random state of 110.<br>
7. Create and fit regression model using appropriate method.<br>
8. Use testing set created in step 6 to find and print the prediction of the outcome.<br>
9. Find and print coefficient and mean squared error of the regression model.</h4>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
# 1. Load Dataset

df = pd.read_csv("RealEstate.csv")
# df
df.head()
# df.info()
# df.describe()
```

```python
# 2. Check datatypes

print(df.dtypes)
```

```python
# 3. Drop "No"

df = df.drop("No", axis = 1)
df.head()
df
```

```python
# 4. Check null values

print(df.isna().sum())
```

```python
# 5. Featurs & Targets 

X = df.drop("Y house price of unit area", axis = 1)
y = df["Y house price of unit area"]

print(X.shape, y.shape)
```

```python
# 6. train_test_split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
     X, y, test_size = 0.7, random_state = 110)
  
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# OPTIONAL STEP


from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

from sklearn import metrics

for i in range(1,11):
    poly = PolynomialFeatures(degree = i)
    X_poly = poly.fit_transform(X)
    model = LinearRegression()
    model.fit(X_poly, y)
    y_pred = model.predict(X_poly)
    print(f"R2_score:  {i} {metrics.r2_score(y_pred, y)}")
```

```python
# 7. Polynomial transformation (IMPORTANT)

poly = PolynomialFeatures(degree = 10)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

```python
# 8. Train Model

model = LinearRegression()
model.fit(X_train_poly, y_train)
```

```python
# 9. Prediction

y_pred = model.predict(X_test_poly)
print("Prediction:", y_pred)
```

```python
# 10. Metrics 

print("Coefficient:",model.coef_)
print("Intercept:", model.intercept_)

print("Mean Suqared Error:", mean_squared_error(y_test, y_pred))

print("Mean Absolute Error:", mean_absolute_error(y_test,y_pred))

print("Accuracy:",metrics.r2_score(y_test,y_pred))
```

```python
# ----(OPTIONAL)-----For Model Saving / Saving Python objects to a file for later use.

import pickle

with open("RealEstate.pkl", "wb") as file:
    pickle.dump((poly,model), file)
    
print("Polinomial Model Saved Successfully")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
# 1. Load Dataset

df = pd.read_csv("car.csv")
df.head()
df.info()
df.describe()
```

```python
# 2. Check Datatypes

print(df.dtypes)
```

```python
# 3. Fill missing values with mean

print(df.isna().sum())
```

```python
# OPTIONAL STEP

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn import metrics

for i in range(1,15):
    poly = PolynomialFeatures(degree = i)
    X_poly = poly.fit_transform(X)
    model = LinearRegression()
    model.fit(X_poly, y)
    y_pred = model.predict(X_poly)
    print(f"R2 Score: {i} {metrics.r2_score(y_pred, y)}")
```
