# Task 1: Create a Series

```python
import pandas as pd
```

```python
# 1. Create a Pandas Series from this tuple:
# ("Python", "Django", "Pandas", "NumPy")

data = ["Python", "Django", "Pandas", "NumPy"]
print(pd.Series(data))
```

```python
# 2. Set the index manually as: ['A', 'B', 'C', 'D']

series = pd.Series(data,index=['A', 'B', 'C', 'D'])

print(series)
```

# Task 2: Create a Simple DataFrame


```python
# 1. Create a DataFrame from the dictionary:
# {
# "Name": ["Amit", "Riya", "Sumit"],
# "Age": [20, 21, 22],
# "Marks": [80, 85, 90]
# }

import pandas as pd

data = {
        "Name": ["Amit", "Riya", "Sumit"],
        "Age": [20, 21, 22],
        "Marks": [80, 85, 90]
}

# 2. Print the complete DataFrame

df = pd.DataFrame(data)
print("Dataframe:")
print(df)

# 3. Print only the Name column
print("\n Name Column:")
print(df["Name"])
```

```python
# Task 3: Rows & Columns Count
# 1. Print:
# o Number of rows
# o Number of columns
# of the DataFrame

import pandas as pd

data = {
        "Name": ["Amit", "Riya", "Sumit"],
        "Age": [20, 21, 22],
        "Marks": [80, 85, 90]
}

df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)

print('\nNumber of rows:')
print(df.shape[0])

print("\nNumber of columns:")
print(df.shape[1])
```
