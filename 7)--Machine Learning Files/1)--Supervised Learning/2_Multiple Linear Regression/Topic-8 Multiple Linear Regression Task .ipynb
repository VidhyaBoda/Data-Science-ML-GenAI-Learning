# Multiple Linear Regression – RealEstate Dataset

```python
# 1. Import required libraries.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.linear_model import LinearRegression
```

```python
# 2. Load RealEstate dataset, create a dataframe and check datatypes of its attributes using appropriate method.

df = pd.read_csv("RealEstate.csv")

print(df.head())
print("\n-----------------------------")
print(df.info())
print("\nData Types:---------------------------------")
print(df.dtypes)
df
```

```python
# 3. Remove ‘No’ column from the dataframe.

df.drop("No", axis = 1, inplace = True)
```

```python
# 4. Check for any null values in features using appropriate method.

print("\nNull Values:---------------------")
print(df.isnull().sum())
```

```python
# 5. Create feature variables x and y as given above.

# Typical RealEstate dataset columns look like:

# X1 transaction date
# X2 house age
# X3 distance to MRT
# X4 number of convenience stores
# X5 latitude
# X6 longitude
# Y house price of unit area


X = df.drop("Y house price of unit area", axis = 1)
y = df["Y house price of unit area"]
```

```python
# 6. Create training and testing sets of feature variables with 70% of data for training and with random state of 110.

from sklearn.model_selection import train_test_split

# 70% of data for training and with random state of 110.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.7, 
    random_state=110
)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# 7. Create and fit regression model using appropriate method.

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
# 8. Use testing set created in step 6 to find and print the prediction of the outcome.

y_pred = model.predict(X_test)
print("\nPredicted Values:--------------")
print(y_pred)
```

```python
# 9. Find and print coefficient and mean squared error of the regression model.

print("Regression Coefficients / Slope:", model.coef_)
print("\nIntercept:", model.intercept_)

mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test, y_pred)

print("\nMean Squared Error:", mse)
print("\nMean Absolute Error:", mae)
```

```python
# ----(OPTIONAL)-----For Model Saving / Saving Python objects to a file for later use.

import pickle

# Save only the trained model (no poly here)
with open("RealEstate.pkl", "wb") as file:
    pickle.dump(model, file)

print("Multiple Linear Regression Model Saved Successfully")
```

# Linear Regression – Wine Quality Prediction

```python
# 1. Do the required process in the data before making a model.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error 
```

```python
# 2. Load Dataset

df = pd.read_csv("winequality.csv")
df.head()
```

```python
# 3. Check Dataset Information

print(df.info())
print("--------------------------")
print(df.describe())
print("------------------------------")
print(df.dtypes)
```

```python
# 4. Handle Missing Values 

print(df.isna().sum())

df = df.fillna(df.mean(numeric_only=True))
```

```python
# 5. Define Feature Variables (X) and Target (y)

# In wine dataset:
# Features → chemical properties
# Target → quality

X = df.drop(["quality",'Id'],axis = 1)
y = df["quality"]
```

```python
# 6. Split Data into Training and Testing Sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.7, random_state = 42)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# 7. Create and Train Linear Regression Model

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# 8. Predict Wine Quality on Test Data

y_pred = model.predict(X_test)

print("Predicted Values:-----------------\n",y_pred)
```

```python
# 9. Model Evaluation

print("Coefficient:---------\n", model.coef_)
print("\nIntercept:--------\n", model.intercept_)

mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:\n", mse)
```

```python
# 10. Predict the quality of red wine for the following data:

# Input Values:
    
# | Feature              | Value |
# | -------------------- | ----- |
# | fixed acidity        | 8     |
# | volatile acidity     | 0.4   |
# | citric acid          | 0.40  |
# | residual sugar       | 15    |
# | chlorides            | 0.048 |
# | free sulfur dioxide  | 40    |
# | total sulfur dioxide | 150   |
# | density              | 0.99  |
# | pH                   | 3     |
# | sulphates            | 0.45  |
# | alcohol              | 10.5  |

# sample = [[8, 0.4, 0.40, 15, 0.048, 40, 150, 0.99, 3, 0.45, 10.5]]

sample = pd.DataFrame(
    [[8,0.4,0.40,15,0.048,40,150,0.99,3,0.45,10.5]],
    columns=X.columns
)


prediction = model.predict(sample)

print("Predicted Wine Quality:",prediction[0])
```
