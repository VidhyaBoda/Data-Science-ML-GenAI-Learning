<h1 style="text-align: center;">Introduction to Multiple Regression</h1>

<h4>🧐 Have you ever wondered how multiple factors influence a single outcome, like how a house's price depends on both size and location?</h4>

<h3>➡️ Explain:</h3>
<h4>Just like predicting house prices based on multiple factors, machines can also learn from multiple inputs to predict an outcome. When a model considers multiple independent variables, it's called <b>Multiple Linear Regression</b>.

<h3>🖥️ Definition:</h3>
<h4><b>Multiple Linear Regression</b> is a type of regression where the output depends on two or more independent variables.</h4>


<h3>📝 Key Characteristics:</h3>
<h4>📌 <b>Multiple Linear Regression:</b></h4>
<ul>
    <li>Uses two or more independent variables to predict an outcome.</li>
    <li>Assumes a linear relationship between the input variables and the output.</li>
    <li>Example: Predicting a house's price using area, location, and number of rooms.</li>
</ul>

<h3>A. Multiple Linear Regression (Predicting with Multiple Features)</h3>

<h3>➡️ Explain:</h3>
<h4>Imagine you want to estimate the price of a car based on its mileage, age, and brand. Instead of using just one factor, you consider multiple ones—this is <b>Multiple Linear Regression</b>.</h4>

<h3>🖥️ Definition:</h3>
<h4>Multiple Linear Regression is a type of regression where multiple independent variables are used to predict a single continuous outcome.</h4>

<h3>📌 Examples of Multiple Linear Regression Problems:</h3>
<ul>
    <li>House Price Prediction → Based on area, location, and number of bedrooms.</li>
    <li>Student Performance Prediction → Based on study hours, sleep, and diet.</li>
    <li>Car Price Prediction → Based on mileage, brand, and manufacturing year.</li>
    <li>Sales Forecasting → Based on advertising spend, season, and competitor pricing.</li>
</ul>

<h3>🎯 Quick Summary:</h3>
<h4>🔹 If multiple factors influence an outcome in a linear way → It’s <b>Multiple Linear Regression</b>.</h4>

<h3><center>Train the Linear Regression Model</h3></center>

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

<h3>🔹 What happens here? </h3>

<h4>
    from sklearn.linear_model import LinearRegression → Imports Multiple Linear Regression from scikit-learn. <br>
    model = LinearRegression() → Creates an instance of the Multiple Linear Regression model. <br>
    model.fit(x_train, y_train) → Trains the model using multiple input features (x_train) and the target variable (y_train). 
</h4>

<h3>📌 How It Works? </h3>

<h4>The Multiple Linear Regression algorithm finds the best-fit equation: <br>
𝑦 = 𝑚₁𝑥₁ + 𝑚₂𝑥₂ + ... + 𝑚ₙ𝑥ₙ + 𝑐 <br>
where: <br>
y = Predicted output (Target Value) <br>
x₁, x₂, ..., xₙ = Independent variables (Feature Values) <br>
m₁, m₂, ..., mₙ = Coefficients (Weights assigned to each feature) <br>
c = Intercept (starting value when all x = 0) <br> <br>

🔹 The <b>fit()</b> method calculates the optimal coefficients (m values) and intercept (c) using the <b>Least Squares Method</b>. <br>
🔹 The model considers the combined effect of multiple features to make accurate predictions. 
</h4>


```python
print("Coefficients: ",model.coef_)
print("Intercept: ",model.intercept_)
```
