# TASK SET 1: Single Line Plot (Very Easy) 

Task 1 — Student Performance Line 

```python
#  Create x = [1,2,3,4,5] (tests) 
#  Create y = [45, 50, 60, 70, 80] (marks) 
#  Plot a line graph 
#  Use: 
# o marker='o' 
# o color='blue' 
# o linestyle='dashed' 
#  Add title, x-label, y-label, grid 

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [45, 50, 60, 70, 80]

plt.plot(x,y, marker='o', linestyle = 'dashed', color = 'blue')

plt.title('Student Performance Chart')

plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')

plt.grid()
plt.show()
```

Task 2 — Temperature Change 

```python
#  Create temperature values for 7 days 
#  Rotate x-axis labels by 30 degrees 
#  Use marker='s' 
#  Show grid 

import matplotlib.pyplot as plt

days = ["Mon","Tue", "Wed","Thu","Fri","Sat","Sun"]
temperature = [28, 30, 35, 31, 34, 37, 36]

plt.plot(days,temperature, marker = 's', )

plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')
plt.xticks(rotation = 30)
plt.show()
```

# TASK SET 3: Multiple Line Plot (Most Important)

Task 5 — Two Students Marks

```python
#  Student A marks → one line
#  Student B marks → second line
#  Use:
# o different colors
# o different markers
#  Add legend()

import matplotlib.pyplot as plt
import numpy as np

student_A = [85, 78, 90, 88, 92]
student_B = [80, 82, 85, 84, 89]

exam = np.arange(1, 6)

plt.plot(exam, student_A, marker='o',color='blue')
plt.plot(exam, student_B, marker='D',color='red')

plt.xlabel("Exam Number")
plt.ylabel("Marks")
plt.legend(loc='upper left', title = 'Student marks')
plt.title("Student Marks Comparison")
plt.show()
```

Task 6 — Company Sales Comparison

```python
#  x = [2019,2020,2021,2022,2023]
#  Company A sales → red line
#  Company B sales → green line
#  Legend position: upper left


import matplotlib.pyplot as plt

x = [2019,2020,2021,2022,2023]

company_a = [50, 55, 70, 65, 80]
company_b = [45, 60, 68, 72, 75]

plt.plot(x,company_a,color='red',marker='o',label='Company A')
plt.plot(x,company_b,color='green',marker='s',label='Company B')

plt.xlabel("Year")
plt.ylabel("Sales")

plt.title("Company Sales Comparison")

plt.legend(loc="upper left")

plt.xticks(rotation=30)

plt.show()
```

# TASK SET 4: Styling Practice (Very Easy)

Task 7 — Marker Change

```python
# Plot the same data 3 times, changing only:
#  marker (o, s, ^)
#  color

import matplotlib.pyplot as plt

x = [2019, 2020, 2021, 2022, 2023]
sales = [50, 65, 70, 60, 80]

plt.plot(x,sales, marker='o', color='blue', label='Style 1')
plt.plot(x,sales, marker='s', color='green', label='Style 2')
plt.plot(x,sales, marker='^', color='red', label='Style 3')

plt.xlabel("Year")
plt.ylabel("sales")

plt.title("Same Data with Different Styles")

plt.legend()

plt.show()
```
