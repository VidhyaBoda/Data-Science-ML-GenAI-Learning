<h1 style="text-align: center;">Introduction to Polynomial Regression</h1>

<h4>🧐 Have you ever wondered how multiple factors influence a single outcome, like how a house's price depends on both size and location?</h4>

<h3>➡️ Explain:</h3>
<h4>Just like predicting house prices based on multiple factors, machines can also learn from multiple inputs to predict an outcome. When a model considers multiple independent variables, it's called <b>Multiple Linear Regression</b>. If the relationship between inputs and output is nonlinear, we use <b>Polynomial Regression</b>.</h4>

<h3>🖥️ Definition:</h3>
<h4><b>Polynomial Regression</b> is a regression technique where the relationship between input and output is modeled as an nth-degree polynomial instead of a straight line.</h4>

<h3>📝 Key Characteristics:</h3>
<h4>📌 <b>Polynomial Regression:</b></h4>
<ul>
    <li>Uses polynomial features (e.g., squared, cubic terms) to model complex relationships.</li>
    <li>Can capture non-linear trends in data.</li>
    <li>Example: Predicting salary growth based on years of experience.</li>
</ul>

<h3>B. Polynomial Regression (Modeling Non-Linear Relationships)</h3>

<h3>➡️ Explain:</h3>
<h4>If you observe that salary growth doesn't increase linearly with experience (e.g., salary jumps after promotions), then a straight-line model won't work. Instead, a curved line (polynomial) fits better—this is <b>Polynomial Regression</b>.</h4>

<h3>🖥️ Definition:</h3>
<h4>Polynomial Regression is a type of regression where the relationship between input and output is modeled using higher-degree polynomials to capture non-linearity.</h4>

<h3>📌 Examples of Polynomial Regression Problems:</h3>
<ul>
    <li>Salary Prediction → Based on years of experience (non-linear growth).</li>
    <li>Disease Progression → Based on time and multiple health factors.</li>
    <li>Stock Market Trends → Capturing fluctuating trends over time.</li>
    <li>Traffic Flow Prediction → Based on time of day and weather conditions.</li>
</ul>

```python
y = mx + c
y = mx2+ mx + c
```

```python
mx + c ----> mx^2 + mx + c
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn import metrics
for i in range(1,30):
    poly = PolynomialFeatures(degree=i)
    x_poly = poly.fit_transform(x)
    model = LinearRegression()
    model.fit(x_poly,y)
    y_pred = model.predict(x_poly)
    print(f"R2 Score: {i} {metrics.r2_score(y_pred,y)}")
```

```python
poly = PolynomialFeatures(degree=21)
x_poly = poly.fit_transform(x)
model = LinearRegression()
model.fit(x_poly,y)
y_pred = model.predict(x_poly)

print("MAE: ",metrics.mean_absolute_error(y,y_pred))
print("MSE: ",metrics.mean_squared_error(y,y_pred))
print("R2 Score: ",metrics.r2_score(y,y_pred))

import pickle

with open('polynomial_model.pkl', 'wb') as file:
    pickle.dump((poly, model), file)

print("Model Saved Successfully!")
```

<h3>📌 How It Works? </h3>

<h4>The Polynomial Regression algorithm finds the best-fit curve using the equation: <br>
𝑦 = 𝑚₁𝑥 + 𝑚₂𝑥² + 𝑚₃𝑥³ + ... + 𝑚ₙ𝑥ⁿ + 𝑐 <br>
where: <br>
y = Predicted value (Target Variable) <br>
x, x², x³, ... xⁿ = Transformed feature values (Polynomial Terms) <br>
m₁, m₂, m₃, ... mₙ = Coefficients for each polynomial term <br>
c = Intercept (starting value when all x = 0) <br> <br>
