# Step 3: Import Libraries

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import MaxPooling2D
from tensorflow.keras.layers import Flatten
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
```

# Step 4: Load Dataset

TensorFlow can load images directly from folders.


```
train_ds = tf.keras.utils.image_dataset_from_directory(
    '/content/dataset/train',
    image_size=(128,128),
    batch_size=32,
    shuffle=True
)
```

# Step 5: Inspect Dataset Shape

```
for images, labels in train_ds.take(1):
    print(images.shape)
    print(labels.shape)
```

Expected:

(32,128,128,3)

Meaning:
- 32 images per batch
- Height = 128
- Width = 128
- RGB channels = 3


# Step 6: Normalize Images

Pixel values are 0–255.

We scale them to 0–1.


```
normalization = tf.keras.layers.Rescaling(1./255)

train_ds = train_ds.map(
    lambda x,y:(normalization(x),y)
)
```

# Step 7: Visualize Images

```
plt.figure(figsize=(8,8))

for images, labels in train_ds.take(1):

    for i in range(9):
        plt.subplot(3,3,i+1)
        plt.imshow(images[i])
        plt.axis("off")

plt.show()
```

# Step 8: Build CNN Architecture

Architecture:

Image
↓
Conv2D
↓
ReLU
↓
Pooling
↓
Conv2D
↓
Pooling
↓
Flatten
↓
Dense
↓
Output


```
model = Sequential()

model.add(
    Conv2D(
        32,
        (3,3),
        activation='relu',
        input_shape=(128,128,3)
    )
)

model.add(MaxPooling2D((2,2)))

model.add(
    Conv2D(
        64,
        (3,3),
        activation='relu'
    )
)

model.add(MaxPooling2D((2,2)))

model.add(
    Conv2D(
        128,
        (3,3),
        activation='relu'
    )
)

model.add(MaxPooling2D((2,2)))

model.add(Flatten())

model.add(Dense(128, activation='relu'))

model.add(Dropout(0.5))

model.add(Dense(1, activation='sigmoid'))
```

# Step 9: Model Summary

Always inspect shapes and parameters.


```
model.summary()
```

# Step 10: Compile Model

Binary Classification:

Cat vs Dog


```
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

# Step 11: Train Model

```
history = model.fit(
    train_ds,
    epochs=10
)
```

# Step 12: Plot Accuracy and Loss

```
plt.plot(history.history['accuracy'])
plt.title('Training Accuracy')
plt.show()
```

# Step 13: Evaluate Model

```
model.evaluate(train_ds)
```

# Step 14: Predict on New Image

```
img = tf.keras.utils.load_img(
    '/content/test.jpg',
    target_size=(128,128)
)

img_array = tf.keras.utils.img_to_array(img)

img_array = np.expand_dims(img_array, axis=0)

img_array = img_array / 255.0

prediction = model.predict(img_array)

print(prediction)
```

```
if prediction > 0.5:
    print("Dog")
else:
    print("Cat")
```
