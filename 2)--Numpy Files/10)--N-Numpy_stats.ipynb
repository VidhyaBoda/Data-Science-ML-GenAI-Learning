2. np.median() –Middle number after sort

```python
import numpy as np

arr = np.array([10, 50, 30, 20])
print(np.median(arr))   
```

3. mode (Most Frequent Value) –

```python
from scipy import stats
import numpy as np

arr = np.array([1, 2, 2, 3, 3, 3, 4,4,4,4,4,4])
print(stats.mode(arr))   # Output: mode = 3, count = 3
```

4. np.std() – Standard Deviation (डेटा कितना फैला हुआ है)

```python
arr = np.array([10, 20, 30, 40])
print(np.std(arr))   
```

5. np.var() – Variance (Standard deviation का square)

```python
print(np.var(arr))   # Output: 125.0
```

6. np.sum() –

```python
print(np.sum(arr))   # Output: 100
```

7. np.min() & np.max() –

```python
print(np.min(arr))   # 10
print(np.max(arr))   # 40
```

9. np.unique()

```python
arr = np.array([1,2,2,3,3,3])
print(np.unique(arr))  # [1 2 3]
```
