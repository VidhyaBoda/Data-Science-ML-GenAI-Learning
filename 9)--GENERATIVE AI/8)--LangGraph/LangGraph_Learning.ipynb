# 08--LangGraph
Stateful, graph-based LLM workflows and agents.

## Learning Flow
State → Nodes → Edges → Conditional Routing → Loops → Persistence → Human-in-the-loop → Agents → RAG → Multi-agent → Production

```python
def process(state):
    state['message'] = state['message'].upper()
    state['steps'].append('process')
    return state

state = {'message': 'hello langgraph', 'steps': []}
state = process(state)
print(state)
```

## Core Mental Model
**State + Nodes + Edges + Conditional Routing = Graph Workflow**

```python
graph = {
    'START': 'process',
    'process': 'decide',
    'decide': 'END'
}
for source, target in graph.items():
    print(f'{source} -> {target}')
```
