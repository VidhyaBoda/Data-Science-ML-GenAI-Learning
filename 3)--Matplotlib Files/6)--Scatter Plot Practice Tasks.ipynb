# Task 1 — Size Based on Data

```python
# Create:
#  study_hours
#  marks
# Set:
# sizes = marks * 5
# 👉 Higher marks = bigger points.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(7)

study_hours = np.random.randint(20,50,15)
marks = np.random.randint(20,50,15)

sizes = marks * 5

plt.scatter(study_hours,marks, s = sizes)

plt.xlabel("Study_hours")
plt.ylabel("Marks")

plt.title("Study hours vs Marks")

plt.show()
```

# Task 2 — Conditional Colors 

```python
# Apply colors using conditions:
#  marks < 40 → red
#  marks 40–70 → orange
#  marks > 70 → green
# Hint:
# colors = np.where(marks>70,'green',np.where(marks>40,'orange','red'))

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(5)

study_hours = np.random.randint(1, 10, 25)
marks = np.random.randint(20, 100, 25)

# Apply conditional colors

colors = np.where(marks>70,'green',np.where(marks>40,'orange','red'))

plt.scatter(study_hours,marks, c = colors, s =100)

plt.xlabel("Study_hours")
plt.ylabel("Marks")

plt.title("Study hours vs Marks")

plt.show()
```

# Task 3 — Alpha Comparison

```python
# Create 3 scatter plots:
#  alpha = 1
#  alpha = 0.5
#  alpha = 0.2
# 👉 Observe overlapping effects.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(50)

x = np.random.randn(50, 10, 200)
y = np.random.randn(50, 10, 200)

# alpha = 1
plt.scatter(x, y, alpha=1)
plt.title("Alpha = 1 (No Transparency)")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()


# alpha = 0.5
plt.scatter(x, y, alpha=0.5)
plt.title("Alpha = 0.5 (Medium Transparency)")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()


# alpha = 0.2

plt.scatter(x, y, alpha=0.2)
plt.title("Alpha = 0.2 (High Transparency)")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

# Task 5 — Three Groups Scatter

```python
# Create 3 groups:
#  Beginners
#  Intermediate
#  Advanced

# Each group should have:
#  different color
#  different size
#  label + legend title

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(10)

x_beg = np.random.randint(1,4,20)
y_beg = np.random.randint(10,40,20)

x_int = np.random.randint(4,7,20)
y_int = np.random.randint(40,70,20)

x_adv = np.random.randint(7,10,20)
y_adv = np.random.randint(70,100,20)

plt.scatter(x_beg,y_beg, c = 'orange', s = 60, label = "Beginners", marker = 'D')
plt.scatter(x_int,y_int, c = 'blue', s = 70, label = "Intermediate", marker = '*')
plt.scatter(x_adv,y_adv, c = 'green', s = 80, label = "Advanced", marker = 'o')

plt.xlabel("Practice Hours")
plt.ylabel("Skill Score")

plt.title("Skill Level Distribution")

plt.legend(loc="upper left",title = "Exprience Level")

plt.show()
```

# Task 6 — Add Mean Lines


```python
# Add average lines:
# plt.axhline(mean_y)
# plt.axvline(mean_x)
# 👉 Makes plots look professional

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(30)

# Sample data
x = np.random.randint(10, 100, 50)
y = np.random.randint(10, 100, 50)

# Calculate means
mean_x = np.mean(x)
mean_y = np.mean(y)

# Scatter plot
plt.scatter(x, y, alpha=0.6)

# Add mean lines
plt.axhline(mean_y, color='red', linestyle='--', label='Mean Y')
plt.axvline(mean_x, color='blue', linestyle='--', label='Mean X')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot with Mean Lines")

plt.legend()
plt.show()
```

# Task 7 — Bubble Chart

```python
# Task 7 — Bubble Chart
# Instead of random size:
# revenue = np.random.randint(1000,5000,size=50)
# sizes = revenue/10

import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100,1000,50)
y = np.random.randint(50,500,50)

revenue = np.random.randint(1000,5000,size=50)
sizes = revenue/10

plt.scatter(x,y, s = sizes, alpha = 0.6)

plt.xlabel("Marketing Spend")
plt.ylabel("Profit")

plt.title("Marketing Spend vs Profit")

plt.show()
```
