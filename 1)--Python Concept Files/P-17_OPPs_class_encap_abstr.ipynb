```python
# property decorator(@property)

class Product:
    def __init__(self,price):
        self.__price=price

    @property
    def price(self):
        return self.__price
    
    @price.setter
    def price(self,value):
        if value>0:
            self.__price=value
        else:
            print("Invalid Price!")

p=Product(500)
print(p.price) #getter call
p.price=100 #setter call
print(p.price)
```

```python
# Example:-

class ATM:

    def __init__(self, balance, pin):
        self.__balance = balance   # private variable
        self.__pin = pin           # private variable

    def check_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        else:
            return "Wrong PIN!"

    def withdraw(self, pin, amount):
        if pin == self.__pin:
            if 0 < amount <= self.__balance:
                self.__balance -= amount   # balance se minus karna hai
                return f"Withdrawn {amount}, Remaining {self.__balance}"
            else:
                return "Insufficient Balance!"
        else:
            return "Wrong PIN!"
        
atm = ATM(5000, 1234)
print(atm.check_balance(1234))     # 5000
print(atm.withdraw(1234, 2000))    # Withdrawn 2000, Remaining 3000
print(atm.check_balance(1234))     # 3000
print(atm.withdraw(1234, 4000))  # Insufficient Balance!
print(atm.check_balance(1111))     # Wrong PIN!
```

# Abstraction in Python – Axe Wali Clarity


---

1
👉 Abstraction ka matlab hai:

Sirf important cheez dikhana, details chhupana.

What to do dikhana, par How to do chhupana.


📌 Example (Life me):

Jab tu bike chalata hai: tu bas accelerator ghumata hai → speed badh jaati hai.

Tu ko pata nahi andar fuel injection, piston, combustion kaise kaam karte hain.
👉 Ye hi abstraction hai: user ko sirf zaroori cheez dikhao, andar ki detail mat dikhao.



---

2. Python Me Abstraction Kaise Karte Hain?

Python me abstraction ko ABC (Abstract Base Class) se implement karte hain.

abc module ka use hota hai.

Abstract class → ek aisi class jisme abstract methods (methods jinka sirf naam hai, body nahi).

Child class ko un methods ko implement karna hi hoga.

```python
from abc import ABC,abstractmethod

#abstract class
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# child classes must implement abstract methods

class Car(Vehicle):
    def start(self):
        return "car is starting"
    
    def stop(self):
        return "car is stopped"
    
class Bike(Vehicle):

    def start(self):
        return "Bike is running"
    
    def stop(self):
        return "bike is stoped"
    
vehicles=[Car(),Bike()]

for v in vehicles:
    print(v.start())
    print(v.stop())
```

Vehicle = abstract class (rule-book).

start aur stop = abstract methods (rule define).

Car aur Bike = un rules ko apne tareeke se implement kar rahe hain.

```python
from abc import ABC,abstractmethod

class ATM(ABC):

    @abstractmethod
    def withdraw(self,amount):
        pass

class SBI(ATM):

    def withdraw(self, amount):
        return f"SBI ATM: Withdraw {amount}"
    
class HDFC(ATM):

    def withdraw(self, amount):
        return f"HDFC ATM: Withdraw {amount}"
    

# user ko bas rule dikhte hai andar ka system alag hota hai

atm=HDFC()
print(atm.withdraw(1000))
```

5. Why Use Abstraction?

✔ Code clean hota hai.

✔ Reusability badhta hai.

✔ Har child class apna implementation deta hai.

✔ Large system me complexity kam hoti hai.


---

6. Abstraction vs Encapsulation 

Feature	Abstraction (What)	Encapsulation (How)

Definition	Sirf zaroori cheez dikhana, details chhupana	Data aur methods ko ek unit me bandh karna

Focus	Design level (what to do?)	Implementation level (how to do?)

Tools in Python	Abstract class, Interfaces (abc module)	Private/Public/Protected variables
Example	Bike ka accelerator → speed dikhata hai	Bike ka engine → andar data protected hai

```python
# Q-1: Create an abstract class Appliance with abstract method turn_on().
# ->Implement two subclasses: WashingMachine and AirConditioner.
# ->Each should define its own version of turn_on().

from abc import ABC, abstractmethod

class Appliance(ABC):
    
    @abstractmethod
    def turn_on(self):
        pass
    
class WashingMachine(Appliance):
    def turn_on(self):
        print("Washing Machine is now ON")
        
class AirConditioner(Appliance):
    def turn_on(self):
        print("Air Conditioner is now ON")
        
wm = WashingMachine()
ac = AirConditioner()

wm.turn_on()
ac.turn_on()
```

```python
# Q-2: Create an abstract class Instrument with abstract method play_sound().
# ->Implement subclasses Guitar and Piano.
# ->Demonstrate playing sounds.

from abc import ABC, abstractmethod

class Instrument(ABC):
    
    @abstractmethod
    def play_sound(self):
        pass
    
class Guitar(Instrument):
    def play_sound(self):
        print("Guitar is playing: Strum Strum")
        
class Piano(Instrument):
    def play_sound(self):
        print("Piano is playing : Plink plonk")

g = Guitar()
p = Piano()

g.play_sound()
p.play_sound()
```

```python
# Q-3: Create an abstract class Sport with abstract methods rules() and equipment().
# ->Implement subclasses Cricket and Football.
# ->Show how each sport has diffrent rules and equipment.

from abc import ABC, abstractmethod

class Sport(ABC):
    
    @abstractmethod
    def rules(self):
        pass
        
    @abstractmethod
    def equipment(self):
        pass
        
class Cricket(Sport):
    def rules(self):
        print("Cricket Rules: Each team has 11 players, runs are scored by hitting the ball.")
        
    def equipment(self):
        print("Cricket Equipment: Bat, ball, stumps, pads, helmet")
        
    
class Football(Sport):
    
    def rules(self):
        print("Football Rules: Each team has 11 players, goals are scored by kicking the ball.")
        
    def equipment(self):
        print("Football Equipment: Football, goalpost, shoes, shin guards")


sports = [Cricket(), Football()]

for s in sports:
    s.rules()
    s.equipment()
    
    print("-"*40)
```

```python
# Q-4: Create an abstract class FoodDelivery with abstract method deliver_order().
# ->Implement subclasses ZomatoDelivery and SwiggyDelivery.
# ->Pass Order id to the function.
# ->Demonstrate order delivery from different services.

from abc import ABC, abstractmethod
class FoodDelivery(ABC):
    
    @abstractmethod
    def deliver_order(self):
        pass
    
class ZomatoDelivery(FoodDelivery):
    def deliver_order(self,order_id):
        print(f"Zomato order is #{order_id} on the way")
        
class SwiggyDelivery(FoodDelivery):
    def deliver_order(self,order_id):
        print(f"Swiggy order is #{order_id} on the way")
        

deliveries = [ZomatoDelivery(), SwiggyDelivery()]

for services in deliveries:
    services.deliver_order(101)
```

```python
# Q-5: Create an abstract class UniversityMember with abstract method get_role()
# ->Implement subclasses Professor,Student and Staff.
# ->Each class should describe its role differently.


from abc import ABC, abstractmethod

class UniversityMember(ABC):
    
    @abstractmethod
    def get_role(self):
        pass
    
class Professor(UniversityMember):
    def get_role(self):
        return "Professor: Teaches students and conducts research"
            
class Student(UniversityMember):
    def get_role(self):
        return "Student: Attends classes and completes assignments"

class Staff(UniversityMember):
    def get_role(self):
        return "Staff: Manages administrative and support tasks"
    

members = [Professor(), Student(), Staff()]

for m in members:
    print(m.get_role())
    
```
