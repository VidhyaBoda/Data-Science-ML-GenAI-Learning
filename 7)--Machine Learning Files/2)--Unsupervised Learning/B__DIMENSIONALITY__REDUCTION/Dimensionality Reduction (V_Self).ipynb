
# Dimensionality Reduction Models

| Model                | Description                       |
| -------------------- | --------------------------------- |
| PCA                  | Principal Component Analysis      |
| Kernel PCA           | Non-linear PCA                    |
| Incremental PCA      | Large datasets                    |
| Sparse PCA           | Sparse features                   |
| t-SNE                | Visualization                     |
| UMAP                 | Advanced visualization            |
| Factor Analysis      | Latent factors                    |
| ICA                  | Independent Component Analysis    |
| Truncated SVD        | Sparse matrix reduction           |
| LDA (Topic Modeling) | Text dimensionality reduction     |
| NMF                  | Non-negative Matrix Factorization |
