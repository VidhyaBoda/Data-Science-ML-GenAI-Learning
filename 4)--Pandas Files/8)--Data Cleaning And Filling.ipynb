# Task 1

```python
import pandas as pd
```

```python
# Create a DataFrame: 
data = { 
        "Name": ["Raj", "Simran", None, "Ankit"], 
        "Age": [22, None, 25, None], 
        "Marks": [80, 90, None, 85] 
} 

df = pd.DataFrame(data)

print(df)

print("--------------")

# Do the following: 

# 1. Fill missing Name with "Unknown" 

df["Name"] = df["Name"].fillna("Unknown")
print(df)

print("--------------")

# 2. Fill missing Age with mean 

df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)

print("--------------")

# 3. Fill missing Marks with median 

df["Marks"] = df["Marks"].fillna(df["Marks"].median())
print(df)
```

# Task 2

```python
# Given: 
df = pd.DataFrame(
    { 
        "Product": ["Pen", "Pencil", "Eraser", None], 
        "Price": [10, None, 5, None] 
    }
) 

print(df)

print("--------------")

# Tasks: 

#  Fill missing Product with "Not Available"

df["Product"] = df["Product"].fillna("Not Available")
print(df)

print("--------------")

#  Fill missing Price with 0 

df["Price"] = df["Price"].fillna(0)
print(df)
```

# Task 3

```python
# Create a DataFrame: 
data = { 
        "Name": ["A", None, "C", None], 
        "Age": [20, None, 22, None], 
        "Salary": [30000, None, None, 40000] 
    } 

df = pd.DataFrame(data)

print(df)

print("----------")

# Task: 

# Fill missing values using one fillna() call: 

# Name → "No Name" 
# Age → column mean
# Salary → 25000

df = df.fillna({
    "Name": "No Name",
    "Age": df["Age"].mean(),
    "Salary": 25000
})
print(df)
```

# Task 4

```python
# Given: 

df = pd.DataFrame(
    { 
        "City": ["Delhi", None, "Mumbai", None], 
        "Population": [None, 15000000, None, 12000000] 
    }
) 

print(df)

print("----------")

# Tasks: 

#  Replace missing City with "Unknown City" 
#  Replace missing Population with median

df = df.fillna({
    "City" : "Unknown City",
    "Population" : df["Population"].median() 
})

print(df)
```

# Level 3: Conditional Fill using loc 

# Task 5

```python
df = pd.DataFrame(
    { 
        "Name": ["A", "B", "C", "D"], 
        "Age": [None, 25, None, None], 
        "City": ["Delhi", "Mumbai", "Delhi", "Mumbai"] 
    }
) 

print(df)

print("----------")
# Tasks: 

#  If City == 'Delhi', fill missing Age with 18 

df.loc[(df["City"] == "Delhi") & (df["Age"].isna()), "Age"] = 18

print(df)

print("----------")

#  If City == 'Mumbai', fill missing Age with 30

df.loc[(df["City"] == "Mumbai") & (df["Age"].isna()),"Age"] = 30

print(df)
```

# Task 6

```python
df = pd.DataFrame(
    { 
        "Employee": ["E1", "E2", "E3", "E4"], 
        "Salary": [None, 40000, None, None], 
        "Department": ["HR", "IT", "HR", "IT"] 
    }
) 

print(df)
print("---------")

# Tasks: 

#  Fill missing Salary: 

# o HR → 25000 

df.loc[(df["Department"] == "HR") & (df["Salary"].isna()), "Salary"] = 25000
print(df)

print("-----------")

# o IT → 35000

df.loc[(df["Department"] == "IT") & (df["Salary"].isna()), "Salary"] = 35000
print(df)
```
