# 8.Working with Dates and Time

1.Converting to DateTime:- Pandas me dates by default string ("2025-01-15") ke form me hote hai.unhe datetime object banana padta hai,taki hum unse year, month,day resampling kar saken.

```python
import pandas as pd

data={
    "Order ID":[101,102,103,104,105],
    "OrderDate":["2025-01-15","2025-02-20","2025-03-05","2025-03-18","2025-04-28"],
    "Sales":[250,300,400,200,350]
}

df=pd.DataFrame(data)
df.dtypes
# # Convert to datetime
df['OrderDate']=pd.to_datetime(df['OrderDate'])

print(df.dtypes)

# Output me OrderDate ka dtype datetime64[ns] hojayega
```

2.Extracting year,month,day

```python
df['Year']=df['OrderDate'].dt.year
df['Month']=df['OrderDate'].dt.month
df['Day']=df['OrderDate'].dt.day
print(df)
```

3.Resampling & Frequency Conversion

ye time-series ML me bahot use hota hai.Resampling ka matlab hai ek frequency se dusri frequency me data ko convert krna (daily->monthly,weekly->yearly).

```python
# Index ko date banana pdta hai resampling ke liye
# df=df.set_index("OrderDate")

# # Monthly sales total
# monthly_sales=df['Sales'].resample("M").sum()
# print(monthly_sales)
# weekly sales total

weekly_sales=df['Sales'].resample("W").sum()
print(weekly_sales)
```

# Q3. Monthly & Weekly Sales

```python
# You have this DataFrame:

data = { 
        "OrderID": [201,202,203,204,205,206], 
        "OrderDate": ["2025-01-02","2025-01-15","2025-01-29","2025-02-10","2025-02-25","2025-03-05"], 
        "Amount": [120,250,300,500,150,400] 
}

df = pd.DataFrame(data)
print(df)
```

```python
# 👉 Tasks:

# 1. Convert OrderDate to datetime and set it as the index.


df["OrderDate"] = pd.to_datetime(df["OrderDate"])

```

```python

df.set_index("OrderDate",inplace = True)
df
```

```python
# 2. Find monthly total sales.

monthly_sales = df["Amount"].resample("M").sum()
print(monthly_sales)
```

```python
# 3. Find weekly total sales.

weekly_sales = df["Amount"].resample("W").sum()
print(weekly_sales)
```

# Q4. Quarterly Analysis

```python
# You have stock price data:

data = { 
        "Date": ["2025-01-10","2025-02-18","2025-03-25","2025-04-05","2025-04-20","2025-05-15"], 
        "Price": [105,110,115,120,118,125] 
}

df = pd.DataFrame(data)
print(df)
```

```python
# 👉 Tasks:

# 1. Convert Date into datetime and set it as index.

df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.set_index("Date", inplace = True)
df
```

```python
# 2. Calculate quarterly average price.

quarterly_avg = df["Price"].resample("Q").mean()
print(quarterly_avg)
```
