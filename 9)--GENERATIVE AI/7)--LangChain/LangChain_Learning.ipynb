# 07--LangChain
LangChain — LLM application development framework.

## Learning Flow
Models → Prompts → Messages → Parsers → Runnables → Chains → Retrievers → Tools → Agents → RAG → Production

```python
def prompt_template(topic):
    return f'Explain {topic} in three practical points.'

def model(prompt):
    return f'MODEL RESPONSE -> {prompt}'

def parser(response):
    return {'response': response, 'status': 'parsed'}

result = parser(model(prompt_template('LangChain')))
print(result)
```

## Core Idea
A LangChain application composes reusable components. A common pattern is **Prompt → Model → Parser**, while RAG and agentic applications add retrievers and tools.
