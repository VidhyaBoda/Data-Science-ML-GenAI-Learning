# 1️⃣ How to load dataset (Scikit-learn)

=> Scikit-learn me datasets mainly 3 tarah se load hote hain:

A. Toy / Built-in Datasets (load_*)

B. Fetch / Real-world Datasets (fetch_*)

C. Generated / Synthetic Datasets (make_*)


# 🔸A. Toy / Built-in Datasets (load_*)

=>Small datasets already sklearn me available hote hain

Examples: 
    
    load_iris(), 
    load_wine(), 
    load_digits(), 
    load_breast_cancer(), 
    load_diabetes()

```python
from sklearn.datasets import load_iris

iris=load_iris()

print(iris)
```

Variables Provided:


| Variable             | Description                                     |
| -------------------- | ----------------------------------------------- |
| `iris.data`          | Features (X), usually NumPy array               |
| `iris.target`        | Labels / target (y)                             |
| `iris.feature_names` | List of feature names                           |
| `iris.target_names`  | List of target class names (for classification) |
| `iris.DESCR`         | Full description of dataset                     |
| `iris.filename`      | Path of the dataset (for local file, optional)  |


```python
print("data :\n",iris.data[:5])
print("---------------------")

print("target :\n",iris.target[:5])
print("---------------------")

print("feature_names :\n",iris.feature_names)
print("---------------------")

print("target_names :\n",iris.target_names)
print("---------------------")

print("Full description of dataset :\n",iris.DESCR)
print("---------------------")

print("filename :",iris.filename)
```

# 🔸B. Fetch / Real-world Datasets (fetch_*)

=>Large datasets, internet se download hote hain

Examples: 
    
    fetch_california_housing(), 
    fetch_20newsgroups(), 
    fetch_lfw_people()

```python
from sklearn.datasets import fetch_california_housing

california=fetch_california_housing()

print(california)
```

Variables Provided:

| Variable                   | Description              |
| -------------------------- | ------------------------ |
| `california.data`          | Features (X)             |
| `california.target`        | Target (y)               |
| `california.feature_names` | Feature names            |
| `california.DESCR`         | Description              |
| `california.data_module`   | Source module / location |


# 🔸C. Generated / Synthetic Datasets (make_*)

=>Artificial datasets → regression, classification, clustering

Examples: 
    
    make_classification(), 
    make_regression(), 
    make_blobs(), 
    make_moons()

```python
from sklearn.datasets import make_regression

X,y=make_regression(n_samples=100,n_features=4,noise=0.1)
print(X)
print("---------------------------------")
print(y)
```

Variables Provided:

- X → Features (NumPy array)

- y → Target variable (continuous for regression, discrete for classification)

Note: Generated datasets directly return X, y, unlike load_* which return dictionary-like object
