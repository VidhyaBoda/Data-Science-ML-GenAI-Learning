# Clustering Models

| Model                        | Description                |
| ---------------------------- | -------------------------- |
| K-Means Clustering           | Most popular clustering    |
| K-Medoids                    | Robust version of K-Means  |
| Hierarchical Clustering      | Dendrogram-based           |
| Agglomerative Clustering     | Bottom-up clustering       |
| Divisive Clustering          | Top-down clustering        |
| DBSCAN                       | Density-based clustering   |
| HDBSCAN                      | Improved DBSCAN            |
| Mean Shift                   | Mode seeking               |
| OPTICS                       | Density-based              |
| Birch                        | Large datasets             |
| Gaussian Mixture Model (GMM) | Probabilistic clustering   |
| Spectral Clustering          | Graph-based clustering     |
| Affinity Propagation         | Message-passing clustering |
| Fuzzy C-Means                | Soft clustering            |
