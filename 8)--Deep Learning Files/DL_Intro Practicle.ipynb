```python
model = Sequential()

model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D((2,2)))

model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

Sequential → model where layers are added one by one

Conv2D → convolution layer for image feature detection

MaxPooling2D → reduces image size

Flatten → converts image to 1D

Dense → fully connected layer

```python
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)))
```

32 → number of filters (feature detectors)

(3,3) → filter size

activation='relu' → removes negative values

input_shape=(28,28,1)

28 → height

28 → width

1 → grayscale image

```python
model.add(MaxPooling2D((2,2)))
```

Takes maximum value from 2×2 area

Reduces image size

Keeps important features only

```python
28×28 → 14×14
```

```python
model.add(Conv2D(64, (3,3), activation='relu'))
```

Second convolution layer

Uses 64 filters

Learns deeper features

shapes

curves

patterns

```python
model.add(MaxPooling2D((2,2)))
```

Again reduces size

Makes model faster

Prevents overfitting

```python
model.add(Flatten())
```

Converts 2D feature maps into 1D vector


```python
model.add(Dense(128, activation='relu'))
```

Fully connected hidden layer

128 neurons

Learns final relationships between features

```python
model.add(Dense(10, activation='softmax'))
```

Output layer

10 → number of classes (0–9 digits)

softmax → gives probability for each class

```python
[0.01, 0.02, 0.90, ...]
```

Highest value = final prediction.

| Line       | Meaning               |
| ---------- | --------------------- |
| Conv2D     | Detect image features |
| ReLU       | Add non-linearity     |
| MaxPooling | Reduce size           |
| Flatten    | 2D → 1D               |
| Dense      | Classification        |
| Softmax    | Final prediction      |
