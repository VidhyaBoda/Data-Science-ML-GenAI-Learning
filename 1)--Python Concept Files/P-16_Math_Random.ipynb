# 🔹 1️⃣ What is Math Module?

Definition:
The math module in Python provides mathematical functions like trigonometry, logarithm, power, rounding, etc.


How to use:

import math

# 2️⃣ Commonly Used Functions in Math Module

Mathematical Constants

```python
import math

print(math.pi) #just like π.
print(math.e)   #e≈2.718281828459045
print(math.atan)
```

Power & Roots

```python
print(math.pow(2,3)) # 8.0 → 2^3
print(math.sqrt(16))  #4.0 → square root
```

Rounding Functions

```python
import math

print(math.ceil(4.3)) #5 → round up
print(math.floor(4.8)) #4 → round down
print(math.trunc(4.9))  # 4 -> Remove decimal part
print(round(4.8)) #normal rounding
```

Absolute & Factorial

```python
-3 -2 -1 0 1 2 3
```

```python
print(math.fabs(9))

print(math.factorial(5)) #120 → 5! = 5*4*3*2*1
```

# What is random in Python?

The random module is part of Python’s standard library — it helps you generate random numbers, shuffle data, or pick random elements.


Importing the module

import random

# Common Functions in random

# 1️⃣ random.random()

Returns a random float between 0.0 and 1.0.

```python
import random

print(random.random())  
```

# 2️⃣ random.randint(a, b)

Returns a random integer between a and b (both inclusive).

```python
x=random.randint(1000, 10000) # Example output: 7

print(x)
```

# 3️⃣ random.uniform(a, b)

Returns a random float between a and b.

```python
print(random.uniform(1, 10))  # Example output: 3.674
```

# 4️⃣ random.choice(sequence)

Returns a random element from a list, tuple, or string.

```python
colors = ['red', 'green', 'blue']
print(random.choice(colors))  # Example output: 'green'
```

# 5️⃣ random.choices(sequence, k=n)

Returns a list of n random elements (with replacement).


```python
nums = [1, 2, 3, 4, 5]
print(random.choices(nums, k=3) ) # Example: [4, 1, 5]
```

# 6️⃣ random.sample(sequence, k=n)

Returns n unique elements (without replacement).

```python
nums = [1, 2, 3, 4, 5]
print(random.sample(nums, k=3))  # Example: [5, 1, 3]
```

# 7️⃣ random.shuffle(list)

Shuffles a list in place.

```python
cards = ['A', 'K', 'Q', 'J']
random.shuffle(cards)
print(cards)  # Example: ['J', 'Q', 'K', 'A']
```

# 8️⃣ random.seed(value)

Sets the seed for reproducibility — ensures the same random results every time.

```python
random.seed(7)
print(random.randint(1, 100))  # Will always give same number for same seed
```
