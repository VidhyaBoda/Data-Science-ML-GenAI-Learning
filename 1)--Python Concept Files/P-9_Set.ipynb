# 1️⃣ What is a Set?

Set is a collection of unique elements.

Duplicates are automatically removed.

Unordered (no index)

Written using {} braces.

```python
my_set={1,1,2,2,3,3,3,4,4,5,6,7}
print(my_set)
```

# 2️⃣ Creating a Set

```python
# Empty set
my_set=set()
print(type(my_set))

# from List
my_set_list=set([1,2,3])
print(my_set_list)

# from string
my_set_string=set("Pytthon")
print(my_set_string)
```

# 3️⃣ Adding Elements

```python
numbers={1,2,3,4}
numbers.add(5)
numbers.add(100)
print(numbers) 
```

# 5️⃣ Removing Elements

```python
numbers={1,2,3,4}
numbers.remove(4)
print(numbers)

numbers.discard(5)  # No error if not found
print(numbers)

popped = numbers.pop()  # Removes a random element
print("Popped:", popped)
print(numbers)
```

# 7️⃣ Membership Test

```python
fruits={"apple","mango","orange"}
print("apple" in fruits)
print("orange" in fruits)
print("pinapple" in fruits)
print("pinapple" not in fruits)
```

# 8️⃣ Set Operations

Python sets support mathematical operations like union, intersection, etc.


 Union → Combine unique elements from both sets

```python
A={1,2,3}
B={3,4,5}

union=A.union(B)
union1=A|B

print(union)
print(union1)
```

Intersection → Common elements

```python
a={1,2,3,4}
b={2,4,6,8}

intersection=a.intersection(b)
inter=a & b
print(intersection)
print(inter)
```

Difference → Elements in A not in B

```python
a={1,2,3,4,5}
b={2,4,6,8,9}

no=a.difference(b)
not_in=a-b
print(no)
print(not_in)
```

Symmetric Difference → Elements not common to both

```python
a={1,2,3,4,5}
b={1,4,5,6}

symatry=a.symmetric_difference(b)
sym=a^b
print(symatry)
print(sym)
```

```python
# Q-1 Create a set of 5 colors and print it.

# Create a set of 5 colors
colors = {"red", "blue", "green", "yellow", "orange"}

# Print the set
print(colors)
```

```python
# Q-2 Add two new colors to it.

# Existing set of colors
colors = {"red", "blue", "green", "yellow", "orange"}

# Add two new colors
colors.add("purple")
colors.add("pink")

# Print updated set
print(colors)
```

```python
# Q-3 Remove one color using remove() and another using discard().

# Existing set of colors
colors = {"red", "blue", "green", "yellow", "orange", "purple", "pink"}

# Remove a color using remove()
colors.remove("green")

# Remove a color using discard()
colors.discard("pink")

# Print updated set
print(colors)
```

```python
# Q-4 Check if "red" is in the set or not.

# Existing set of colors
colors = {"red", "blue", "yellow", "orange", "purple"}

# Check if "red" is in the set
if "red" in colors:
    print("red is present in the set")
else:
    print("red is not present in the set")
```

```python
# Q-5 Create two sets:

# A = {10, 20, 30, 40, 50} B = {30, 40, 50, 60, 70}

# Perform:

# Union

# Intersection

# Difference

# Symmetric Difference


# Create two sets
A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}

# Union
print("Union:", A | B)

# Intersection
print("Intersection:", A & B)

# Difference (A - B)
print("Difference (A - B):", A - B)

# Symmetric Difference
print("Symmetric Difference:", A ^ B)
```

```python
# Q-6 Find whether A is subset of B.

# Given sets
A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}

# Check if A is a subset of B
print(A.issubset(B))
```

```python
# Q-7 Write a program to remove duplicates from a list using set.

# Original list with duplicates
nums = [1, 2, 2, 3, 4, 4, 5, 6, 6]

# Remove duplicates using set
unique_nums = list(set(nums))

# Print result
print("Original List:", nums)
print("List after removing duplicates:", unique_nums)
```

```python
# Q-8 Given two sets of student names (science & math), find:

# Students who study both subjects

# Students who study only one subject



# Sets of student names
science_students = {"Amit", "Neha", "Ravi", "Priya", "Karan"}
math_students = {"Ravi", "Priya", "Anita", "Karan", "Sonal"}

# Students who study both subjects
both_subjects = science_students & math_students

# Students who study only one subject
only_one_subject = science_students ^ math_students

# Print results
print("Students who study both Science and Math:", both_subjects)
print("Students who study only one subject:", only_one_subject)
```

```python
# Q-9 Combine two sets and remove all elements greater than 50.

# Given two sets
set1 = {10, 20, 30, 40, 60}
set2 = {25, 35, 45, 55, 70}

# Combine both sets (union)
combined_set = set1 | set2

# Remove elements greater than 50
result_set = {x for x in combined_set if x <= 50}

# Print result
print("Combined Set:", combined_set)
print("After removing elements greater than 50:", result_set)
```

```python
# Q-10 Check if two sets have no elements in common.

# Given two sets
set1 = {10, 20, 30}
set2 = {40, 50, 60}

# Check if sets have no elements in common
if set1.isdisjoint(set2):
    print("The two sets have no elements in common.")
else:
    print("The two sets have some elements in common.")
```
