# 1️⃣ What is a Lambda Function?
✅ Definition

A lambda function is a one-line function created using the keyword lambda.

```python
# 🔹 Syntax :
    
lambda arguments: expression
```

# 📌 Important Rules

No function name

No return keyword (it returns automatically)

Only one expression

Can take multiple arguments

# 2️⃣ Normal Function vs Lambda Function

```python
# 🔸 Normal Function

def square(x):
    return x * x

# 🔸 Lambda Function

square = lambda x: x * x

print (square(5))
```

# 3️⃣ Simple Lambda Examples

```python
# Example 1: Add two numbers

add = lambda a, b: a + b
print(add(10, 20))
```

```python
# Example 2: Find maximum

max_num = lambda a, b: a if a > b else b
print(max_num(8, 15))
```

```python
# Example 3: Check even or odd

even_odd = lambda n: "Even" if n % 2 == 0 else "Odd"
print(even_odd(7))
```

# 4️⃣ Lambda with Multiple Arguments

```python
area = lambda l, w: l * w
print(area(5, 4))
```

# 5️⃣ Lambda with Default Arguments

```python
power = lambda x, y=2: x ** y
print(power(5))
print(power(5, 3))
```

# 6️⃣ Lambda with map() Function
🔹 What is map()?

Applies a function to each element of a list.

```python
# Square all numbers

nums = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, nums))
print(squares)
```

# 7️⃣ Lambda with filter() Function
🔹 What is filter()?

Filters elements based on a condition.

```python
# Example: Get even numbers

nums = [10, 15, 20, 25, 30]

evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
```

# 8️⃣ Lambda with reduce() Function
🔹 What is reduce()?

Reduces a list to a single value.

📌 Need to import:

```python
from functools import reduce

# Example: Sum of all elements

nums = [1, 2, 3, 4]

total = reduce(lambda a, b: a + b, nums)
print(total)
```

# 9️⃣ Lambda with Sorting
🔹 Sort by second element

```python
data = [(1, 3), (4, 1), (2, 2)]

data.sort(key=lambda x: x[1])
print(data)
```

# 🔟 Lambda Inside Another Function

```python
def multiplier(n):
    return lambda x: x * n

double = multiplier(2)
print(double(5))
```

# 1️⃣1️⃣ Lambda with Conditional Logic

```python
result = lambda x: "Pass" if x >= 40 else "Fail"
print(result(35))
```

# 1️⃣2️⃣ Limitations of Lambda ❌

❌ Cannot contain multiple expressions

❌ No loops (for, while)

❌ No statements (print, try, assign)

❌ Hard to debug if complex

# 1️⃣3️⃣ When to Use Lambda? ✅

✔️ Short operations

✔️ Temporary functions

✔️ Inside map, filter, reduce, sort

✔️ Cleaner code for one-time logic

```python
# Q1. Given lst = [3, 7, 2, 9, 12, 17, 4]
# Filter out numbers greater than 5.

lst = [3,7,2,9,12,17,4]

Greater = list(filter(lambda x:x>5,lst))
print(Greater)
```

```python
# Q2. Given lst = [10, 21, 32, 43, 54, 65]
# Filter only the odd numbers.

lst = [10, 21, 32, 43, 54, 65]
odd = list(filter(lambda x:x%2 != 0, lst))
print(odd)
```

```python
# Q3. Given names = ["Sumit", "Ranu", "Raj", "Aman", "Sita"]
# Filter names that have more than 4 letters.

names = ["Sumit", "Ranu", "Raj", "Aman", "Sita"]
number = list(filter(lambda name: len(name)>4,names))

print(number)
```

```python
# Q4. Given words = ["apple", "cat", "banana", "dog"]
# Filter words that start with 'b'.

words = ["apple", "cat", "banana", "dog"]
b_words = list(filter(lambda word : word.startswith('b'),words))
print(b_words)
```

```python
# Q5. Given lst = [1, 2, 3, 4, 5]
# Use map to create a list of each number multiplied by 10.

lst = [1, 2, 3, 4, 5]
multiply = list(map(lambda x:x*10 , lst))

print(multiply)
```

```python
# Q6. Given prices = [100, 200, 300]
# Increase each price by 15% using map.

prices = [100, 200, 300]
Increase = list(map(lambda x: x*1.15, prices))
print(Increase)
```

```python
# Q7. Given words = ["hello", "python", "world"]
# Convert each word to uppercase using map.



# Using map() with upper()
# uppercase_words = list(map(str.upper, words))
# print(uppercase_words)


# using lambda
words = ["hello", "python", "world"]
uppercase = list(map(lambda word : word.upper(),words))

print(uppercase)
```

```python
# Q8. Given lst = [2, 4, 6, 8]
# Convert each number to its string using map.

lst = [2,4,6,8]
numToStr = list(map(lambda x : str(x), lst))
print(numToStr)
```

```python
# Q9. Given lst = [1, 2, 3, 4, 5]
# Use reduce to find the product of all numbers.

from functools import reduce 

lst = [1,2,3,4,5]
product = reduce(lambda x,y : x*y, lst)
print(product)
```

```python
# Q10. Given lst = [100, 20, 50]
# Use reduce to find the largest number.
# (Hint: lambda x, y: x if x > y else y)

lst = [100, 20, 50]
largest = reduce(lambda x,y : x if x>y else y , lst)
print(largest)
```

```python
# Q11. Given words = ["I", "love", "Python"]
# Use reduce to join words into a sentence ➝ "I love Python"

words = ["I", "love", "Python"]
joinWord = reduce(lambda a,b : a + " " + b , words)
print(joinWord)
```

```python
# Q12. Given lst = [5, 10, 15, 20]
# Use reduce to find the difference like this: ((((5 - 10) - 15) - 20))

lst = [5, 10, 15, 20]
difference = reduce(lambda a,b : a-b,lst)
print(difference)
```
