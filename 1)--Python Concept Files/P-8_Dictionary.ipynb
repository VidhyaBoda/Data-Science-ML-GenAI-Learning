# 1️⃣ What is a Dictionary?

Dictionary stores data in key–value pairs.

Like a mini database: each key has a corresponding value.

Written using {} curly braces.

Keys must be unique and immutable (string, int, tuple)

Values can be anything (list, dict, etc.)

```python
student={
    "name":"Sumit",
    "age":22,
    "course":"Data Science" 
}

print(student)


person={"Name":"Om","Age":22,"Marks":100}
print(person["Marks"])
```

# 2️⃣ Accessing Dictionary Elements

You can access values using keys:

```python
student={
    "names":"Sumit",
    "age":22,
    "course":"Data Science" 
}

print(student["names"])
print(student["age"])
print(student["course"])
print(student.get("name")) #its safer if no key found it returns None.
```

# 3️⃣ Adding / Updating Key-Value Pairs

```python
student={
    "names":"Sumit",
    "age":22,
    "course":"Data Science" 
}

student['city']='Baroda' # Add new key-value
print(student)

student['age']=20  # Update existing key
print(student)
```

# 5️⃣ Looping through Dictionary

```python
person={"name":"Rajesh","age":20,"gender":"Male","course":"DS"}

for key in person: #printing only keys
    print(key)

for value in person.values(): # print only values
    print(value)

for key,value in person.items(): # print both key and value
    print(key,value)

```

| Method                  | Description                        |
| ----------------------- | ---------------------------------- |
| `dict.keys()`           | Returns all keys                   |
| `dict.values()`         | Returns all values                 |
| `dict.items()`          | Returns list of (key, value) pairs |
| `dict.get(key)`         | Get value safely                   |
| `dict.update(new_dict)` | Update dictionary                  |
| `dict.pop(key)`         | Remove item by key                 |
| `dict.popitem()`        | Remove last item                   |
| `dict.clear()`          | Remove all items                   |
| `dict.copy()`           | Copy dictionary                    |


# 8️⃣ Nested Dictionary

You can store dictionaries inside another dictionary.

```python
students={
    "student1":{"name":"pooja","age":18,"course":"AIML"},
    "student2":{"name":"Rohit","age":20,"course":"DS"}
}

print(students["student1"]["name"])

```

# 9️⃣ Dictionary from List of Tuples

```python
pairs=[("a",1),("b",2),("c",3)]
dict_from_list=dict(pairs)
print(dict_from_list)
```

```python
# Q-6 Create a dictionary:

# emp = {"name": "Ravi", "salary": 50000, "dept": "IT"}

# Increase salary by 10%

# Add a new key "city": "Pune"

# Delete "dept" key

emp = {
    "name": "Ravi",
    "salary": 50000,
    "dept": "IT"
}

# Increase salary by 10%
emp["salary"] = emp["salary"] + (emp["salary"] * 10 / 100)

# Add new key
emp["city"] = "Pune"

# Delete dept key
del emp["dept"]

print(emp)
```

```python
# Q-7 Create a dictionary from two lists:

keys = ["name", "age", "city"]
values = ["Amit", 25, "Delhi"]

# Create dictionary
person = dict(zip(keys, values))

print(person)
```

```python
# Q-8 Merge two dictionaries and remove all keys having value < 10.

dict1 = {"a": 5, "b": 15, "c": 25}
dict2 = {"d": 8, "e": 12, "f": 30}

# Merge dictionaries
merged_dict = dict1 | dict2   # Python 3.9+

# Remove keys with value < 10
result = {}
for key, value in merged_dict.items():
    if value >= 10:
        result[key] = value

print(result)
```
