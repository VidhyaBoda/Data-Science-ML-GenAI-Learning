```python
import numpy as np

arr=np.array([10,20,30,40,50])
result=np.where(arr%2==0)
result=np.where((arr>10) & (arr<40))
print(result)

resulted=np.searchsorted(arr,[35,23])
print(resulted)


nonzeros=np.nonzero(arr)
print(nonzeros)


arr=np.array([5,10,20,1])

print(np.any(arr==0)) #true-at least one zero
print(np.all(arr>0))


arr=np.array([50,20,30,10,40])
sorted_arr=np.sort(arr)

print("Original:",arr)
print("Sorted:",sorted_arr)


arr2d=np.array([[50,20,30],
               [10,60,40]])

# sort each row(axis=1)

np.sort(arr2d,axis=1)


arr=np.array([10,19,14,50,30])
st=np.argsort(arr)
print(st)


names=np.array(['Alice','Bob','Charlie','David',"Eva"])
marks=np.array([50,20,30,10,40])

sorted_names=names[np.argsort(marks)]
print(sorted_names)
```

# 1.np.where() - find indexes based on condition

syntax:-np.where(condition)

ye function batata hai kis index par condition True hai.

```python
import numpy as np

arr=np.array([10,20,30,40,50,60])

result=np.where(arr>30)
print(result)


arr=np.array([5,10,15,10,20,0])
result=np.where(arr==10)
print(result)
```

```python
# using multiple conditions
import numpy as np

arr=np.array([2,5,8,12,15,20])
result=np.where((arr>5) & (arr<15))
print(result)


# conditional replacement

# you can also use np.where() to replace values conditionally

arr=np.array([10,20,30,40,50])
new_arr=np.where(arr>30,'High','Low')
print(new_arr)
```

# 2.np.searchsorted()-Find where to insert

np.searchsorted(sorted_array,value)

ye function batata hai kis index par ek value insert karni chahaiye taki array sorted rahe.

```python
arr=np.array([10,20,30,40,50])
x=np.searchsorted(arr,35)
print(x)


# means ->if u insert 35 at index 3 ,array remains sorted.
# result=10,20,30,35,40,50


# multiple values insert karna

arr=np.array([10,20,30,40,50])
x=np.searchsorted(arr,[25,45])
print(x)


# using side='right'
arr=np.array([10,20,30,40,50])
x=np.searchsorted(arr,30,side='left')
print(x)

# if duplicate value exist side='right' gives position after duplicates
```

```python
arr=np.array([10,20,30,40,50])

sorted=np.searchsorted(arr,[25,45])
print(sorted)
```

# 5.np.any() and np.all() for checking conditions

```python
# np.any()->checks if any element matches condition.

# np.all()->checks if all elements match condition.

arr=np.array([5,10,20,1,0])

print(np.any(arr==0)) #true-at least one zero
print(np.all(arr>0)) #false -beacuse 0 is not >0
```

# Numpy Array Sorting

array ke elements ko ascending ya descending order me arrange karna.

Numpy me ye kaam hum np.sort(),sort method or argsort() se krte hai.

np.sort()->Returns a sorted copy


np.sort(array,axis=-1,kind='quicksort',order=None)

array->input array
axis->along which axis to sort (-1=last axis by default)
kind->sorting algorithm(quicksort,mergesort,heapsort)
order->only for strucutred arrays

np.sort() returns a new sorted array,original array unchanged.

```python
import numpy as np

arr=np.array([50,20,30,10,40])
sorted_arr=np.sort(arr)

print("Original:",arr)
print("Sorted:",sorted_arr)


# Descending order

arr=np.array([50,20,30,10,40])
sorted_arr=np.sort(arr)[::-1]
print(sorted_arr)
```

2.array.sort()->sort in-place

array.sort(axis=-1,kind='quicksort',order=None)

->array.sort() modifies the original array (in-place)

->does not return a new array

```python
arr=np.array([50,20,30,10,40])
arr.sort()
print(arr)

# original array sorted in-place
```

4.argsort()-get indexes instead of values

np.argsort(array)->returns indexes that would sort the array

useful for tracking original positions.

```python
arr=np.array([50,20,30,10,40])
indices=np.argsort(arr)
print(indices)

# at index 3->10(smallest)

# ar index 1->20

# at index 0->50

# using these indexes you can reorder thr array other related arrays


# using argsort to sort another array

names=np.array(['Alice','Bob','Charlie','David',"Eva"])
marks=np.array([50,20,30,10,40])

sorted_names=names[np.argsort(marks)]
print(sorted_names)

# sorted by marks names follow the same order 
```

Dataset 1 — Students Marks

```python
# Q5: Sort marks in ascending order.
# Sort names according to marks using argsort().

names = np.array(["Alice", "Bob", "Charlie", "David", "Eva"]) 
marks = np.array([50, 20, 30, 10, 40])


sorted_marks =  np.sort(marks)
result = names[np.argsort(marks)]
print(sorted_marks)
print(result)
```

```python
# Q6: Sort only students who scored >20, both marks and names.
    
names = np.array(["Alice", "Bob", "Charlie", "David", "Eva"]) 
marks = np.array([50, 20, 30, 10, 40])

index = np.argsort(marks[marks>20])

print(names [ marks > 20 ][index])
print(marks [ marks >20 ][index])
```

Dataset 2 — Employee Salaries

```python
# Q7: Sort salaries in ascending order.
# Sort employees according to salaries using argsort().

employees = np.array(["John", "Sara", "Mike", "Anna", "David"]) 
salaries = np.array([45000, 35000, 50000, 30000, 40000])

sort_salary = np.sort(salaries)
print(sort_salary)

sort_employee = employees[np.argsort(salaries)]
print(sort_employee)

```

```python
# Q8: Find employees earning more than 40000, sort them ascendingly, and print both names and salaries.

employees = np.array(["John", "Sara", "Mike", "Anna", "David"]) 
salaries = np.array([45000, 35000, 50000, 30000, 40000])

idx = np.argsort(salaries[salaries > 40000])

print(employees[salaries > 40000][idx])
print(salaries[salaries > 40000][idx])
```

```python
# Q9: Sort prices in ascending order.
# Sort products according to prices using argsort().

products = np.array(["Laptop", "Phone", "Tablet", "Monitor", "Keyboard"]) 
prices = np.array([75000, 35000, 45000, 15000, 5000])

asce = np.sort(prices)
print(asce)

result = products[np.argsort(prices)]

print(result)
```

```python
# Q10: Sort only products costing more than 20000, both names and prices.

products = np.array(["Laptop", "Phone", "Tablet", "Monitor", "Keyboard"]) 
prices = np.array([75000, 35000, 45000, 15000, 5000])

cost = np.argsort(prices[prices>20000])

print(cost)

print(products[prices>20000][cost])
print(prices[prices>20000][cost])
```
