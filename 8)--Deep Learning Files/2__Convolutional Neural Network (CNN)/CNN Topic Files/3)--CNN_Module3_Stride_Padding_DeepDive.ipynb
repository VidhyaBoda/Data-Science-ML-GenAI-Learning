# 1. Why Does Convolution Shrink Images?

Input: 5×5 image
Kernel: 3×3

The kernel cannot move outside the image boundary.

Therefore the output becomes smaller after convolution.


# 2. What is Stride?

Stride = Number of pixels the kernel moves each step.

Stride = 1
- More information retained
- More computation

Stride = 2
- Faster
- Smaller output
- Some information lost


# Example: Stride = 1

Input Size = 5×5
Kernel = 3×3

Kernel positions:
1 → 2 → 3

Output becomes:

3×3


# Example: Stride = 2

Input Size = 5×5
Kernel = 3×3

Kernel jumps 2 pixels at a time.

Output becomes:

2×2


# 3. What is Padding?

Padding means adding extra pixels around the image border.

Usually zeros are added.

Purpose:
- Preserve border information
- Control output size


# Example of Padding

Original:

1 2 3
4 5 6
7 8 9

Padding = 1

0 0 0 0 0
0 1 2 3 0
0 4 5 6 0
0 7 8 9 0
0 0 0 0 0


# 4. Types of Padding

## Valid Padding

No padding.

TensorFlow:

padding='valid'

Output shrinks.

---

## Same Padding

Automatically adds padding.

TensorFlow:

padding='same'

Output height and width remain approximately the same.


# 5. Output Shape Formula

Output = ((W - F + 2P) / S) + 1

Where:

W = Input size
F = Filter size
P = Padding
S = Stride


# Manual Calculation Example 1

W = 5
F = 3
P = 0
S = 1

Output:

((5-3+0)/1)+1

= 3

Output = 3×3


# Manual Calculation Example 2

W = 5
F = 3
P = 1
S = 1

Output:

((5-3+2)/1)+1

= 5

Output = 5×5


# Real CNN Example

Input:

28×28×1

Layer:

Conv2D(32,(3,3),padding='valid')

Output:

26×26×32

Because:

((28-3+0)/1)+1 = 26


# Same Padding Example

Input:

28×28×1

Layer:

Conv2D(32,(3,3),padding='same')

Output:

28×28×32


```python
from tensorflow.keras.layers import Conv2D

layer = Conv2D(
    filters=32,
    kernel_size=(3,3),
    strides=(1,1),
    padding='same'
)

print(layer)
```
