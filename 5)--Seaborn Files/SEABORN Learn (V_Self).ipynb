# 1⃣  What is Seaborn?

=>Seaborn is a Python data visualization library built on top of Matplotlib.
It is designed for statistical data visualization and works extremely well with Pandas DataFrames.

Seaborn simplifies complex plots like:

Distribution plots
Statistical comparisons
Correlation visualization
Multi-variable plots

=> Without seaborn, these require a lot of manual Matplotlib code.

# 3⃣ Seaborn Built-in Dataset

```python
import seaborn as sns

df = sns.load_dataset("tips")
print(df.head())
```

# 4⃣ Seaborn Plot Categories

# 1) Distribution Plots

```python
# import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.histplot(df["total_bill"],bins = 20)

plt.title("distribution of total Bill")
plt.show()
```

```python
sns.kdeplot(df["total_bill"],fill = True)

plt.title("Density Distribution of Total bill")
plt.show()
```

# 2) Relation Plots

```python
sns.scatterplot(x="total_bill", y="tip", data = df)

plt.title("Total Bill vc Tip")
plt.show()
```

```python
sns.regplot(x="total_bill", y="tip", data = df)

plt.title("Regression Relationship")
plt.show()
```

#  3) Categorical Plots

```python
sns.barplot(x="day", y="total_bill", data = df)

plt.title("Average Bill per Day")
plt.show()
```

```python
sns.boxplot(x = "day", y = "total_bill", data = df)

plt.title("Bill Distribution per Day")
plt.show()
```

```python
sns.violinplot(x = "day", y = "total_bill", data = df)

plt.title("Violin plot of Bills")
plt.show()
```

# 5⃣ Correlation Heatmap (Most Important)

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(corr, annot = True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

# 6⃣ Pair Plot

# 9⃣ Real Industry Use Cases

# 1⃣0⃣ Complete Example (Real Workflow) 

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Distribution
sns.histplot(df["total_bill"])

# Relation
sns.scatterplot(x = "total_bill", y = "tip", data = df)

# Categorical
sns.boxplot(x = "day", y = "total_bill", data = df)

# #correlation
numeric_df = df.select_dtypes(include = ["number"])
sns.heatmap(numeric_df.corr(), annot = True)

plt.show()



# => This gives quick understanding of dataset.
```
