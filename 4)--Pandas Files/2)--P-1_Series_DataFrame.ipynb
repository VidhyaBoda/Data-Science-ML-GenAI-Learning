What is Pandas?

Pandas is a python library used for working with data sets.

it has function for analyzing,cleaning,exploring and manipulating data.

the name "Pandas" has a reference to both "Panel Data" and Python Data Analysis" and was created by Wes mckinney in 2008.


Why Use Pandas?

Pandas allows us to analyze big data and make conclusions based on statical theories.

pandas can clean messy data sets, and make them readable and relevant.

Relevant data is very important in data science.


What Can Pandas Do?

Pandas gives you answer about the data.

Like:

->is there a correlation between two or more columns?

->what is average value.

->max value?

->min value?


Pandas are also able to delete rows that are not relevant, or contains wrong values , like empty or NULL values. This is called cleaning the data.

# What is Series?

a pandas series is like a column in a table.
Its a one-dimensional array holding data of any type. 


```python
import pandas as pd

a=[1,2,3,4]
myvar=pd.Series(a)
print(myvar)
```

# Labels


if nothing else is specified, the values are labeled with their index number.First value has index 0, second value has index 1 etc.

this label can be used to access a specified value.

```python
import pandas as pd

a=[10,20,30,40,50]

myseries=pd.Series(a)
print(myseries)

```

```python
# Create Labels


# with the index argument you can name your own labels.


a=[1,7,2]

myvar=pd.Series(a,index=["x","y","z"])

print(myvar)
```

```python
# Key/Value object as series

# you can also use key value object like a dictionary when creating a series.

calories={"day1":420,"day2":340,"day3":230}

df=pd.Series(calories,index=["day1","day2"])
print(df)
```

# Viewing the data


One of the most used method for getting a quick overview of the DataFrame is the head() method.

the head() method returns the headers and a specific number of rows,starting from the top.


```python
import pandas as pd

df=pd.read_csv("iris.csv")
print(df.head())
print(df.tail())
```

```python
# Q-1: Create a pandas series from a python list[10,20,30,40].
# ->Print the series.
# ->Print only the first 2 elements using head().

import pandas as pd

data = [10,20,30,40]
series = pd.Series(data)

print("Full series:")
print(series)

print("\nFirst 2 elements:")
print(series.head(2))
```

```python
# Q-2: Create a Dataframe for student marks: Name:["Amit","Riya","Sumit","Pooja"], Math:[78,89,67,90],Science:[88,78,95,80].
# ->Print first 3 Rows.
# ->Show column names and data types.

data = {
    "Name":["Amit","Riya","Sumit","Pooja"],
    "Math":[78,89,67,90],
    "Science":[88,78,95,80]
}

df = pd.DataFrame(data)

print("First 3 Rows:")
print(df.head(3))

print("\nColums Name:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
# Q-3: Read a CSV file named employee.csv into a DataFrame.
# ->Print first 5 rows.
# ->Print last 2 rows.
# ->Show number of rows and columns using shape.

import pandas as pd

df = pd.read_csv("employee.csv") 
print(df)


print("\nFisrt 5 rows:")
print(df.head(5))

print("\nLast 2 rows:")
print(df.tail(2))

print("Number of rows and columns:")
print(df.shape)
```

```python
# Q-4:From the employees.csv show:
# ->Only column names.
# ->Summary statistics(mean,std,min,max etc) using describe()

import pandas as pd

df = pd.read_csv("employee.csv")

print("\nColumn names:")
print(df.columns)

print("\nSummary statistics(mean,std,min,max):")

print(df.describe())
```
