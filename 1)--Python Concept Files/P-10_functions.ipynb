# 🔹 1️⃣ What is a Function?

Definition:

A function is a block of reusable code that performs a specific task.

Why use Functions?

✅ To avoid repeating code

✅ To organize program into smaller parts

✅ To make debugging and testing easier

# 🔹 2️⃣ Syntax of a Function


def function_name(parameters):

    # code block

    return result

```python
def greet():
    print("Hello welcome to datascience class.")

greet()
```

# 🧩 Explanation:

def → keyword used to define function

greet → function name

() → parameters (empty means no input)

print() → function body

greet() → function call

# 4️⃣ Function with Parameters

```python
def add(a,b):
    result=a+b
    print (result)

add(5,8)

# a and b are parameters.

# When we call add(5,10), those values are arguments.
```

# 5️⃣ Function with Return Value

```python
def multiply(a,b):
    return a*b

product=multiply(3,4)
print(product)


# Why return?

# Returns the result for further use.

# Without return, you can’t store the output in a variable.
```

# 6️⃣ Default Parameters

```python
def greet(name="sumit"):
    print("hello",name)

greet("pritam")
```

# 7️⃣ Keyword Arguments

```python
def student(name,age):
    print("name:",name)
    print("age:",age)

student(age=23,name="sumit")


def num(*a):
    print(a)

num(1,2,3,4)
```

# 8️⃣ Variable-Length Arguments

a) args → multiple positional arguments

```python
def add(*num):
    total=sum(num)
    print(total)

add(1,2,3,4,5)


# *args → converts all values into a tuple.
```

*b) kwargs → multiple keyword arguments

```python
def show_info(**details):
    for key,value in details.items():
        print(f"{key}: {value}")


show_info(name="sumit",age=22 ,course="java")

    
# **kwargs → converts arguments into a dictionary.
```

# 9️⃣ Function Returning Multiple Values

```python
def calculate(a,b):
    addition=a+b
    substraction=a-b
    return addition,substraction

x,y=calculate(3,4)

print("addition:",x)
print("substraction:",y)
```

```python
# Q-8 Write a function area_of_circle(radius) that returns the area of a circle.

# Formula: πr² (use 3.14 for π)

def area_of_circle(radius):
    return 3.14*radius*radius

result = area_of_circle(7)
print("Area of Circle: ", result)
```

```python
# Q-9 Write a function average_of_three(a, b, c) that returns the average of three numbers.

def average_of_three(a,b,c):
    return (a+b+c)/3

result = average_of_three(10,30,50)
print("Average of three numbers: ", result)
```

```python
# Q-10 Write a function is_even(num) that returns True if number is even, otherwise False.

def is_even(num):
    return num%2==0

print(is_even(10))
print(is_even(7))
```

```python
# Q-11 Write a function power(base, exponent=2) that returns base raised to exponent.

# Example: power(4) → 16, power(2, 3) → 8


# Define the function with default argument
def power(base, exponent=2):
    return base ** exponent

#function call

print(power(4))
print(power(2,3))
```

```python
# Q-13 Write a function student_info(name, age, course) and call it using keyword arguments (in any order).

def student_info(name,age,course):
    print("Name: ", name)
    print("Age: ", age)
    print("course: ", course)
    
# Call the function using keyword arguments (any order)

student_info(age=21, name="Vidhya",course="Data Science , AI-ML")
```

```python
# Q-14 Write a function employee(name, salary, company) and pass arguments in different orders using keywords.

def employee(name, salary, company):
    print("Name: ",name)
    print("Salary: ", salary)
    print("Company: ",company)
  
# Calling function using keyword arguments (different orders)

employee(salary=70000,company="TCS",name="Siya")


employee(company="Google", name="Vishwa", salary=100000)
```

```python
# Q-15 Write a function add_all(*numbers) that returns the sum of all given numbers.

# Example: add_all(2, 4, 6) → 12

def add_all(*numbers):
    return sum(numbers)

#Function calls
print(add_all(2,4,6))
print(add_all(1,5,6,9,10))
```

```python
# Q-16 Write a function find_max(*nums) that returns the maximum number among all arguments.

def find_max(*nums):
    return max(nums)

print(find_max(10,50,80,15))
print(find_max(5,10,40,0))
```

```python
# Q-17 Write a function show_details(**info) that prints key-value pairs of employee details.

# Example: show_details(name="Sumit", age=23, city="Delhi")

def show_details(**info):
    for key,value in info.items():
        print(f"{key} : {value}")
        
show_details(name="Sumit", age=23, city="Delhi")
```

```python
# Q-19 Write a function calculate_marks(english, maths, science) that returns total marks and average marks.

def calculate_marks(english, maths, science):
    total = english + maths + science
    average = total/3
    return total,average

#Call the functions 
result = calculate_marks(80,20,10)
print(result)

# Optional : Unpacking
total_marks,average_marks = calculate_marks(80,20,10)
print("total_marks:", total_marks)
print("average_marks:",average_marks)
```
