# Else in For Loop
The else keyword in a for loop specifies a block of code to be executed when the loop is finished:

```python
for x in range(6):
    print(x)
else:
    print("Program Completed!")
```

```python
# Nested Loops

colors=["red","big","tasty"]
fruits=["apple","bannana","cherry"]

for color in colors:
    for fruit in fruits:
        print(color,fruit)
```

Q2. Use a while loop to print numbers from 1 to 50.

👉 If the number is divisible by 3 → print "Fizz", 👉 If divisible by 5 → print "Buzz", 👉 If divisible by both 3 and 5 → print "FizzBuzz".

```python
i=0
while i<=50:
    if (i%3==0) & (i%5==0):
        print(i,"fizzbuzz")
    elif i%3==0:
        print(i,"Fizz")
    elif i%5==0:
        print(i,"buzz")
    
    else:
        print(i)
    i+=1
```

Q6. Take a number as input and check if it is Prime or not.

(Hint: A prime number is divisible only by 1 and itself.)

```python
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a Prime number")
else:
    is_prime = True   # assume number is prime
    for i in range(2, num):
        if num % i == 0:   # if divisible by any number
            is_prime = False
            break
    
    if is_prime:
        print(num, "is a Prime number")
    else:
        print(num, "is not a Prime number")
```

Q8. Use a loop and condition to find the largest and smallest number from a list.

numbers = [12, 45, 2, 67, 34, 89, 1]

```python
numbers = [12, 45, 2, 67, 34, 89, 1]

max=numbers[0]
min=numbers[0]

for number in numbers:
    if number>max:
        max=number
    if number<min:
        min=number
print("Max:",max)
print("Min:",min)
```

Q9. Use a while loop to reverse the digits of a number.

Example:

Input: 12345 Output: 54321

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10        # get last digit
    reverse = reverse * 10 + digit
    num //= 10              # remove last digit

print("Reversed number:", reverse)
```
