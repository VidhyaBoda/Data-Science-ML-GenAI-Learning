## 1. What are Embeddings?\n\nAn embedding is a numerical vector representation produced by a model. Similar or related inputs can be represented by vectors that are close under an appropriate similarity metric.

```python
example_vector = [0.12, -0.31, 0.77, 0.25]\nprint(example_vector)\nprint('Dimensions:', len(example_vector))
```

## 2. Similarity and Distance\n\nCommon approaches include cosine similarity, Euclidean distance and dot product. The appropriate metric depends on the embedding model and retrieval setup.

```python
import math\n\ndef cosine_similarity(a, b):\n    dot = sum(x*y for x, y in zip(a, b))\n    na = math.sqrt(sum(x*x for x in a))\n    nb = math.sqrt(sum(y*y for y in b))\n    return dot / (na * nb)\n\nv1 = [1, 2, 3]\nv2 = [1, 2, 3]\nprint(cosine_similarity(v1, v2))
```

## 3. Normalization\n\nL2 normalization scales a non-zero vector so its Euclidean norm becomes 1. This can make dot-product calculations correspond to cosine similarity.

## 4. Document and Query Embeddings\n\nDocuments/chunks are embedded and stored. A user query is embedded using a compatible embedding approach. Similarity search then retrieves candidate chunks.

## 5. Chunking\n\nChunking is not a trivial preprocessing step. Chunk size, overlap, semantic boundaries and metadata can materially affect retrieval quality.

```python
text = 'Embeddings help applications compare meaning using vector representations.'\nwords = text.split()\nchunk_size = 5\nchunks = [' '.join(words[i:i+chunk_size]) for i in range(0, len(words), chunk_size)]\nchunks
```

## 6. Semantic Search\n\nSemantic search retrieves content based on representation similarity rather than relying only on exact keyword overlap.

## 7. Embeddings in RAG\n\nDocuments → chunks → embeddings → vector store → query embedding → similarity search → relevant context → LLM.

## 8. Embedding Limitations\n\nEmbedding similarity does not guarantee factual correctness. Model choice, domain coverage, chunking, language support and retrieval configuration all affect results.
