# Task 1 — Draw a Scatter Plot with Different Shapes

```python
# Create a scatter plot for: 
# x = [5,10,15,20,25] 
# y = [10,30,20,40,60] 
# Conditions: 
# • Each point must have different color 
# • Each point must have different size 
# • Add grid, title, labels 
# • Add a dotted line (--) joining the points 

import matplotlib.pyplot as plt

x = [5,10,15,20,25] 
y = [10,30,20,40,60] 

colors = ['blue','orange','red','yellow','green']
sizes = [100,200,250,300,450]

plt.scatter(x,y, c = colors, s=sizes,edgecolor = 'black', )
plt.plot(x,y, linestyle = '--')
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.grid()
plt.title("Customized Scatter Plot")
plt.show()
```

# Task 2 — Two Groups, Two Scatter Plots

```python
# Make 2 groups: 
# Group A 
# x = [1,3,5,7] 
# y = [10,18,12,25] 
# Group B 
# x = [2,4,6,8] 
# y = [22,28,30,35] 
# Conditions: 
# • Plot both groups using scatter 
# • Give different colors 
# • Add legend 
# • Add title: "Performance Comparison" 
# • Add x label: "Time" 
# • Add y label: "Score"

x1 = [1,3,5,7] 
y1 = [10,18,12,25]

x2 = [2,4,6,8] 
y2 = [22,28,30,35]

plt.scatter(x1,y1, c='red',label='Group A',marker='*',s=60)
plt.scatter(x2,y2, c='blue',label = 'Group B',marker='D',s=40)

plt.legend()

plt.title("Performance Comparison")

plt.xlabel("Time")
plt.ylabel("Score")

plt.show()
```

# Task 3 — Horizontal Bar Chart 

```python
# Categories: 
# ['Apple','Banana','Mango','Orange'] 
# Values: 
# [50, 80, 40, 90] 
# Conditions: 
# • Use different colors 
# • Show value on each bar using plt.text() 
# • Add title: "Fruit Sales Report"

import matplotlib.pyplot as plt

Categories = ['Apple','Banana','Mango','Orange']

Values = [50, 80, 40, 90]

colors = ['red','yellow','purple','green']

plt.barh(Categories,Values, color= colors,alpha=0.7,edgecolor = 'black',height = 0.5)


for i in range(len(Categories)):
    plt.text(Values[i] + 2, i, str(Values[i]), va='center')

    
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.xlabel("Fruits")
plt.title("Fruit Sales Report")
plt.show()
```

# Task 4 — Histogram of Student Marks 

```python
# Create 100 random student marks between 1–100: 
# marks = np.random.randint(1,100,size=100) 
# Conditions: 
# • Show histogram with 10 bins 
# • Color = blue 
# • Edge color = black 
# • Add title: "Marks Distribution"

import matplotlib.pyplot as plt

import numpy as np

marks = np.random.randint(1,100,size=100)

plt.hist(marks,bins = 10,color='blue',edgecolor = 'black',alpha=0.7)

plt.title("Marks Distribution")

plt.show()
```

```python
# Task 5 — Mixed Chart (Line + Scatter) 
# Data: 
# x = [1,2,3,4,5] 
# y = [5,12,18,25,40] 
# Conditions: 
# • Draw line plot 
# • Draw scatter plot on same graph 
# • Line color = green 
# • Scatter color = red 
# • Add title, labels, grid 

import matplotlib.pyplot as plt

import numpy as np

x = [1,2,3,4,5] 
y = [5,12,18,25,40]

plt.plot(x,y,color = 'green')
plt.scatter(x,y, color = 'red')

plt.title("Mixed Chart (Line + Scatter)")

plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.grid()

plt.show()
```

# Task 6 — Compare 3 Bar Charts in One Graph 

```python
# Categories: 
# ['Math','Science','English'] 
# Scores: 
# student1 = [80,60,75] 
# student2 = [85,70,90] 
# student3 = [78,68,82] 
# Conditions: 
# • Use bar chart for 3 students in same graph 
# • Different colors 
# • Add legend 
# • Add title: "Student Performance Comparison" 

import matplotlib.pyplot as plt

Categories = ['Math','Science','English'] 

student1 = [80,60,75] 
student2 = [85,70,90] 
student3 = [78,68,82]


plt.bar(student1,height = 0.5, width = 0.5,color = 'black',label='Student 1')
plt.bar(student2,height = 0.7,width = 0.5,color = 'orange',label = 'Student 2')
plt.bar(student3,height=0.9,width = 0.5,color = 'green',label = 'Student 3')


plt.legend()
plt.title("Student Performance Comparison")
```

# Task 7 — Multiple Histograms

```python
# Create 2 datasets:
# d1 = np.random.randint(1,100,50)
# d2 = np.random.randint(1,100,50)
# Conditions:
# • Plot both histograms on same graph
# • Make them slightly transparent using alpha
# • Give different colors
# • Add legend
# • Add title

import matplotlib.pyplot as plt

import numpy as np

d1 = np.random.randint(1,100,50)
d2 = np.random.randint(1,100,50)

plt.hist(d1, alpha = 0.7, color = 'green',label='Hist d1')
plt.hist(d2, alpha = 0.7,color = 'purple',label='Hist d2')

plt.legend()

plt.xlabel = 'X-Values'
plt.ylabel = 'Y-Values'

plt.title("")
plt.show()
```
