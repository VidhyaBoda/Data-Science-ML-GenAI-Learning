# 03--Prompt-Engineering\n\nA practical notebook for designing, refining and evaluating prompts.

## 1. What is Prompt Engineering?\n\nPrompt Engineering is the design of instructions, context, constraints and output requirements that guide a generative AI system toward a useful result.

```python
prompt_parts = ['Task', 'Context', 'Constraints', 'Examples', 'Output Format']\nfor part in prompt_parts:\n    print('-', part)
```

## 2. Anatomy of a Good Prompt\n\nA strong prompt makes the objective and success criteria explicit.

```python
prompt = {\n    'role': 'Data Analyst',\n    'task': 'Analyze monthly revenue',\n    'context': 'Revenue increased for three months',\n    'constraints': ['Do not invent data', 'Be concise'],\n    'output_format': '3 bullet points'\n}\nprompt
```

## 3. Zero-shot, One-shot and Few-shot\n\nExamples can communicate output patterns that are difficult to specify with instructions alone.

## 4. Structured Input and Output\n\nUse explicit sections, delimiters and schemas when reliability matters.

```python
schema = {\n    'sentiment': 'positive | neutral | negative',\n    'priority': 'low | medium | high',\n    'reason': 'string'\n}\nschema
```

## 5. Prompt Templates\n\nTemplates make prompts reusable across users, datasets and application workflows.

```python
template = '''You are a {role}.\nTask: {task}\nContext: {context}\nOutput: {output_format}'''\nprint(template.format(\n    role='Data Analyst',\n    task='Explain a KPI trend',\n    context='Revenue increased for three months',\n    output_format='3 concise bullets'\n))
```

## 6. Task Decomposition\n\nBreak complex tasks into explicit stages such as define → inspect → analyze → validate → report.

## 7. RAG Prompting\n\nA RAG prompt should explicitly instruct the model to ground its answer in retrieved context and define what to do when the context is insufficient.

## 8. Prompt Evaluation\n\nEvaluate correctness, groundedness, format compliance, consistency, safety and usefulness across representative test cases.

## 9. Prompt Injection Basics\n\nTreat retrieved or user-supplied content as untrusted data. Separate trusted instructions from untrusted content and minimize tool permissions.

## Key Takeaways\n\n1. Be explicit about task, context and constraints.\n2. Use examples when the desired pattern is difficult to describe.\n3. Use structured outputs when downstream software consumes the response.\n4. Evaluate prompts systematically.\n5. In RAG systems, ground answers in retrieved context.\n6.
