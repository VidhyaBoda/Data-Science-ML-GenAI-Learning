```python
# 13. Student Marks
#  Generate marks for 12 students (0–100)
#  Find:
# o highest mark
# o lowest mark
# o average mark
#  Randomly select 3 students for a prize


marks = np.random.randint(0, 101, 12)
print("Student Marks:", marks)



print("Highest mark:", marks.max())
print("Lowest mark:", marks.min())
print("Average mark:", marks.mean())



selected_students = np.random.choice(len(marks), 3, replace=False)

print("Selected student index for prize:", selected_students)
print("Their marks:", marks[selected_students])
```

```python
# 14. Weekly Sales
#  Generate sales for 7 days (200–1000)
#  Find:
# o highest sales day
# o lowest sales day
#  Shuffle the sales data

sales = np.random.randint(200, 1001, 7)
print("Weekly Sales:", sales)



highest_day = np.argmax(sales)
lowest_day = np.argmin(sales)
print("Highest sales:", sales[highest_day], "on day index:", highest_day)
print("Lowest sales:", sales[lowest_day], "on day index:", lowest_day)



np.random.shuffle(sales)
print("Shuffled Sales:", sales)
```

```python
# 15. Dice Game
#  Simulate rolling a dice 20 times
#  Count how many times 6 appear

dice_rolls = np.random.randint(1, 7, 20)
print("Dice Rolls:", dice_rolls)


count_six = np.sum(dice_rolls == 6)
print("Number of times 6 appeared:", count_six)
```

```python
# 17. Create random salaries for 8 employees (30,000–80,000) and:
#  find lowest salary
#  randomly pick 2 employees for bonus


salaries = np.random.randint(30000, 80001, 8)

print("Employee Salaries:", salaries)

print("Lowest Salary:", salaries.min())



selected_indices = np.random.choice(len(salaries), 2, replace=False)

print("Selected Employee Indices:", selected_indices)
print("Their Salaries:", salaries[selected_indices])
```

```python
# 20. Why do we use np.random.seed()?


# np.random.seed() is used to make random number generation reproducible.

# Without a seed → different output every run.
# With a seed → same output every run.



# Example Without Seed

print(np.random.randint(1, 10, 5))


# Example With Seed

np.random.seed(42)
print(np.random.randint(1, 10, 5))
```
