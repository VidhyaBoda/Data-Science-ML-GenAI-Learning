# What is a String?

A string is a sequence of characters enclosed in single quotes '...', double quotes "...", or triple quotes '''...''' / """...""".

Strings are immutable, which means once created, you cannot change individual characters.

```python
s1='hello shivansh'

s2="hello shivansh"

s3='''Hello Shivansh 
is my name'''


print(s1)
print(s2)
print(s3)
```

# 2️⃣ String Characteristics

Ordered → Characters have an index.

Indexed → Access via positions: 0,1,2… or -1,-2…

Immutable → Cannot change a character.

Iterable → Can loop through characters.

```python
s="Pythonnnnn"
print(s[2])
print(s[-1])

for ch in s:
    print(ch)
```

# 4️⃣ String Indexing & Slicing

```python
# Indexing

s='Python'
print(s[1])
print(s[0])
print(s[-1])
print(s[-2])


s="vidhya"
print(s[-1])
```

```python
# Slicing

s="Python"
# print(s[start:stop:step])
print(s[0:4])
print(s[2:])
print(s[:4])
print(s[-4:-1])
print(s[0::2])
print(s[::-1])


print(s[2:-2])
print(s[:-2])

print(s[::-1])
```

# 5️⃣ String Methods

a) Case Conversion

```python
s='Python'
print(s.upper())
print(s.lower())
print(s.title())
print(s.capitalize())
```

```python
# Membership

s="My name is sumit"
print("is" in s)
```

b) Strip / Remove Whitespaces

```python
s=' python '
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

c) Replace / Find / Count

```python
s = "I like Python. Python is Fun! Python"

print(s.replace("Python","Data Science"))
print(s.find("Python"))
print(s.count("Python"))
```

d) Split & Join

```python
s="Python is Fun!"
words=s.split()
print(words)

new_s=",".join(words)
print(new_s)
```

e) Check Contents

```python
s="Python"
print(s.isalpha())
print(s.isdigit())
print(s.isalnum())
```

# 6️⃣ String Formatting

(a)Using format()

```python
name="Sumit"
age=22

print("My Name Is {} and My Age is {}.".format(name,age))
print("My Name is {1} and my age is {0} ".format(name,age))
print("My Name is {n} and my age is {m}.".format(n=age,m=name))
```

c) f-Strings (Python 3.6+)

```python
name=input("enter your name")
age=23

print(f"my name is {name}. And my age is {age}.")
# print("my name is" ,name,". And my age is ",age,".")


# name="hello\nworld"
# print(name)
```

```python
# 3️⃣ Convert "hello world" → "Hello World" using .title().

text = input("Enter a string: ")

result = text.title()

print("Converted string:", result)
```

```python
# 4️⃣ Reverse words of a sentence → "I love Python" → "Python love I". 


sentence = input("Enter a sentence: ")

words = sentence.split()      # split sentence into words
reversed_words = words[::-1]  # reverse the list of words

result = " ".join(reversed_words)

print("Reversed sentence:", result)
```

```python
# 5️⃣ Check if a sentence starts with "Hello" using .startswith().


sentence = input("Enter a sentence: ")

if sentence.startswith("Hello"):
    print("The sentence starts with 'Hello'")
else:
    print("The sentence does not start with 'Hello'")
```
