```python
# python me polymorphism ke types

# A.Function polymorphism(metho overloading)

# ek hi function alag alag data types ke liye alg result deta hai

print(len("hello"))
print(len([10,20,30]))
print(len((10,20,30)))

# len() ek hi function hai lekin input type ke hisab se result change hota hai
```

```python
# operator polymorphism

# ek hi operator alag sitution me alag kaan karta hai

print(5+10)
print("hello"+"world")
print([10,20]+[30,40])

# same operator but diffrent results                                                          

```

```python
# class polymorphism(method overriding)

# same method name,lekin child class me uska behiour alag hota hai.

class Animal:
    def sound(self):
        print("Animals Make Sound.")

class Dog(Animal): 
    def sound(self):
        print("Dog Barks")

class Cat(Animal):
    def sound(self):
        print("Cat Makes Sound")

# Polymorphism in action

animals=[Dog(),Cat()]

for a in animals:
    a.sound()


# ek hi method sound() lekin alg objects ke liye result alag
```

```python
# Duck Typing(special in python)

# python me object ki types se jyada usake behiour(method/attribute)pe dhyan dete hai.

class Duck:
    def swim(self):
        print("Duck swims in water.")

class Fish:
    def swim(self):
        print("Fish swim in water.")

def make_swim(func):
    func.swim()

make_swim(Duck())
make_swim(Fish())

# dono class me swim method hai python ne dono ko same treat kiya.

# Agar koi insaan guitar bajata hai → usse guitarist bol dete ho, chahe uska naam kuch bhi ho.

# Agar koi object me fly() method hai → Python use fly karne wala maan lega, chahe wo Bird ho ya Airplane.
```

```python
# Q-1: Create class Calculator with a method add() that can:
# ->add two integers.
# ->add three integers.
# ->add two strings(concatation).
# Here you need to use method overloading(by default arguments in python).

class Calculator:
    def add(self,a,b,c=None):
        return a+b if c is None else a+b+c
    
calc=Calculator()

print(calc.add(10,12)) #add two integers.
print(calc.add(10,20,30)) #add three integers.
print(calc.add("hello ","everyone"))
```

```python
# Q-2: Create a base class Employee with method get_role().
# ->Subclass Developer should override get_role() to return "Writes Code".
# ->Subclass Designer should override get_role() to return "Creates Designs".
# ==>Here You Use method overriding.

class Employee:
    def get_role(self):
        return "Employee role"
    
class Developer(Employee):
    def get_role(self):
        return "Writes Code"
    
class Designer(Employee):
    def get_role(self):
        return "Creates Designs"
    
emp = Employee()
dev = Developer()
des = Designer()
 
print(emp.get_role())
print(dev.get_role())
print(des.get_role())
```

```python
# Q-3: Create a class Vector with attributes x and y
# ->Overload thr + operator so that two vectors can be added.
# ->Example: v1+v2 should return a new Vector with summed coordinates.
# =>here you must use operator overloading(add).

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
        
    def __add__(self,other):
        return Vector(self.x + other.x, self.y + other.y)
    
v1 = Vector(2,3)
v2 = Vector(4,5)

v3 = v1 + v2

print(v3.x,v3.y)
```

```python
# Q-4: Create two classes PDFReader and DocReader, each with a method read().
# ->Write a function open_file(reader) that calls reader.read(),regardless of the object type.
# =>here you use duck typing.

class PDFReader:
    def read(self):
        print("Reading PDF file")
        
class DocReader:
    def read(self):
        print("Reading Doc file")
        
def open_file(reader):
    reader.read()
    
pdf = PDFReader()
doc = DocReader()

open_file(pdf)
open_file(doc)
```

```python
# Q-5: Create an abstract class Applliance with abstract method turn_on().
# ->subclass Fan,Light and Heater must implement turn_on() differently.
# ->Write a function start_all(Appliance) that loops over a list of appliances and calls turn_on().
# =>here you must use Abstract Class(ABC)+Overriding.

from abc import ABC,abstractmethod    #ABC : Abstract Base Class

# Abstract base class
class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass
    
#Subclasses
class Fan(Appliance):
    def turn_on(self):
        print("Fan is sppinning")
        
class Light(Appliance):
    def turn_on(self):
        print("Light is growing")
        
class Heater(Appliance):
    def turn_on(self):
        print("Heater is warming")
   

#Function using polymorphism

def start_all(appliances):
    for appliance in appliances:
        appliance.turn_on()
        
fan = Fan()
light = Light()
heater = Heater()

appliance_list = [fan,light,heater]

start_all(appliance_list)
```

```python
# Q-6: Create a class hierarchy for a Payment System.
# ->Base Class: Payment(abstract method make_payment(amount)).
# ->Subclass: CreditCard,Paypal,UPI-->override make_payment().
# ->Overloading: Add a method transaction() that can take 1 or 2 arguments(amount only, or amount+currency).
# ->Operator Overloading: Overloading+so two payments can be combined (eg: PayPal+UPI=combined balance).
# ->Duck Typing: Create a function process_payment(payment_method) that works with any class that has make_payment().


from abc import ABC,abstractmethod

# Abstract base class
class Payment(ABC):
    def __init__(self,balance):
        self.balance = balance 
        
    @abstractmethod
    def make_payment(self,amount):
        pass
        
        
    def transaction(self,amount,currency = "INR"):
        print(f"Transaction of {amount} {currency} initiated")
        
    def __add__(self,other):
        return PaymentSystem(self.balance + other.balance)
        
# Helper class for combined payment
class PaymentSystem(Payment):
    def make_payment(self, amount):
        print(f"Combined payment of {amount} processed")
        
# Subclasses
class CreditCard(Payment):
    def make_payment(self, amount):
        print(f"Credit Card payment of {amount} done")


class Paypal(Payment):
    def make_payment(self, amount):
        print(f"Paypal payment of {amount} done")


class UPI(Payment):
    def make_payment(self, amount):
        print(f"UPI payment of {amount} done")
        
#Duck typing function
def process_payment(payment_method):
    payment_method.make_payment(500)
    
    
#Objects
cc = CreditCard(2000)
pp = Paypal(1500)
upi = UPI(1000)

#Method overloading
pp.transaction(300)
pp.transaction(300, "USD")

#Operator overloading
combined = pp + upi
print("Combined Balance:",combined.balance)

#Duck typing
process_payment(cc)
process_payment(pp)
process_payment(upi)
```

3.Super()

super() parent class ke method ko call krne ke liye use hota hai.
ye mainly inheritance me kaam ata hai.


```python
class Parent:
    def show(self):
        print("This is Parent Class.")

class Child(Parent):
    def show(self):
        super().show() #parent ka method call 
        print("this is child class")
c=Child()
c.show()

# super().show() ne pahele parent ka show() call liya fir apna.
```

```python
# 3.__str__-->String representation

# object ko human readable string banata hai(jab print() karte ho)

class Book:
    def __init__(self,title):
        self.title=title

    def __str__(self):
        return f"Book:{self.title}"

b=Book("Python OOP")
print(b)
```

```python
# 4.__len__->length return

# len() function ke liye object ka length batata hai.

class Course:
    def __init__(self,students):
        self.students=students
    def __len__(self):
        return len(self.students)

c=Course(["A","B","C"])
print(len(c))
```

```python
# 5.__add__->operator overloading

# object ko + operator ke sath add karne deta hai

class Bank:
    def __init__(self,balance):
        self.balance=balance

    def __add__(self,other):
        return self.balance+other.balance

b1=Bank(1000)
b2=Bank(500)
print(b1+b2)
        
```

```python
# 6.__eq__->Equality check(==)
# objects ko == ke sath compare krne ke liye

class Pen:
    def __init__(self,color):
        self.color=color
    
    def __eq__(self,other):
        return self.color==other.color
    
p1=Pen("blue")
p2=Pen("blue")
print(p1==p2)
```
