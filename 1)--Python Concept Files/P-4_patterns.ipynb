# Pattern printing with loops

```python
#simple line

for i in range(5):
    print("*")
```

```python
#row me 5 stars

for i in range(5):
    print("*",end=' ')
```

```python
# square pattern

for i in range(5):
    for j in range(5):
        print("*",end=' ')
    print()
```

```python
# right triangle

for i in range(1,6):
    for j in range(i):
        print("*",end=' ')
    print()

```

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=' ')
    print()
```

```python
# inverted triangle

for i in range(5,0,-1):
    for j in range(i):
        print("*",end=' ')
    print()
```

```python
# pyramid pattern

for i in range(1,6):
    print(" "*(5-i),end='') #spaces
    print("* "*i) #stars
```

```python
# inverted pyramid

for i in range(5,0,-1):
    print(" "*(5-i),end='')
    print("* "*i)
```

```python
# diamond pattern
n=5

# upper pyramid
for i in range(1,n+1):
    print(" "*(n-i),end='')
    print("* "*i)

# lower inverted pyrmid

for i in range(n-1,0,-1):
    print(" "*(n-i),end='')
    print("* "*i)

print()

n=5

# upper pyramid
for i in range(1,n+1):
    print(" "*(n-i),end='')
    print("* "*i)

print()
# lower inverted pyrmid

for i in range(5,0,-1):
    print(" "*(n-i),end='')
    print("* "*i)
    
```

```python
# number pyramid

for i in range(1,6):
    print(" "*(5-1),end='')
    for j in range(1,i+1):
        print(j, end=' ')
    print()
```

```python
# floyds triangle

num=1
for i in range(1,6):
    for j in range(i):
        print(num,end=' ')
        num+=1
    print()
```

```python
# Extra Pattern Programs
```

```python
# 1. Square Star Pattern

n = 5
for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

```python
# 2. Hollow Square Pattern

n=9
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 or j==0 or j==n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 3. Right Angle Triangle (Numbers)

n = 5
for i in range(1, n+1):
    for j in range(1, i+1):
        print(j, end=" ")
    print()
```

```python
# 4. Pyramid Pattern

n = 7
for i in range(n):
    print(" "*(n-i-1) + "*"*(2*i+1))
```

```python
# 5. Inverted Pyramid

n = 6
for i in range(n):
    print(" "*i + "*"*(2*(n-i)-1))
```

```python
# 6. Number Pyramid

n=5
for i in range(1,n+1):
    print(" "*(n-i), end="")
    for j in range(1, i+1):
        print(j, end = "")
    for j in range(i-1,0,-1):
        print(j, end="")
    print()
```

```python
# 7. Alphabet Triangle

n=9
for i in range(n):
    for j in range(i+1):
        print(chr(65+j), end=" ")
    print()    
    
```

```python
# 8. Same Number Triangle

n=5
for i in range(1, n+1):
    print(str(i)*i)
```

```python
# 9. Cross Pattern

n = 9
for i in range(n):
    for j in range(n):
        if i == j or i + j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 10. Butterfly Pattern

n=7
for i in range(1, n+1):
    print("*"*i + " "*(2*(n-i)) + "*"*i)
    
for i in range(n, 0, -1):
    print("*"*i + " "*(2*(n-i)) + "*"*i)
```

```python
# 11 Zig-Zag Pattern

n = 3
for i in range(n):
    for j in range(9):
        if (i+j)%4==0 or (i==1 and j%4==2):
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 12. Hollow Diamond

n = 8
for i in range(n):
    for j in range(2*n):
        if j==n-i or j==n+i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
for i in range(n-2, -1, -1):
    for j in range(2*n):
        if j==n-i or j==n+i:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 13. Continuous Alphabet Pattern

ch = 65
n = 7
for i in range(1, n+1):
    for j in range(i):
        print(chr(ch), end=" ")
        ch += 1
    print()
```
