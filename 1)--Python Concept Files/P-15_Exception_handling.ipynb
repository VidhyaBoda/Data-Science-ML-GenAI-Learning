# What is Exception Handling?

An exception is an error that occurs during program execution and interrupts the normal flow.



```python
print(10/0)
print("hello")


# To prevent program crash, we use Exception Handling.
```

# 1. try and except

```python
try:
    x=10
    print(x/0)

except ZeroDivisionError:
    print("You Can not Divide by Zero.")

print("Hello")
    
# If an error occurs → except block runs.
# If no error → except is skipped.
```

# 2. Handling Multiple Exceptions

```python
try:
    x=int(input("Enter Number: "))
    print(x/0)

except ValueError:
    print("Please Enter a Number!")

except ZeroDivisionError:
    print("The number is not divided by zero")

```

# 3. Using else Block

```python
try:
    x=int(input("Enter a number: "))
    print(10/x)
except ZeroDivisionError:
    print("Number is Not Divisible by Zero!")

else:
    print("Code Run Succeffuly...")
```

# 4. Using finally Block

finally always executes — whether an error occurs or not.

Usually used to close files, release resources, etc.

```python
try:
    file=open("hello.txt","r")
    file.read()
except Exception as e:
    print("Error",e)

finally:
    file.close()
    print("File Closed Successfully!")
```

# 5. Catching All Exceptions (Generic)

```python
try:
    x = int("abc")
except Exception as e:
    print("Error occurred:", e)
```

# 6. Raising Exceptions Manually

You can raise custom errors using raise.

```python
age=12

if age<18:
    raise ValueError("Age must be greater than 18!")
```

7. Creating Custom Exception Class

You can define your own exception types.

```python
class NegativeNumberError(Exception):
    pass

try:
    num = int(input("Enter a positive number: "))
    if num < 0:
        raise NegativeNumberError("Negative numbers not allowed!")
except NegativeNumberError as e:
    print(e)
```

```python
# Q-1 Write a program that handles division by zero error.


try:
    x=int(input("Enter Number: "))
    print(x/0)

except ZeroDivisionError:
    print("The number is not divided by zero")
```

```python
# Q-2 Take user input and handle ValueError if input is not a number.

try:
    a = int(input("Enter a number:"))
    print("Your Entered:", a)
    
except ValueError:
    print("Error: Please enter valid integer.")
    
finally:
    print("Program Finished.")
```

```python
# Q-3 Handle both ZeroDivisionError and ValueError together.    
    
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    
    result = a / b
    print("Result:", result)

except (ZeroDivisionError, ValueError) as e:
    print("Error occurred:", e)

else:
    print("Calculation successful.")

finally:
    print("Program ended.")
```

```python
# Q-4 Try reading a non-existing file and handle the FileNotFoundError.

try:
    file = open("non_existing_file.txt", "r")
    content = file.read()
    print(content)
    file.close()
    
except FileNotFoundError:
    print("Error: The file does not exist.")
```

```python
# Q-5 Add an else block that prints a success message if no error occurs.

try:
    with open("source.txt", "r") as file:
        content = file.read()
        print(content)
        
except FileNotFoundError:
    print("Error: File not found.")
    
else :
    print("File read successfully. No errors occured.")
```

```python
# Q-6 Write a program to handle multiple specific exceptions in one try.

try:
    a=int(input("Enter First Number: "))
    b=int(input("Enter Second Number: "))
    
    result = a/b
    print("result:", result)
    
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

except ValueError:
    print("Error: Please enter valid integers.")

else:
    print("Calculation completed successfully.")

finally:
    print("Program execution finished.")
```

```python
# Q-7 Create a program that always prints "Program ended" using finally.

try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

except ValueError:
    print("Error: Invalid input. Please enter a number.")

finally:
    print("Program ended")
```

```python
# Q-8 Use Exception as e and print the error message.

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    
    result = a / b
    print("Result:", result)

except Exception as e:
    print("An error occurred:", e)
```

```python
# Q-9 Handle IndexError when accessing an invalid list index.

try:
    numbers = [10, 20, 30, 40]
    print(numbers[10])   # Invalid index

except IndexError:
    print("Error: List index out of range.")
```

```python
# Q-10 Raise a ValueError if the user enters a negative number.

try:
    num = int(input("Enter a number: "))

    if num < 0:
        raise ValueError("Negative numbers are not allowed.")

    print("You entered:", num)

except ValueError as e:
    print("Error:", e)
```

```python
# Q-11 Define a custom exception LowBalanceError for banking app if balance < 500.

class LowBalanceError(Exception):
    pass


def withdraw(balance, amount):
    if balance - amount < 500:
        raise LowBalanceError("Withdrawal denied! Minimum balance of 500 must be maintained.")
    
    return balance - amount

try:
    current_balance = 1000
    withdraw_amount = int(input("Enter withdrawal amount: "))

    new_balance = withdraw(current_balance, withdraw_amount)
    print("Withdrawal successful!")
    print("Remaining balance:", new_balance)

except LowBalanceError as e:
    print("LowBalanceError:", e)

    
except ValueError:
    print("Invalid amount entered.")
```

```python
# Q-12 Write a program that opens a file, writes data, and ensures it closes using finally.

file = None

try:
    file = open("data.txt", "w")
    file.write("This data is written to the file.\n")
    file.write("File handling with finally block.\n")
    print("Data written successfully.")

    with open("data.txt","r") as file:
        data=file.read()
        print(data)
        
        
except Exception as e:
    print("Error:", e)

finally:
    if file:
        file.close()
        print("File closed successfully.")
```
