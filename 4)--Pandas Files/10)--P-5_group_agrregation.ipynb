# 5. Grouping & Aggregation

idxmax(),idxmin():give me the index(row/label)of the maximum value in this series or dataframe.

```python
# 1.groupby()

# Used to split data into groups and apply functions(mean,sum,count,etc)

import pandas as pd

df=pd.DataFrame({
    "Department":["IT","IT","HR","HR","Sales"],
    "Salary":[50000,60000,45000,47000,55000]
})

# Average Salary per department

print(df.groupby('Department')['Salary'].mean())
```

```python
# 2. Multiple Aggregations(agg()):many functions at once.

df.groupby("Department")["Salary"].agg(['mean','sum','count'])
```

```python
# 3. Pivot Table:summary table 

print(df)
pd.pivot_table(df,values='Salary',index='Department',aggfunc='mean')
```

```python
# 4. Crosstab: compare categories (like gender vs department)

# for categorical data comparision.

df2=pd.DataFrame({
    "Gender":['M','F','M','F','M'],
    "Department":['IT','HR','IT','Sales','HR']
})

pd.crosstab(df2['Gender'],df2['Department'])
```

# Task-1: Student Marks

```python
data={ 
    "Class":[10,10,10,11,11,12,12], 
    "Student":["Rahul","Aman","Priya","Sneha","Kiran","Meena","Raju"], 
    "Subject":["Math","English","Math","English","Math","English","Math"], 
    "Marks":[88,75,92,85,80,90,70] 
}

df =pd.DataFrame(data)

print(df)
```

```python
# ->Find Average marks per class.

print(df.groupby("Class")["Marks"].mean())
```

```python
# ->Find highest marks per subject.

print(df.groupby("Subject")["Marks"].max())
```

```python
# ->Count how many students are in each class.

print(df.groupby("Class")["Student"].count())
```

```python
# ->Create a pivote table:Class Vs Subject with average marks.

print(df)

pd.pivot_table(df,values="Class",index = "Subject",aggfunc = "mean")

# pivot = pd.pivot_table(
#     df,
#     values="Marks",
#     index="Class",
#     columns="Subject",
#     aggfunc="mean"
# )

# print(pivot)
```

# Task-2:Online Store Orders

```python
import pandas as pd

data={ 
    "OrderID":[1,2,3,4,5,6],
    "Quantity":[5,2,2,3,1,4],
    "Product":["Pen","Book","Laptop","Mouse","Shirt","Jeans"], 
    "Category":["Stationery","Stationery","Electronics","Electronics","Clothing","Clothing"], 
    "Price":[10,100,50000,1000,700,1200] 
}

df = pd.DataFrame(data)
print(df)
```

```python
# ->Find total sales(Price*Quantity) for each category.

df["Sales"] = df["Price"] * df["Quantity"]

total_sales = df.groupby("Category")["Sales"].sum()

print(total_sales)
```

```python
# ->Find average price per category.

print(df.groupby("Category")["Price"].mean())
```

```python
# ->Use agg() to calculate min,max and mean price per category.

print(df)
df.groupby("Category")["Price"].agg(["min","max","mean"])
```

```python
# ->Pivot table: Category Vs Product->total sales.

print(pd.pivot_table(df,
                     values = "Category",
                     index = "Product", 
                     aggfunc = "sum"
                    )
     )

print("-------------")
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print(pivot)
```

# Task-3: Employees data

```python
# ->Find Average experience per department

df.groupby("Dept")["Experience"].mean()
```

# Task-4: Movie Ratings

```python
data={ 
    "Movie":["Inception","Inception","Titanic","Titanic","Avengers","Avengers"], 
    "Genre":["Sci-Fi","Sci-Fi","Romance","Romance","Action","Action"], 
    "User":["U1","U2","U3","U4","U5","U6"], 
    "Rating":[9,8,9,7,8,9] 
}

df = pd.DataFrame(data)

df
```

```python
# ->Find average rating per genre.

df.groupby("Genre")["Rating"].mean()
```

```python
# ->Count how many ratings each movie received.

df.groupby("Movie")["Rating"].count()
```

```python
# ->Which genre got the highest average rating?

highest_genre = df.groupby("Genre")["Rating"].mean().idxmax()
print(highest_genre)
```

```python
# ->Pivot table: Genre Vs Movie->average rating.

pd.pivot_table(df,values = "Rating",index = "Genre", columns = "Movie",aggfunc = "mean")

```

# Task-5: Sales by region

```python
data={ 
    "Region":["North","North","South","South","East","East"], 
    "Salesman":["Ramesh","Suresh","Raj","Meena","Aman","Sneha"], 
    "Product":["Pen","Book","Pen","Book","Pen","Bag"], 
    "Sales":[200,300,150,250,180,400] 
}

df = pd.DataFrame(data)

df
```

```python
# ->Find total sales per region.

df.groupby("Region")["Sales"].sum()
```

```python
# ->Find total sales per product.

df.groupby("Product")["Sales"].sum()
```

```python
# ->Pivot table: Region Vs Product-> total sales.

pd.pivot_table(df,values = "Sales", index = "Region", columns = "Product",aggfunc = "sum")
```

```python
# ->Crosstab: Region Vs Salesman->count of sales records.

pd.crosstab(df["Region"],df["Salesman"])
```
