```python
# Q11. Student Marks Simulation
# Simulate marks (0–100) of 15 students using np.random.randint()
# Find the top 5 marks (sorted descending).
# Randomly select 3 students for a prize.

np.random.seed(42)

marks = np.random.randint(0,101, size = 15)

top_5 = np.sort(marks)[::-1][:5]

selected_students = np.random.choice(len(marks), size = 3,replace = False)


# print("Marks:",marks)
print("Top 5 marks (decending):",top_5)
# print("Selected student indices:",selected_students)
# print("Selected students marks:",marks[selected_students])
```

```python
# Q12. Daily Sales Data
# Generate daily sales for 7 days between 200 and 1000 units.
# Find the day with the highest sales.
# Randomly pick 2 days for a special offer.
# Shuffle the sales data.

daily_sales = np.random.randint(200,1001, size = 7)

max_day = np.argmax(daily_sales)       
max_sales = daily_sales[max_day]

special_days = np.random.choice(len(daily_sales), size=2, replace=False)

np.random.shuffle(daily_sales)

print(daily_sales)
print(max_sales,max_day)
print(special_days)
```

```python
# Q13. Employee Salary Data
# Create salary data for 8 employees between 30,000 and 80,000.
# Find the employee with the lowest salary.
# Randomly pick 2 employees for a bonus.

salary_data = np.random.randint(30000,80001, size = 8)

min_emp_index = np.argmin(salary_data)

min_salary = salary_data[min_emp_index]

bonus_employees = np.random.choice(len(salary_data), size = 2, replace = False)

print(salary_data)
print(min_salary,min_emp_index)
print(bonus_employees)
print(salary_data[bonus_employees])
```

```python
# Q14. Product Prices
# You have 10 products. Generate random prices between ₹100 and ₹1000.
# Sort the prices in ascending order.
# Pick 3 random products for a discount.

import numpy as np

np.random.seed(11)

prices = np.random.randint(100, 1001, size=10)

sorted_prices = np.sort(prices)

discount_products = np.random.choice(len(prices), size=3, replace=False)

print(prices)
print(sorted_prices)
print(discount_products)
print(prices[discount_products])
```

```python
# Q15. Weather Temperature Data
# Simulate 10 random daily temperatures between 25°C and 45°C.
# Find 3 hottest days using argsort().
# Randomly shuffle the temperature data.

import numpy as np

# Set seed for reproducibility (optional but recommended in real projects)
np.random.seed(42)

# 1️⃣ Simulate 10 random daily temperatures between 25°C and 45°C
temps = np.random.randint(25, 46, 10)
print("Daily Temperatures:", temps)

# 2️⃣ Find 3 hottest days using argsort()
# argsort gives indices sorted in ascending order
hottest_indices = np.argsort(temps)[-3:]   # last 3 indices (largest values)
print("Indices of 3 Hottest Days:", hottest_indices)
print("Top 3 Hottest Temperatures:", temps[hottest_indices])

# 3️⃣ Randomly shuffle the temperature data
np.random.shuffle(temps)
print("Shuffled Temperatures:", temps)
```

```python
# Q16. Exam Simulation
# You are simulating marks of 1000 students (0–100).
# Find the average mark.
# Find how many scored above 80.
# Randomly select 10 for interview.

import numpy as np

# Reproducibility (important in analytics work)
np.random.seed(42)

# 1️⃣ Simulate marks of 1000 students (0–100 inclusive)
marks = np.random.randint(0, 101, 1000)

# 2️⃣ Find the average mark
average_mark = np.mean(marks)

# 3️⃣ Count how many scored above 80
above_80_count = np.sum(marks > 80)

# 4️⃣ Randomly select 10 students for interview (without replacement)
selected_students = np.random.choice(len(marks), 10, replace=False)

print("Average Mark:", average_mark)
print("Number of Students Scored Above 80:", above_80_count)
print("Selected Student Indices for Interview:", selected_students)
print("Their Marks:", marks[selected_students])
```

```python
# Q17. Random Sampling in Population
# Create an array representing 1000 people IDs (1–1000).
# Randomly select 50 people for a survey (no repetition).
# Randomly select 5 people for a prize (with repetition).

import numpy as np

# Reproducibility (important in analytics experiments)
np.random.seed(42)

# 1️⃣ Create array of 1000 people IDs (1–1000)
people_ids = np.arange(1, 1001)

# 2️⃣ Randomly select 50 people for survey (no repetition)
survey_sample = np.random.choice(people_ids, 50, replace=False)

# 3️⃣ Randomly select 5 people for prize (with repetition allowed)
prize_sample = np.random.choice(people_ids, 5, replace=True)

print("Survey Sample (50 unique people):")
print(survey_sample)

print("\nPrize Winners (duplicates possible):")
print(prize_sample)
```

```python
# Q18. Game Simulation
# A dice game is played 10 times.
# Simulate the results using np.random.randint(1, 7, 10).
# Find how many times 6 appeared.

import numpy as np

# Reproducibility (optional but recommended)
np.random.seed(42)

# 1️⃣ Simulate dice rolled 10 times (1 to 6)
dice_rolls = np.random.randint(1, 7, 10)

# 2️⃣ Count how many times 6 appeared
six_count = np.sum(dice_rolls == 6)

print("Dice Rolls:", dice_rolls)
print("Number of times 6 appeared:", six_count)
```
