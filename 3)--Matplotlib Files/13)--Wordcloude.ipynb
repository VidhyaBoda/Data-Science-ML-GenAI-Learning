```python
from wordcloud import WordCloud


text="Python is easy to learn.I love python.python helps a lot in datascience."


wc=WordCloud(width=800,height=500,background_color='white').generate(text)


import matplotlib.pyplot as plt

plt.imshow(wc,interpolation='bilinear')
plt.axis('off')
plt.show()
```

```python
text=open('wordcloud.txt').read()


wc=WordCloud(width=900,height=700,max_words=100,background_color='black',colormap='coolwarm').generate(text)


plt.imshow(wc,interpolation='bilinear')
plt.axis('off')
plt.show()
```
