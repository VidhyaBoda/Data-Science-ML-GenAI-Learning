# Task 1️⃣– Horizontal Bar Chart

```python
#  Create a horizontal bar chart for:
# o Categories: Tea, Coffee, Juice, Water
# o Quantity: any 4 numbers
#  Change:
# o Bar color
# o Title
# o X-label and Y-label

Categories = ['Tea','Coffee','Juice','Water']

Quantity = [20,70,50,100]

colors = ['orange','green','blue','black']

plt.barh(Categories,Quantity, color = colors,height = 0.5)

plt.xlabel("Quantity")
plt.ylabel("Categories")

plt.title("Horizontal Bar Chart")

plt.show()
```

# Task 3️⃣– Histogram (Basic)

```python
#  Generate 50 random numbers between 1–50
#  Plot a histogram
#  Add:
# o X label
# o Y label
# o Title

import matplotlib.pyplot as plt
import numpy as np

num = np.random.randint(1,50,50)

plt.hist(num, color='purple',edgecolor='black',alpha=0.6)


plt.xlabel("X-Values")
plt.ylabel("y-Values")

plt.title("Histrogram")
plt.show()
```

# Task 5️⃣– Horizontal Bar Chart (Stock Data)

```python
#  Products: Pen, Pencil, Eraser, Scale
#  Stock values: random numbers
#  Plot barh
#  Change:
# o Transparency
# o Title

import matplotlib.pyplot as plt

import numpy as np

Products = ['Pen','Pencil','Eraser','Scale']

stock = np.random.randint(10,100, size = 4)

plt.figure(figsize=(8,5))
plt.barh(Products, stock, alpha = 0.6, height = 0.5)

plt.title("Current Product Stock Levels", fontsize=14)

plt.xlabel("Stock Quantity")
plt.ylabel("Products")

plt.show()
```

# Task 7️⃣– Compare Two Charts

```python
#  Plot:
# o One vertical bar chart
# o One horizontal bar chart
#  Use same data
#  Observe: Which looks clearer?


import matplotlib.pyplot as plt

products = ['Pen', 'Pencil', 'Eraser', 'Scale']
stock = [50, 75, 30, 60]

plt.figure(figsize=(12,5))

# Vertical Bar Chart
plt.subplot(1,2,1)
plt.bar(products, stock, width=0.5)
plt.title("Vertical Bar Chart")
plt.xlabel("Products")
plt.ylabel("Stock")

# Horizontal Bar Chart
plt.subplot(1,2,2)
plt.barh(products, stock, height=0.5)
plt.title("Horizontal Bar Chart")
plt.xlabel("Stock")
plt.ylabel("Products")


plt.show()
```

# Task 8️⃣– Histogram Meaning

```python
#  Generate random marks of 100 students
#  Plot histogram
#  Answer (in comments):
# o Which marks range has the highest frequency?

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(12)
marks = np.random.randint(0,101,100)

plt.hist(marks, bins = 10,edgecolor = 'black',alpha = 0.7)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

# Task 9️⃣– Real-Life Data

```python
#  Create a bar chart for:
# o Days: Mon, Tue, Wed, Thu, Fri
# o Study hours per day
#  Add:
# o Title
# o Labels
# o Colors


Days = ['Mon','Tue','Wed','Thu','Fri']

Study_hours = [10,12,5,7,9]

Colors = ['Blue','Green','Red','Yellow','Orange']
plt.bar(Days,Study_hours, color = Colors, width = 0.5,alpha = 0.7,edgecolor='black')

plt.title("Real-Life Data")

plt.xlabel("Days")
plt.ylabel("Study Hour")

plt.show()
```
