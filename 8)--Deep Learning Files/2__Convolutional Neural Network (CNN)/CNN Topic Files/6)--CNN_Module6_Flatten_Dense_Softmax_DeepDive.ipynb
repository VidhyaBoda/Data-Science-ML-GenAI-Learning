# 1. Where Are We in CNN?

Typical CNN Flow:

Image
↓
Convolution
↓
ReLU
↓
Pooling
↓
Convolution
↓
ReLU
↓
Pooling
↓
Flatten
↓
Dense
↓
Softmax
↓
Prediction

The convolution part extracts features.

The Dense + Softmax part makes the final decision.


# 2. Why Flatten is Needed

Convolution and Pooling produce 3D outputs.

Example:

14 × 14 × 128

Dense layers expect 1D vectors.

Therefore we need Flatten.


# Example of Flatten

Input:

4 × 4 × 3

Total values:

4 × 4 × 3

= 48

After Flatten:

Shape:

(48,)

All values become a single vector.


# Important Facts About Flatten

Flatten:

- Has no trainable parameters
- Performs no learning
- Only reshapes data

Purpose:

Convert 3D feature maps into 1D vectors.


```
from tensorflow.keras.layers import Flatten

layer = Flatten()

print(layer)
```

# 3. Dense Layer

Dense Layer = Fully Connected Layer

Every input neuron is connected to every output neuron.

This is similar to a traditional Artificial Neural Network (ANN).


# Example

Flatten Output:

48 values

Dense Layer:

Dense(10)

Means:

48 Inputs
↓
10 Neurons


# Neuron Calculation

Each neuron computes:

y = wx + b

Example:

x1 = 2
x2 = 3

w1 = 4
w2 = 5

bias = 1

Output:

(2×4)+(3×5)+1

= 24


# What Does Dense Layer Learn?

Convolution learns:

- Edges
- Curves
- Eyes
- Ears
- Shapes

Dense Layer combines these features.

Example:

Eye + Ear + Nose
↓
Cat

This is the reasoning stage of CNN.


# Multiple Dense Layers

Example:

Dense(128)
↓
Dense(64)
↓
Dense(32)

Each layer learns increasingly complex relationships.


# 4. Output Layer

Suppose dataset:

Cat
Dog
Horse

Total Classes:

3

Output Layer:

Dense(3)

One neuron for each class.


# Raw Outputs (Logits)

Example:

[4.5, 2.1, 0.8]

These are called logits.

They are not probabilities.


# 5. Softmax Layer

Softmax converts logits into probabilities.

Input:

[4.5, 2.1, 0.8]

Output:

[0.89, 0.09, 0.02]

Meaning:

Cat = 89%
Dog = 9%
Horse = 2%


# Property of Softmax

Probabilities always sum to 1.

Example:

0.89 + 0.09 + 0.02

= 1


# Binary Classification

Example:

Cat vs Dog

Output Layer:

Dense(1, activation='sigmoid')

One output neuron is enough.


# Multi-Class Classification

Example:

Cat
Dog
Horse
Car

Output Layer:

Dense(4, activation='softmax')

One neuron per class.


# Dense Layer Parameter Calculation

Formula:

(Input Neurons × Output Neurons)
+ Output Neurons

Example:

Input = 100
Output = 50

Parameters:

(100 × 50)+50

= 5050


# Why Dense Layers Have Many Parameters

Flatten:

25088 values

Dense:

128 neurons

Parameters:

(25088 × 128)+128

Very large.

Dense layers often contain most CNN parameters.


# CNN Decision Making Process

Convolution:
Finds features

ReLU:
Keeps useful activations

Pooling:
Reduces dimensions

Flatten:
Converts 3D → 1D

Dense:
Combines features

Softmax:
Produces probabilities

Prediction:
Final class label


```
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

model = Sequential()

model.add(Flatten())

model.add(
    Dense(
        128,
        activation='relu'
    )
)

model.add(
    Dense(
        3,
        activation='softmax'
    )
)
```
