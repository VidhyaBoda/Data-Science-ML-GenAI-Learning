# DBSCAN vs K-Means

| Feature               | K-Means | DBSCAN    |
| --------------------- | ------- | --------- |
| K Required            | Yes     | No        |
| Noise Detection       | No      | Yes       |
| Arbitrary Shapes      | No      | Yes       |
| Outlier Handling      | Poor    | Excellent |
| Speed                 | Faster  | Slower    |
| Customer Segmentation | Good    | Good      |
| Anomaly Detection     | Poor    | Excellent |
