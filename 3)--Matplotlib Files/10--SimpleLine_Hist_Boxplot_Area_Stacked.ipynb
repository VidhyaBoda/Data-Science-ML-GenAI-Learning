# Simple Line Chart 

```python
import matplotlib.pyplot as plt


xpoint=[1,2,3,4,5]
ypoint=[2,4,6,8,9]


x1=[10,20,30,40]
y1=[12,14,25,23]


plt.plot(xpoint,ypoint,marker='^',color='red',linestyle='dashed',label='Group 1')
plt.plot(x1,y1,color='green',marker='o',linestyle='-',label='Group-2')

plt.legend(loc='lower right')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Simple Line Chart")
plt.xticks(rotation=45)
plt.show()
```

# Multiple Boxplot

```python
import matplotlib.pyplot as plt


maths=[12,13,15,20,25]
english=[45,47,49,50,32]
science=[90,91,93,94,95]


plt.boxplot([maths,english,science],labels=['Maths','English','Science'])
plt.show()
```

# Area Line Plot

```python
xpoint=[1,2,3,4,5,6]
ypoint=[2,4,9,12,13,14]

plt.plot(xpoint,ypoint,color='black')
plt.fill_between(xpoint,ypoint,color='skyblue')

plt.show()
```

# Stacked Area Chart

```python
product_a=[12,13,14,15,19]
product_b=[20,34,25,29,35]
product_c=[87,85,83,82,80]


plt.stackplot(product_a,product_b,product_c,labels=['Product 1','Product 2','Product 3'])
plt.legend()
plt.show()
```
