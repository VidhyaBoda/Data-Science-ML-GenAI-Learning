# 🧩 1. What is a Tuple?

👉 Definition:
A tuple is an ordered, immutable (cannot be changed) collection of elements in Python.

It allows different data types (like int, str, float, etc.).

Written using round brackets ( ).

student = ("Shivansh", 22, "India")


✅ Ordered
✅ Allows duplicates
✅ Immutable ❌ (cannot modify elements)

# 2. How to Create a Tuple?

# # normal tuple
t1 = (1, 2, 3, 4)

# tuple with mixed data types
t2 = ("Vidhya", 22, True, 78.5)

# single element tuple (important!)
t3 = (5,)   # must include comma


⚠️ Without the comma → Python treats it as an integer, not a tuple.

x = (5)     # ❌ not a tuple
print(type(x))  # <class 'int'>

y = (5,)    # ✅ tuple
print(type(y))  # <class 'tuple'>

# 📦 3. Accessing Elements

Tuples are index-based, just like lists.

fruits = ("apple", "banana", "mango")
print(fruits[0])   # apple
print(fruits[-1])  # mango

# ✂️ 4. Tuple Slicing

```python
nums = (10, 20, 30, 40, 50)

print(nums[1:4])   # (20, 30, 40)

print(nums[::-1])  # (50, 40, 30, 20, 10)
```

# 🔒 5. Tuple is Immutable (Cannot Change)
t = (1, 2, 3)
t[0] = 99  # ❌ Error: 'tuple' object does not support item assignment


But you can convert it to a list to modify:

t = (1, 2, 3)
lst = list(t)
lst[0] = 99
t = tuple(lst)
print(t)


🧾 Output → (99, 2, 3)

# 🔄 6. Tuple Operations

```python
Operation	Example	Output
Concatenation	(1,2) + (3,4)	(1,2,3,4)

 
Repetition	("Hi",)*3	('Hi', 'Hi', 'Hi')
Membership	3 in (1,2,3)	True
Length	len((1,2,3))	3
```

# 🧰 7. Tuple Methods (Only Two!)
Method	Description	Example

.count(x)	Count of value	(1,2,2,3).count(2) → 2

.index(x)	Index of first value	(1,2,3).index(3) → 2

# 🧮 8. Tuple Functions

```python
nums = (10, 20, 5, 40)
print("Length:", len(nums))
print("Max:", max(nums))
print("Min:", min(nums))
print("Sum:", sum(nums))
```

# 🔁 9. Loop Through Tuple

```python
colors = ("red", "blue", "green")
for c in colors:
    print(c)
```

# 🧩 10. Nested Tuples

```python
# Tuples can contain other tuples.

matrix = ((1,2,3), (4,5,6), (7,8,9))
print(matrix[1][2])   # 6
```

# 🔄 11. Tuple Unpacking

```python
# Extract tuple values directly into variables.

student = ("Shivansh", 22, "India")
name, age, country = student
print(name)
print(age)
print(country)
```

# 🧠 12. Tuple vs List (Very Important Difference)

```python
Feature	List	Tuple
Brackets	[ ]	( )
Mutable	✅ Yes	❌ No
Methods	Many	Only 2
Speed	Slower	Faster
Use Case	When data changes	When data should not change
```
