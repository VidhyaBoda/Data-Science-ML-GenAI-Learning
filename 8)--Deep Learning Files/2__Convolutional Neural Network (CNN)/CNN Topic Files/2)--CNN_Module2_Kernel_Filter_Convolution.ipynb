# 1. What is a Kernel?

A kernel is a small matrix that slides over an image.

Example:

[[1,0],
 [0,1]]

CNN uses kernels to detect patterns such as:

- Edges
- Corners
- Curves
- Textures

In real CNNs, kernels are learned automatically during training.


# 2. Why Do We Need Kernels?

An image may contain millions of pixels.

CNN does not look at the entire image at once.

Instead, it examines small local regions using kernels.

This greatly reduces parameters and computation.


# 3. Convolution Concept

Convolution steps:

1. Place kernel on image
2. Multiply corresponding values
3. Add results
4. Store output
5. Slide kernel
6. Repeat

Output is called a Feature Map.



# 4. Manual Convolution Example

Image

\[
\begin{bmatrix}
1&2&3\\
4&5&6\\
7&8&9
\end{bmatrix}
\]

Kernel

\[
\begin{bmatrix}
1&0\\
0&1
\end{bmatrix}
\]

First output:

(1×1)+(2×0)+(4×0)+(5×1)=6


# 5. Feature Maps

The result produced after convolution is called a Feature Map.

Each filter produces one feature map.

Example:

Conv2D(32,(3,3))

Produces:

32 Feature Maps


# 6. Filter vs Kernel

Kernel:
Single weight matrix.

Filter:
Collection of kernels across all image channels.

For RGB images:

One filter contains:
- Red kernel
- Green kernel
- Blue kernel


# 7. RGB Convolution

Input Shape:

32 × 32 × 3

Kernel Size:

3 × 3

Actual filter shape:

3 × 3 × 3

CNN performs convolution on all channels and combines the results.


# 8. Conv2D Example

Conv2D(32,(3,3))

Meaning:

- 32 filters
- Each filter size 3×3
- Produces 32 feature maps

Output depth becomes 32.


# 9. Parameter Calculation

Formula:

Parameters =
((Filter Height × Filter Width × Channels)+1)
× Number of Filters

Example:

Conv2D(32,(3,3))

Input channels = 3

((3×3×3)+1)×32

= 896 parameters
