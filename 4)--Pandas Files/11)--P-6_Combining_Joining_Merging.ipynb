# Concatenation in pandas


Concatenation ka matlab hai 2 ya jyada DataFrame ko ek sath jodna.

ye 2 tareeke se hota hai:

1.Row-wise(axis=0)->Rows ke niche rows chipkana.

2.Column-wise(axis=1)->Columns ke side me chipkana


```python
# Ex:Row-wise Concatenation

import pandas as pd

df1=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Amit","Riya","Sumit"]
})

df2=pd.DataFrame({
    "ID":[4,5,6],
    "Name":["Neha","Raj","Ranu"]
})

# concatenate row-wise(default axis=0)
result=pd.concat([df1,df2])
print(result)
print("-----------")
# Notice:-Index reset nai hua dono df ke apne index aye.
# agar rest chahiye

result=pd.concat([df1,df2],ignore_index=True)
print("After index ignore:\n",result)
```

```python
# Ex-Column-wise concatenation

df3=pd.DataFrame({
    "Age":[11,22,13]
})

result3=pd.concat([df1,df3],axis=1)
print(result3)
```

```python
# Ex:-Handling diffrent columns

df4=pd.DataFrame({
    'Roll':[101,102,103]
})

result=pd.concat([df1,df4],axis=1)
print(result)
```

# Merging in pandas:

pd.merge() do DataFrame ko common column ya index ke basis pr join krta hai.

ye SQL JOIN ke types support krta hai.

->Inner Join: Sirf common rows.

->Left Join: Left DataFrame ki sari rows,right ke matching rows.

->Right Join: Right DataFrame ki saari rows, left ke matching rows.

->Outer Join: dono DataFrame ki Sari rows,missing me NaN.


```python
# Example Data:-

import pandas as pd

df1=pd.DataFrame({
    "EmpID":[1,2,3,4],
    "Name":["Amit","Riya","Sumit","Ranu"]
})

df2=pd.DataFrame({
    "EmpID":[3,4,5,6],
    "Salary":[30000,40000,50000,60000]
})
```

```python
# 1. Inner Join(default)

result=pd.merge(df1,df2,on='EmpID',how='inner')
print(result)
```

```python
# 2.Left join

result=pd.merge(df1,df2,on='EmpID',how='left')
print(result)
```

```python
# 3.Right Join

result=pd.merge(df1,df2,on='EmpID',how='right')
print(result)
```

```python
# 4.Outer Join

result=pd.merge(df1,df2,on='EmpID',how='outer')
print(result)
```

Joining with Index

Agar columns same nahi hai but index same hai, to hum join() use krte hai.


```python
df3=pd.DataFrame({
    'Dept':['IT','HR','Sales','Finance']},index=[1,2,3,4]
)

result=df1.set_index('EmpID').join(df3)
print(result)
```
