# Task 6:

```python
# Change the color of the box plot using:
# patch_artist=True
# boxprops=dict(facecolor='lightblue')

import matplotlib.pyplot as plt

# Data
math =    [65, 70, 75, 80, 85, 90]
science = [60, 68, 74, 78, 82, 88]
english = [72, 76, 79, 83, 87, 91]
hindi =   [70, 72, 75, 80, 85, 90]

data = [math, science, english, hindi]

# Create colored box plot
plt.figure()

plt.boxplot(data,
            patch_artist=True,
            boxprops=dict(facecolor='lightblue'))

plt.xticks([1, 2, 3, 4], ['Math', 'Science', 'English', 'Hindi'])

plt.title("Marks Distribution Across 4 Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

# Task 8:

```python
# Create a word cloud using your own text:
# ➡ “I love Python. Python is awesome. Learning data science is fun.”

from wordcloud import WordCloud

text = "I love Python. Python is awesome. Learning data science is fun."

wc = WordCloud(width = 800,height = 500, background_color = 'white').generate(text)

import matplotlib.pyplot as plt

plt.imshow(wc,interpolation = 'bilinear')

plt.axis('off')

plt.show()
```

# Task 10:

```python
# Generate 3 word clouds using different colormaps:
# • cool
# • plasma
# • spring

import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Text data
text = """
WordCloud in matplotlib is powerful for text visualization.
It helps in representing frequency of words visually.
Data visualization, analysis, python, matplotlib, wordcloud,
analytics, charts, graphs, insights, reporting, dashboard.
"""

# Colormaps to use
colormaps = ['cool', 'plasma', 'spring']

for cmap in colormaps:
    wc = WordCloud(width=600,
                   height=400,
                   background_color='white',
                   colormap=cmap).generate(text)
    
    plt.figure()
    plt.imshow(wc, interpolation='bilinear')  # intepolation = bininear,nearest,none
    plt.axis('off')
    plt.title(f"WordCloud using '{cmap}' colormap")
    plt.show()
```

# Task 11:

```python
# Limit the word cloud to max_words = 20

import matplotlib.pyplot as plt
from wordcloud import WordCloud

text = """
WordCloud in matplotlib is powerful for text visualization.
It helps represent frequency of words visually.
Data visualization, python, matplotlib, analytics,
dashboard, reporting, charts, graphs, insights,
machine learning, statistics, business intelligence,
data science, automation, engineering, programming.
"""

wc = WordCloud(width=600,
               height=400,
               background_color='skyblue',
               colormap='viridis',
               max_words=20).generate(text)

plt.figure()
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud (Top 20 Words)")
plt.show()
```

# Task 12:

```python
# Read text from a new file:
# ➡ new_para.txt
# (Create the file yourself)

import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Step 1: Create and write to file
with open("new_para.txt", "w") as file:
    file.write("""
    WordCloud visualization helps in understanding text data.
    It highlights the most frequent words clearly.
    Data science, analytics, python programming,
    machine learning, statistics, visualization,
    dashboards and reporting are important skills.
    """)

# Step 2: Read text from file
with open("new_para.txt", "r") as file:
    text = file.read()

# Step 3: Generate WordCloud
wc = WordCloud(width=600,
               height=400,
               background_color='white',
               colormap='cool',
               max_words=20).generate(text)

# Step 4: Display
plt.figure()
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud from File: new_para.txt")
plt.show()
```

# Task 13:

```python
# Change the word cloud size to:
# ➡ width = 1000, height = 600

import matplotlib.pyplot as plt
from wordcloud import WordCloud

text = """
WordCloud visualization helps in understanding text data.
It highlights the most frequent words clearly.
Data science, analytics, python programming,
machine learning, statistics, visualization,
dashboards and reporting are important skills.
"""

wc = WordCloud(width=1000,
               height=600,
               background_color='white',
               colormap='plasma',
               max_words=20).generate(text)

plt.figure(figsize=(10, 6))   # Match display size
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud (1000 x 600 Resolution)")
plt.show()
```

# Task 14:

```python
# Save your word cloud image using:
# wc.to_file("my_wordcloud.png")

import matplotlib.pyplot as plt
from wordcloud import WordCloud


text = """
WordCloud visualization helps in understanding text data.
It highlights frequent words clearly.
Data science, analytics, python, machine learning,
statistics, reporting, dashboards.
"""

wc = WordCloud(width=1000,
               height=600,
               background_color='white',
               colormap='plasma',
               max_words=20).generate(text)

# Save image to file
wc.to_file("my_wordcloud.png")


plt.figure(figsize=(10, 6))
plt.imshow(wc, interpolation='nearest')
plt.axis('off')
plt.title("Saved WordCloud Image")
plt.show()
```
