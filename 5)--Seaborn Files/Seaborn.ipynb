```python
import pandas as pd

import seaborn as sns

df=pd.DataFrame({
    "Age":[20,25,30,35,40],
    "Salary":[20000,25000,30000,35000,40000],
    "Gender":['M','F','F','M','M']
})


sns.histplot(df['Salary'],kde=True,palette='dark')
plt.show()
```

```python
sns.histplot(data=df,x='Salary',hue='Gender',kde=True)
plt.show()
```

```python
sns.countplot(x='Gender',data=df,hue='Gender')
plt.show()
```

```python
sns.barplot(x='Gender',y='Salary',data=df,errorbar='ci')
plt.show()
```

```python
df=pd.DataFrame({
    "Age":[20,25,30,35,40],
    "Salary":[20000,25000,30000,35000,40000],
    "Gender":['M','F','F','M','M'],
    "Experience":[2,4,8,10,12]
})


import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x="Gender",y="Salary",data=df,hue="Gender")
plt.show()
```

```python
sns.violinplot(x="Gender",y="Salary",data=df,hue="Gender")
plt.show()
```

```python
sns.stripplot(x="Gender",y="Salary",data=df)
plt.show()
```

```python
sns.boxplot(x="Gender",y="Salary",data=df)
sns.swarmplot(x="Gender",y="Salary",data=df,color='black')
plt.show()
```

```python
sns.scatterplot(x='Age',y="Salary",hue="Gender",size='Experience',data=df,style="Gender")
plt.show()
```
