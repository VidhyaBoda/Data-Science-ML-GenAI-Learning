```python
# Forward fill: Upar vali row ka value niche copy hojata hai

import pandas as pd

data={
    "Name":["Amit","Riya","Sumit","Pooja","Raj"],
    "Math":[78,None,65,88,None],
    "Science":[85,88,None,90,75],
    "English":[80,None,72,85,70]
}

df=pd.DataFrame(data)

print("Original:\n",df)

print("-------------")
# forward fill

df_ffill=df.fillna(method="ffill")
print("\nAfter Forward Fill:\n",df_ffill)

df_ffill
```

```python
# Backward fill

# Neche wali row ka value upar copy ho jata hai.

df_bfill=df.fillna(method="bfill")
print("\nAfter Backward Fill:\n",df_bfill)
```

```python
# Conditional Fill

# Specific condtion ke hisab se fill krna.

# ex: Agar English marks missing hai to 70 daal do

df.loc[df['English'].isnull(),'English']=70
print("\nAfter Conditional Fill:\n",df)
```

```python
# Column Specific missing values handling : har column ke liye alag-alg tarike se missing values bharna.

# 1.Ek column ke liye fillna

import pandas as pd

data={
    "Name":["Aman","Sumit","Ranu",None],
    "Age":[20,None,25,30],
    "Salary":[50000,60000,None,70000]
}

df=pd.DataFrame(data)

# sirf Age column ke NaN ko mean se fill karna

df['Age']=df["Age"].fillna(df['Age'].mean())

# sirf Salary column ke NaN ko 0 se fill krna

df['Salary']=df['Salary'].fillna(0)

print(df)
```

```python
# 2. Alag alag colmns ke liye alag tarike se ek sath

# har column ke liye rule define karo 

fill_rules={
    "Age":df['Age'].median(),
    "Salary":0,
    "Name":"Unknown"
}

df=df.fillna(value=fill_rules)
print(df)
```

```python
# 3.Conditional filling (column specific)

# maan lo salary missing hai to usako age*2000 se bhrna hai

df['Salary']=df.apply(lambda row:row['Age']*2000 if pd.isna(row['Salary']) else row['Salary'],axis=1)
print(df)

```

```python
# 3.1 Fill based on another column

# we want to fill missing Age with default values based on City

import pandas as pd
import numpy as np

data={
    "Name":['Alice','Bob','Charlie','David'],
    "Age":[25,None,30,None],
    "City":['Delhi','Mumbai','Delhi','Mumbai']
}

df=pd.DataFrame(data)
print(df)

print("--------------")
df.loc[df['City']=='Delhi','Age']=df.loc[df['City']=='Delhi','Age'].fillna(28)
df.loc[df['City']=='Mumbai','Age']=df.loc[df['City']=='Mumbai','Age'].fillna(32)
print(df)


# 2.Using np.where:we can also use np.where for conditional filling.

df['Age']=np.where(df['Age'].isna()&(df['City']=='Delhi'),28,df['Age'])
df['Age']=np.where(df['Age'].isna()&(df['City']=='Mumbai'),32,df['Age'])
df
```

```python
# 4. Diffrent fill for numeric vs categorical filling automatically

# Numeric columns fill with mean

for col in df.select_dtypes(include='number'):
    df[col]=df[col].fillna(df[col].mean())

# string columns->fill with "Unknown"

for col in df.select_dtypes(include='object'):
    df[col]=df[col].fillna("Unknown")
print(df)
```

```python
# interpolate():- ye missing values ko fill karne ka ek smart mathematical tarika hai

# c Basic Linear interpolation:-Missing values ko linearly guess krta hai(previous or next value ke beech average)


import pandas as pd

data={
    "Day":[1,2,3,4,5,6,7],
    "Temperature":[30,None,32,None,36,None,40],
}

df=pd.DataFrame(data)

print("Original:\n",df)

# Linear Interpolation

df["Temperature"]=df["Temperature"].interpolate()
print("\nAfter Linear Interpolation:\n",df)
```

# Q-4: Import Employee.csv file.

```python
import pandas as pd

df = pd.read_csv("Employees_1.csv")
df
```

```python
# 1. Show True/False matrix where missing values are present.

df.isnull()
```

```python
# 2. Find the total number of missing values in each column.

df.isna().sum()
```

```python
# 3. Delete rows where Department is missing.

df = df.dropna(subset=["Department"])
```

```python
# 4. Replace missing salaries with the mean salary.

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df
```

```python
# 5. Replace missing experience with the median value of the column.

df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df
```

```python
# 6. Replace missing department with the most frequent department(mode).

mode_dept = df["Department"].mode(0)
df["Department"] = df["Department"].fillna(mode_dept)
df
```

```python
# 7. Diffrent Fill stretegy.

# -> Fill Salary with mean.

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df

# -> Fill Experience with 0.

df["Experience"] = df["Experience"].fillna(0)
df

# -> Fill Department with "Unknown".

df["Department"] = df["Department"].fillna("Unknown")
```

```python
# 8. Find how many rows have more than 1 missing value.

row_missing_count = df.isna().sum(axis=1)
print(row_missing_count)
```

```python
# 9. Custom Conditional fill
# ->If Salary is missing and Department ="IT", fill with 60000.

df.loc[(df["Salary"].isna()) & (df["Department"] == "IT"), "Salary"] = 60000

# ->If Salary is missing and Department="Finance",fill with 58000.

df.loc[(df["Salary"].isna()) & (df["Department"] == "Finance"), "Salary"] = 58000
```

```python
# 10. Interpolation
# Try Using interpolate() to fill numeric missing values(Salary,Experience)

df[["Salary","Experience"]] = df[["Salary","Experience"]].interpolate()

print(df)
```

# Practice Task:-

Use products.csv file and solve following proble:-

1. Duplicates

->Find how many duplicate rows exist.

->Remove duplicates based on all columns.

->Remove duplicates based only on "ProductName".


2. Rename Columns

->Rename "ProductID"->"ID"

->Rename "ProductName"->"Name"

->Convert all column names to uppercase.


3. Changing Data Types

->"ProductID"->string

->"Price"->int(notice row 107 has "45,000" as string with comma)


```python
import pandas as pd

df = pd.read_csv("products.csv")

print(df)
```

# 1. Duplicates

```python
# ->Find how many duplicate rows exist.

duplicates_count = df.duplicated().sum()
print(duplicates_count)
```

```python
# ->Remove duplicates based on all columns.

df_no_duplicates = df.drop_duplicates()
print(df_no_duplicates)
```

```python
# ->Remove duplicates based only on "ProductName".

df_unique_products = df.drop_duplicates(subset = ["ProductName"])
print(df_unique_products)
```

# 2. Rename Columns

```python
# ->Rename "ProductID"->"ID"

print(df.rename(columns = {"ProductID" : "ID"}))
```

```python
# ->Rename "ProductName"->"Name"

print(df.rename(columns = {"ProductName" : "Name"}))
```

```python
# ->Convert all column names to uppercase.

df.columns = df.columns.str.strip().str.upper()
df
```

```python
# 3. Changing Data Types

# ->"ProductID"->string

print(df["ProductID"].astype(str))

# ->"Price"->int(notice row 107 has "45,000" as string with comma)

df["Price"] = df["Price"].astype(str).str.replace(",","")
df["Price"] = df["Price"].astype(int)
df
```
