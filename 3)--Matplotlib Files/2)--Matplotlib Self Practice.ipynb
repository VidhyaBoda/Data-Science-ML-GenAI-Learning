```python
import matplotlib.pyplot as plt
```

```python
Common Matplotlib Marker Styles


'.': Point
',': Pixel
'o': Circle
'v': Triangle Down
'^': Triangle Up
'<': Triangle Left
'>': Triangle Right
's': Square
'p': Pentagon
'P': Plus (filled)
'*': Star
'h': Hexagon 1
'H': Hexagon 2
'+': Plus
'x': X
'X': X (filled)
'D': Diamond
'd': Thin Diamond
'|': Vertical line
'_': Horizontal line 
```

```python
Matplotlib Linestyles

Solid: '-' or 'solid'
Dashed: '--' or 'dashed'
Dotted: ':' or 'dotted'
Dash-Dot: '-.' or 'dashdot'
None: ' ' or ''
```

```python
Ways to Specify Colors

'b' = blue
'g' = green
'r' = red
'c' = cyan
'm' = magenta
'y' = yellow
'k' = black
'w' = white
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [7,8,9,10,11]


xpoint = [21,22,23,24,25]
ypoint = [5,16,17,18,19]


plt.plot(x,y, marker='D', linestyle = '--', color = 'g', label = 'Data 1')
plt.plot(xpoint,ypoint, marker='o', linestyle = '--', color = 'b', label = 'Data 2')

plt.legend(loc = 'upper left', title = 'Data Group')

plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')

plt.title('Simple Line Chart')

plt.xticks(rotation = 35)
plt.yticks(rotation = 20)

plt.grid() #graph background

plt.show()
```
