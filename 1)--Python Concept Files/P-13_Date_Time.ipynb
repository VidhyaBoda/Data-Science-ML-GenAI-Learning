# 3️⃣ Accessing Date and Time Components

```python
import datetime

now=datetime.datetime.now()
today=datetime.date.today()
print(now)
print(today)
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
```

# 4️⃣ Creating Custom Dates & Times

```python
import datetime

#create specific date
d=datetime.date(2025,10,25)
print(d)

# create specific date and time
dt=datetime.datetime(2025,5,24,10,45,3)
print(dt)
```

# 5️⃣ Date Arithmetic (Add/Subtract Days)

We use timedelta for adding or subtracting time.

```python
import datetime

today=datetime.date.today()
print("Today:",today)

#add 10 days

after_10_days=today+datetime.timedelta(days=10)
print("After 10 Days: ",after_10_days)


# Substracting 10 days

before_10_days=today-datetime.timedelta(days=10)
print("Before 10 Days: ",before_10_days)
```

# 6️⃣ Difference Between Two Dates

```python
import datetime

d1=datetime.date(2025,10,1)
d2=datetime.date(2025,10,30)

diff=d1-d2
print("Diffrence:",diff.days)
```

# 7️⃣ Formatting Dates (strftime())

Convert date/time → string in your desired format.

```python
import datetime

now=datetime.datetime.now()
print(now)
print(now.strftime("%y-%m-%d"))
print(now.strftime("%d-%m-%y"))
print(now.strftime("%B %d,%Y"))
print(now.strftime("%I:%M %p"))
```

# Common strftime() Format Codes

| Code | Meaning          | Example   |
| ---- | ---------------- | --------- |
| `%Y` | Year (4 digits)  | 2025      |
| `%y` | Year (2 digits)  | 25        |
| `%m` | Month (01–12)    | 10        |
| `%B` | Full month name  | October   |
| `%b` | Short month name | Oct       |
| `%d` | Day (01–31)      | 08        |
| `%A` | Day name         | Wednesday |
| `%a` | Short day name   | Wed       |
| `%H` | Hour (24-hour)   | 16        |
| `%I` | Hour (12-hour)   | 04        |
| `%M` | Minutes          | 35        |
| `%S` | Seconds          | 42        |
| `%p` | AM/PM            | PM        |


# 8️⃣ Converting String → Date (strptime())

When you have a string date (e.g., from CSV file), convert it into a datetime object.

```python
import datetime

date_str="2025-10-8"
date_obj=datetime.datetime.strptime(date_str,"%Y-%m-%d")
print(date_str)
print(date_obj)

print(type(date_str))
print(type(date_obj))
```

| Function                     | Description               |
| ---------------------------- | ------------------------- |
| `datetime.datetime.now()`    | Current date & time       |
| `datetime.date.today()`      | Current date only         |
| `datetime.date(y, m, d)`     | Create custom date        |
| `datetime.timedelta(days=…)` | Add/subtract days         |
| `strftime()`                 | Format datetime → string  |
| `strptime()`                 | Parse string → datetime   |
| `datetime.datetime()`        | Create custom date & time |
| `datetime.time()`            | Time object only          |

A date object represents a calendar date (year, month, day) without time.

```python
# Q-10 Print current time in 12-hour format with AM/PM.

import datetime

now=datetime.datetime.now().time()
print(now)

print(now.strftime("%H:%M:%S"))   #If you want clean time (HH:MM:SS) without microseconds

print(now.strftime("%I:%M:%S %p"))
```

```python
# Q-11 Display current day name.

import datetime

now=datetime.date.today()
print(now)
day_name=now.strftime("%A")
day_short_name=now.strftime("%a")
print("Today is:",day_name)
print("Today is:",day_short_name)
```

```python
# Q-12 Add 2 hours and 30 minutes to current time.

from datetime import datetime, timedelta

current_time = datetime.now()

new_time = current_time + timedelta(hours=2, minutes=30)

print("Current Time:", current_time.strftime("%I:%M:%S %p"))
print("After 2 hours 30 minutes:", new_time.strftime("%I:%M:%S %p"))
```
