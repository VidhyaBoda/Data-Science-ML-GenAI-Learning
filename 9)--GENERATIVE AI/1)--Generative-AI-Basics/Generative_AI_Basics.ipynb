## 2. AI → ML → DL → Generative AI\n\n- **AI:** broad field of intelligent systems\n- **ML:** systems learn patterns from data\n- **DL:** ML using deep neural networks\n- **Generative AI:** systems that generate new content

## 3. Foundation Models\n\nFoundation models are broadly trained models that can be adapted for multiple downstream tasks. LLMs are a major category of foundation models.

## 4. Important LLM Terms\n\n**Token:** a unit processed by the model.\n\n**Context window:** the amount of context the model can process for a request.\n\n**Temperature:** a generation control that influences randomness/variation.

```python
def show_generation_settings(temperature):\n    if temperature < 0.4:\n        return 'More predictable / focused generation'\n    elif temperature < 0.8:\n        return 'Balanced generation'\n    return 'More varied / creative generation'\n\nfor temperature in [0.2, 0.7, 1.0]:\n    print(temperature, '->', show_generation_settings(temperature))
```
