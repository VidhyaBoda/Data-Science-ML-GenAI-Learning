# Task 1 — Show Missing Values

```python
print(df.isnull())
```

# Task 2 — Count Missing Values per Column 

```python
print(df.isnull().sum())
```

# Task 3 — Count Total Missing Values 

```python
# Count all missing values in entire DataFrame.

print(df.isna().sum().sum())
```

# Task 4 — Remove All Rows with Missing Values

```python
df = df.dropna()
print(df)
```

# Task 5 — Remove All Columns with Missing Values

```python
df =df. dropna(axis=1)
print(df)
```

# Task 6 — Fill All Missing Values With 0

```python
df_filled = df.fillna(0)
print(df_filled)
```

# Task 7 — Fill Age Column With Mean Age

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

# Task 8 — Fill Salary With Median Salary 

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df)
```

# Task 9 — Fill Missing Names With "Unknown"

```python
df["Name"] = df["Name"].fillna("Unknown")
print(df)
```

# Task 10 — Forward Fill the Dataset

```python
df_ffill = df.fillna(method="ffill")
print(df_ffill)
```

# Task 11 — Backward Fill the Dataset

```python
df_bfill = df.fillna(method="bfill")
print(df_bfill)
```

# Task 12 — Conditional Fill:

```python
# If English is missing → fill with 60
```

```python
df.loc[df['English'].isna(),'English'] = 60
print(df)
```

# Task 13 — Conditional Fill 2:

```python
# If Science is missing → fill with subject average 
# You must calculate mean first.
```

```python
science_mean = df["Science"].mean()
print(science_mean)
```

```python
df["Science"] = df["Science"].fillna(science_mean)
print(df)
```

```python
df.loc[df["Science"].isna(), "Science"] = df["Science"].mean()
print(df)
```

# Task 14 — Fill Each Column Using Different Rules

```python
# Use a dictionary: 
    
rules = { 
        "Age": df['Age'].median(), 
        "Name": "No Name", 
        "Salary": 0 
    } 

df = df.fillna(rules)
print(df)
```

# Task 15 — Simple Report 

```python
# Print: 
#  how many missing values original dataset had 

original_missing = df.isna().sum().sum()
print(original_missing)

#  how many missing values after filling 

after_fill_missing = df.isna().sum().sum()
print(after_fill_missing)

#  number of rows before vs after dropna 

rows_before = df.shape[0]
rows_after = df.dropna().shape[0]

print(rows_before)
print(rows_after)

# (Just print numbers — no advanced logic needed)
```
