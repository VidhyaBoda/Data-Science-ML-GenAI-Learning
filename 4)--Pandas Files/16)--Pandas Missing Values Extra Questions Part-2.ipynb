```python
import pandas as pd
```

```python
data = { 
        "Student": ["Aman", None, "Ranu", "Neha", "Tina", None], 
        "City": ["Delhi", "Mumbai", None, "Surat", None, "Pune"], 
        "Age": [18, 19, None, 20, None, 21], 
        "Python": [85, None, 70, 90, None, 75], 
        "Django": [88, 80, None, None, 92, 85], 
        "Attendance": [None, 80, 95, None, 85, 90] 
}

df = pd.DataFrame(data)

df
```

# Task 1 — Show Which Values Are Missing 
Use: 
df.isnull()

```python
df.isnull()
```

# Task 2 — Count Missing Values Column-wise 
Use: 
df.isnull().sum()

```python
df.isnull().sum()
```

# Task 3 — Count Total Missing Values 
Count total NaN in entire dataset.

```python
df.isnull().sum().sum()
```

# Task 4 — Remove All Rows Containing NaN 
Use: 
df.dropna()

```python
df.dropna()
```

# Task 5 — Remove All Columns Containing NaN 
Use: 
df.dropna(axis=1)

```python
df.dropna(axis=1)
```

# Task 6 — Fill Missing Values With 0 
Use: 
df.fillna(0)

```python
df.fillna(0)
```

# Task 7 — Fill Age Column With Mean 
Replace missing Age with: 
df["Age"].fillna(df["Age"].mean())

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

# Task 8 — Fill Python Marks With Median 
Use: 
df["Python"].fillna(df["Python"].median())

```python
df["Python"] = df["Python"].fillna(df["Python"].median())

df
```

# Task 9 — Fill Missing Student Names With "Unknown" 
Use: 
df["Student"].fillna("Unknown")

```python
df["Student"] = df["Student"].fillna("Unknown")

df
```

# Task 10 — Forward Fill 
Use: 
df.fillna(method='ffill')

```python
df = df.fillna(method='ffill')
df
```

# Task 11 — Backward Fill 
Use: 
df.fillna(method='bfill')

```python
df = df.fillna(method='bfill')
df
```

# Task 12 — Conditional Fill (Attendance) 
If Attendance is missing → fill 75

```python
df.loc[df["Attendance"].isna(),"Attendance"] = 75
df
```

# Task 13 — Conditional Fill (City) 
If City missing → fill with "Unknown City"

```python
df.loc[df["City"].isna(),"City"] = "Unknown City" 
df
```

# Task 14 — Different Rules for Different Column

```python
rules = {
        "Student": "Unknown",
        "City": "Unknown City",
        "Age": df["Age"].mean(),
        "Python": df["Python"].median(),
        "Django": df["Django"].mean(),
        "Attendance": 75
    }

df = df.fillna(rules)

print(df)
```
