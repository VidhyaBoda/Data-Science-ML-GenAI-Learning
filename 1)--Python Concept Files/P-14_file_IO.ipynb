# File I/O in Python

File I/O (Input/Output) means reading data from files and writing data to files.

# 1. Opening a File


file=open("Example.txt","r")

| Mode   | Meaning             | Use                         |
| ------ | ------------------- | --------------------------- |
| `"r"`  | Read mode (default) | File must exist             |
| `"w"`  | Write mode          | Overwrites existing content |
| `"a"`  | Append mode         | Adds data to end            |
| `"r+"` | Read & write        | File must exist             |
| `"w+"` | Write & read        | Overwrites                  |
| `"a+"` | Append & read       | Keeps old data              |


# 2. Reading Data

```python
file=open("customers.csv","r")
file.read() #entire file
file.readline() #read one line
file.readlines() #read all lines in list1

# file.close()
```

# 3. Writing Data

```python
file=open("Students.txt","w")
file.write("shiv")

file=open("live.txt","w")
file.write("my name is sumit.")
file.write("my name is shivansh")

# file.close()

# "w" mode overwrites the file each time.
```

# 4. Appending Data

```python
file=open("Students.txt","a")
file.write("Hello My Name Is Shivansh.\n")
file.write("My age is 22")
file.close()
```

# 5. Using with (Best Practice)

```python
with open("Students.txt","r") as f:
    data=f.read()
    print(data)


    # No need to manually close the file — it auto-closes.
```

# 6. Working with CSV Files

```python
import csv

# writing

with open("data.csv","w",newline="") as f:
    writer=csv.writer(f)
    writer.writerow(["Name","Age"])
    writer.writerow(["Rohit",22])
    writer.writerow(["Shivansh",21])
    writer.writerow(["Mohit",12])

# reading

with open("data.csv","r") as f:
    reader=csv.reader(f)
    for row in reader:
        print(row)
```

```python
# Q-1 Create a text file named intro.txt and write your introduction inside it.

file = open("intro.txt", "w")

file.write("My name is Vidhya Boda.\n")
file.write("I am a student learning Python programming.\n")
file.write("I want to become a Data Analyst.\n")

file.close()
```

```python
# Q-2 Write a program to read the entire content of intro.txt and display it.

with open("intro.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Q-3 Write 3 lines into a file, then read and print them line-by-line.

with open("line.txt", "w") as file:
    file.write("This is line one. \n")
    file.write("This is line two. \n")
    file.write("This is line three. \n")
    
with open("lines.txt", "r") as file:
    for line in file:
        print(line, end="")
```

```python
# Q-4 Check if a file exists before trying to open it.

import os

filename = "intro.txt"

if os.path.exists(filename):
    with open(filename, "r") as file:
        print(file.read())
        
else:
    print("file does not exist")
```

```python
# Q-5 Append one new line to an existing file without deleting old content.

with open("intro.txt", "a") as file:
    file.write("\nThis line is added using append mode.")
    
    
with open("intro.txt", "r") as file:
    print(file.read())
```

```python
# Q-6 Write a program that reads a file and converts all text to uppercase.

with open("intro.txt", "r") as file:
    content = file.read()
    
uppercase_text = content.upper()
print(uppercase_text)
```
