# 05--Vector-Databases\n\nA practical introduction to vector databases, indexing, similarity retrieval, filtering and their role in RAG.

## 1. What is a Vector Database?\n\nA vector database stores vector representations and provides mechanisms for efficient similarity search, often combined with metadata filtering and persistence.

## 2. Vector Record Structure\n\nA practical record commonly contains an ID, vector and metadata. Metadata can identify the source document, page, department, language, timestamp and other filtering attributes.

## 3. Similarity Search\n\nA query vector is compared with stored vectors using a compatible metric. Results are ranked and the top K candidates are returned.

```python
import math\n\ndef cosine_similarity(a, b):\n    dot = sum(x*y for x, y in zip(a, b))\n    na = math.sqrt(sum(x*x for x in a))\n    nb = math.sqrt(sum(y*y for y in b))\n    return dot / (na * nb)\n\nquery = [1, 0, 0]\ncandidate = [0.9, 0.1, 0]\nprint(cosine_similarity(query, candidate))
```

## 4. Metadata Filtering\n\nVector similarity can be combined with structured filters such as department, document type, tenant, language or date.

## 5. ANN and HNSW\n\nApproximate nearest-neighbor indexes reduce search cost by avoiding a full comparison against every vector. HNSW is a widely used graph-based ANN approach.

## 6. Hybrid Search and Reranking\n\nHybrid retrieval can combine semantic similarity with lexical search and metadata filters. A reranker can then improve ordering among retrieved candidates.

## 7. RAG Retrieval Pipeline\n\nQuestion → Query embedding → Vector search → Filtering → Optional reranking → Relevant chunks → LLM.

## 8. Retrieval Evaluation\n\nMeasure retrieval independently using metrics such as Precision@K, Recall@K, MRR and NDCG where appropriate.
