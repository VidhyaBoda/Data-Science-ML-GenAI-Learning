# 1. What is Loss?

Loss measures how wrong the model is.

Example:

Actual:

Cat

Predicted:

Dog

The prediction is wrong.

Therefore loss becomes high.

Goal of training:

Reduce loss as much as possible.


# Why Do We Need Loss Functions?

Without a loss function the model would not know:

- How wrong it is
- Whether weights should change
- How much weights should change

Loss acts as a feedback signal.


# 2. Binary Crossentropy

Used for:

Binary Classification

Examples:

- Cat vs Dog
- Spam vs Not Spam
- Yes vs No

Output Layer:

Dense(1, activation='sigmoid')

Compile:

loss='binary_crossentropy'


# Binary Classification Example

Actual:

Cat = 1

Prediction:

0.95

Loss = Very Small

---

Actual:

Cat = 1

Prediction:

0.10

Loss = Large

Bad predictions create larger losses.


# 3. Categorical Crossentropy

Used for:

Multi-Class Classification

Example:

Cat
Dog
Horse

Output:

Dense(3, activation='softmax')

Compile:

loss='categorical_crossentropy'

Labels must be one-hot encoded.


# One-Hot Encoding Example

Cat   → [1,0,0]

Dog   → [0,1,0]

Horse → [0,0,1]


# 4. Sparse Categorical Crossentropy

Most common for image classification.

Example Labels:

Cat = 0
Dog = 1
Horse = 2

No one-hot encoding needed.

Compile:

loss='sparse_categorical_crossentropy'


# 5. Forward Propagation

Training starts with:

Image
↓
Convolution
↓
ReLU
↓
Pooling
↓
Dense
↓
Softmax
↓
Prediction

This process is called Forward Propagation.


# Example Forward Pass

Input:

Cat Image

Prediction:

Cat = 0.20
Dog = 0.80

Prediction:

Dog

Actual:

Cat

Model is wrong.


# 6. Backpropagation

Question:

How does the model improve?

Answer:

Backpropagation

Backpropagation calculates:

- Which weights caused the error?
- How much should they change?


# Basic Idea

Prediction
↓
Loss
↓
Gradient Calculation
↓
Weight Update
↓
Better Prediction

Repeat thousands of times.


# 7. What is a Gradient?

Gradient tells us:

Which direction should weights move?

Think of a mountain.

Goal:

Reach the lowest point.

Gradient tells the direction of steepest descent.


# Example

Weight:

w = 5

Gradient:

2

Learning Rate:

0.1

Update:

w_new = w_old - learning_rate × gradient

w_new = 5 - (0.1×2)

= 4.8


# 8. Gradient Descent

Gradient Descent repeatedly updates weights.

Process:

1. Predict
2. Calculate Loss
3. Compute Gradient
4. Update Weights
5. Repeat

Goal:

Minimize loss.


# Visual Idea

Top of Mountain
↓
Move Down
↓
Move Down
↓
Move Down
↓
Minimum Loss

Gradient Descent searches for the lowest loss.


# 9. Learning Rate

Learning Rate controls step size.

Small Learning Rate:

0.0001

Advantages:
- Stable

Disadvantages:
- Slow

---

Large Learning Rate:

1.0

Advantages:
- Fast

Disadvantages:
- Can overshoot optimum


# Common Learning Rates

0.1
0.01
0.001
0.0001

Most common:

0.001


# 10. Epoch

Epoch = One complete pass through the dataset.

Example:

Dataset:

10,000 Images

Epoch = 1

Means:

Model has seen all 10,000 images once.


# Multiple Epochs

epochs = 10

Means:

The model sees the entire dataset 10 times.


# 11. Batch Size

Instead of processing all images at once,

CNN processes batches.

Example:

Batch Size = 32

Training:

32 Images
↓
Update Weights

Next 32 Images
↓
Update Weights


# Why Use Batches?

Advantages:

- Less memory
- Faster training
- Better optimization

Common Batch Sizes:

32
64
128
256


# 12. Types of Gradient Descent

Batch Gradient Descent:
- Uses entire dataset

Stochastic Gradient Descent (SGD):
- Uses one sample

Mini-Batch Gradient Descent:
- Uses small batches

Most modern CNNs use mini-batches.


# 13. SGD Optimizer

SGD = Stochastic Gradient Descent

Simple update:

Weight
↓
Gradient
↓
Update

Advantages:
- Simple
- Easy to understand


# SGD Example

optimizer='sgd'

Often requires careful tuning of learning rate.


# 14. Adam Optimizer

Most popular optimizer.

Advantages:

- Fast convergence
- Stable training
- Adaptive learning rates
- Works well on most CNN problems

Common choice:

optimizer='adam'


# Why Adam is Popular

Adam automatically adapts learning rates.

Benefits:

- Less tuning
- Faster convergence
- Better practical performance


# Complete CNN Learning Cycle

Image
↓
Forward Propagation
↓
Prediction
↓
Loss Function
↓
Backpropagation
↓
Gradient Calculation
↓
Optimizer
↓
Weight Update
↓
Repeat

This cycle continues throughout training.
