# 8. Common Problems in ANN

**(i) Overfitting :**

Training Accuracy: 99%

Testing Accuracy: 70%

=> Model memorized data.

Solution:

- Dropout
- More Data
- Regularization

**(ii) Underfitting :**

Training: 60%

Testing: 55%

=> Model too simple.

Solution:

- More Layers
- More Epochs

**(iii) Vanishing Gradient :**

=> Occurs in deep networks.

Mostly with:

- Sigmoid
- Tanh

Solution:

- ReLU
- Batch Normalization

# 9. Dropout (IMP)

=> Randomly turns off neurons.

Example:

10 neurons
Dropout=0.2
2 neurons disabled

=> Prevents overfitting.

# 10. Batch Normalization

=> Normalizes data inside network.

Benefits:

- Faster training
- Stable learning
- Better accuracy
