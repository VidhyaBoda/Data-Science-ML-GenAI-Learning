```python
import numpy as np

X = np.array([
    [1,0,1,0],
    [0,2,0,2],
    [1,1,1,1]
])

WQ = np.array([
    [1,0],
    [0,1],
    [1,0],
    [0,1]
])

print("X Shape :", X.shape)
print("WQ Shape :", WQ.shape)

Q = X @ WQ

print(Q)
print("Q Shape :", Q.shape)
```

```python
Q = X @ WQ
K = X @ WK
V = X @ WV

print("Query")
print(Q)

print("\nKey")
print(K)

print("\nValue")
print(V)
```

```python
scores = Q @ K.T

print(scores)
print(scores.shape)
```

```python
dk = K.shape[1]

scores = scores / np.sqrt(dk)

print(scores)
```

```python
def softmax(x):

    exp = np.exp(x - np.max(x, axis=-1, keepdims=True))

    return exp / np.sum(exp, axis=-1, keepdims=True)
```

```python
weights = softmax(scores)

print(weights)
```

```python
output = weights @ V

print(output)
print(output.shape)





# i love india,python is easy to learn

# easy to
```
