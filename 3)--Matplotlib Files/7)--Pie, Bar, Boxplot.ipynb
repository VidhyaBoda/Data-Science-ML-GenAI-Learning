```python
import matplotlib.pyplot as plt


products=['Apple','Mango','Orange','Cherry']

values=[35,25,15,25]


explode=[0,0,0.6,0]

plt.pie(values,labels=products,autopct='%1.2f%%',explode=explode,shadow=True,startangle=270)
plt.legend()
plt.show()
```
