```python
import matplotlib.pyplot as plt
```

```python
x = [20,30,40,50,60]
y = [25,35,45,55,65]

#s = size,
size = [100,200,300,400,500]
colors = ['yellow','blue','green','red','magenta']
plt.scatter(x,y,s=size,color = colors, edgecolor='black',alpha = 0.7,marker = '*')

plt.colorbar(label = "Color Scale")
plt.title("Scatter Data")

plt.show()
```

```python
x = [20,30,40,50,60]
y = [101,202,303,404,505]


xpoint = [100,200,300,400,500]
ypoint = [25,35,45,55,65]

# colors = ['yellow','blue','green','red','magenta']
plt.scatter(x,y,s=100,color = 'green', edgecolor='black',alpha = 0.7,marker = 'o')

plt.scatter(xpoint,ypoint,s=100,color = 'orange', edgecolor='black',alpha = 0.7,marker = 'D')
plt.colorbar(label = "Color Scale")
plt.title("Scatter Data")

plt.show()
```
