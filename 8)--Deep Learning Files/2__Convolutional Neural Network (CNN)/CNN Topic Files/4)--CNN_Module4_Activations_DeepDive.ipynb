# 1. Why Activation Functions are Needed

Without activation functions, every layer performs only linear operations.

Even a deep neural network would behave like a single linear equation.

Activation functions introduce non-linearity, allowing networks to learn:
- Images
- Faces
- Handwriting
- Speech
- Complex patterns


# 2. Sigmoid Activation Function

Formula:

Sigmoid(x) = 1 / (1 + e^(-x))

Output Range:

0 to 1

Examples:

x = -10 → ~0
x = 0 → 0.5
x = 10 → ~1

Advantages:
- Probability-like output

Disadvantages:
- Vanishing gradient
- Slow training


# 3. Tanh Activation Function

Output Range:

-1 to 1

Examples:

x = -10 → ~ -1
x = 0 → 0
x = 10 → ~ 1

Advantages:
- Zero centered

Disadvantages:
- Still suffers from vanishing gradients


# 4. ReLU (Rectified Linear Unit)

Formula:

ReLU(x) = max(0,x)

Examples:

ReLU(-5) = 0
ReLU(0) = 0
ReLU(8) = 8

Advantages:
- Fast
- Simple
- Reduces vanishing gradients
- Most popular activation in CNNs


# ReLU Example on Feature Map

Before ReLU:

[5, -3, 2]
[-7, 8, -1]

After ReLU:

[5, 0, 2]
[0, 8, 0]

Negative values become zero.


# 5. Vanishing Gradient Problem

During backpropagation, gradients travel backward through the network.

In very deep networks:

Small gradients × Small gradients × Small gradients

become extremely tiny.

Result:

- Early layers stop learning
- Training becomes slow
- Accuracy suffers

Sigmoid and Tanh are most affected.


# Why ReLU Helps

Derivative of ReLU:

For x > 0:

Gradient = 1

This helps gradients flow more easily through deep networks.


# 6. Dying ReLU Problem

Problem:

If a neuron always outputs negative values:

ReLU → 0

Gradient → 0

Weight updates stop.

Neuron becomes inactive forever.

This is called Dying ReLU.


# 7. Leaky ReLU

Solution to Dying ReLU.

Formula:

If x > 0:
    f(x)=x

If x < 0:
    f(x)=0.01x

Examples:

x = -10

Output:

-0.1

The neuron still receives gradients and can continue learning.


# Comparison Table

Sigmoid:
- Range: 0 to 1
- Vanishing Gradient: Yes

Tanh:
- Range: -1 to 1
- Vanishing Gradient: Yes

ReLU:
- Range: 0 to infinity
- Fast: Yes
- Most Popular: Yes

Leaky ReLU:
- Fixes Dying ReLU
- Allows small negative gradients
