```python
import pandas as pd

df=pd.read_csv('House_Price.csv')
# df.isnull().sum()

x=df.drop('price',axis=1)
y=df['price']


from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)



from sklearn.linear_model import LinearRegression


model=LinearRegression()
model.fit(x_train,y_train)


y_pred=model.predict(x_test)


from sklearn.metrics import r2_score,mean_absolute_error

print("R2 Score:",r2_score(y_test,y_pred))


print("Mean Absolute Error:",mean_absolute_error(y_test,y_pred))


# 4925,2,4,20,2,2502.261325193366

predictions=model.predict([[1443,3,4,10,4]])
print("Predicted Price:",predictions)

```

```python
import pandas as pd

data={
    "student_marks":[12,13,15,90,92],
    "result":[0,0,0,1,1]
}


df=pd.DataFrame(data)
df


x=df[['student_marks']]
y=df['result']


from sklearn.linear_model import LogisticRegression


model=LogisticRegression()
model.fit(x,y)

y_pred=model.predict(x)
print(y_pred)


from sklearn.metrics import accuracy_score,confusion_matrix


print("Accuracy Score:",accuracy_score(y,y_pred))
print("Confusuion MAtrix:\n",confusion_matrix(y,y_pred))


```

```python
import pandas as pd



df=pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5],
    "sleep_hours": [7, 6, 5, 4, 3],
    "result": ['Fail','Fail','Fail','Pass','Pass']
})

df


x=df[['study_hours','sleep_hours']]
y=df['result']



from sklearn.linear_model import LogisticRegression

model=LogisticRegression()
model.fit(x,y)


predictions=model.predict([[9, 5]])
print(predictions)



from sklearn.metrics import accuracy_score,confusion_matrix

y_pred=model.predict(x)

print(accuracy_score(y,y_pred))
print(confusion_matrix(y,y_pred))



# TP   FP
# FN   TN

```
