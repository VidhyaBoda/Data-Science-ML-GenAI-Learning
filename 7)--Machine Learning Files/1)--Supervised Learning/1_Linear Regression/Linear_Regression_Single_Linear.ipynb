```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('single_linear.csv')
df
```

```python
plt.plot(df['Area_sqft'], df['Price'], marker = 'o',c = 'purple')
plt.show()
```

```python
x = df[['Area_sqft']]
y = df['Price']

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x,y)
```

```python
prediction = model.predict([[4050]])
print(prediction)
```

```python
print("m:",model.coef_)
print("c:",model.intercept_)
```

```python
from sklearn.metrics import r2_score

y_pred = model.predict(x)
print(y_pred)
```

```python
print("r2_score:", r2_score(y, y_pred))
```
