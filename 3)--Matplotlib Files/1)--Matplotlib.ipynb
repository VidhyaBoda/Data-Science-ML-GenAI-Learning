```python
import matplotlib.pyplot as plt
```

```python
xpoint=[1,2,3,4,5]
ypoint=[2,4,6,8,10]

x1=[12,13,14,15]
y1=[16,12,18,19]



plt.plot(xpoint,ypoint,marker='o',linestyle='--',color='red',label='Group 1')
plt.plot(x1,y1,color='green',marker='^',label='Group 2')
plt.legend(loc='lower right',title='Two Groups')
plt.xlabel('X-AXIS')
plt.ylabel('Y-AXIS')
plt.title("Simple Line Chart")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
