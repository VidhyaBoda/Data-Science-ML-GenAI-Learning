```python

QUESTIONS=[

"What is Artificial Intelligence?",
"What is Machine Learning?",
"What is Deep Learning?",
"What is Python?",
"Who developed Python?",
"What is TensorFlow?",
"What is Keras?",
"What is NumPy?",
"What is Pandas?",
"What is Matplotlib?",
"What is OpenCV?",
"What is Natural Language Processing?",
"What is a Neural Network?",
"What is a Dataset?",
"What is Training Data?",
"What is Testing Data?",
"What is an Epoch?",
"What is Batch Size?",
"What is Learning Rate?",
"What is Overfitting?",
"What is Underfitting?",
"What is Classification?",
"What is Regression?",
"What is Clustering?",
"What is Data Science?",
"What is SQL?",
"What is JSON?",
"What is an API?",
"What is Git?",
"What is Streamlit?"]

ANSWERS=[

"Artificial Intelligence is the field of creating computer systems that can perform tasks requiring human intelligence.",
"Machine Learning is a branch of AI that enables computers to learn patterns from data.",
"Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers.",
"Python is a high-level programming language known for its simplicity and versatility.",
"Python was developed by Guido van Rossum.",
"TensorFlow is an open-source library used for machine learning and deep learning.",
"Keras is a high-level API for building and training neural network models.",
"NumPy is a Python library used for numerical and mathematical operations.",
"Pandas is a Python library used for data analysis and manipulation.",
" Matplotlib is a Python library used for creating graphs and visualizations.",
"OpenCV is an open-source library used for computer vision and image processing.",
"Natural Language Processing is the field of AI that enables computers to understand and process human language.",
"A Neural Network is a computational model inspired by the human brain for learning patterns.",
"A Dataset is a collection of data used for analysis or model training.",
"Training Data is the data used to teach a machine learning model.",
"Testing Data is the data used to evaluate a trained machine learning model.",
"An Epoch is one complete pass of the training dataset through the model.",
"Batch Size is the number of training samples processed before updating the model.",
"Learning Rate determines how much the model's weights are updated during training.",
"Overfitting occurs when a model learns the training data too well and performs poorly on new data.",
"Underfitting occurs when a model fails to learn the patterns in the training data.",
"Classification is a machine learning task that predicts categories or class labels.",
"Regression is a machine learning task used to predict continuous numerical values.",
"Clustering is an unsupervised learning technique that groups similar data points.",
"Data Science is the process of extracting meaningful insights from data.",
"SQL is a programming language used to manage and query relational databases.",
"JSON is a lightweight format used for storing and exchanging structured data.",
"An API is an Application Programming Interface that allows software applications to communicate with each other.",
"Git is a version control system used to track changes in source code.",
"Streamlit is a Python framework used to build interactive web applications for data science and machine learning."]
```

```python
encoder = LabelEncoder()

y= encoder.fit_transform(ANSWERS)

y
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(QUESTIONS)
```

```python
X = tokenizer.texts_to_sequences(QUESTIONS)
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(QUESTIONS)

X = tokenizer.texts_to_sequences(QUESTIONS)

print(X)
```

```python
max_length = max(len(i) for i in X)

X = pad_sequences(

    X,

    maxlen=max_length,

    padding="post"

)

print(X)
```

```python
model = Sequential()
model.add(

    Embedding(

        input_dim=len(tokenizer.word_index)+1,

        output_dim=16

    )

)

model.add(

    LSTM(64)

)

model.add(

    Dense(

        len(np.unique(y)),

        activation="softmax"

    )

)
```

```python
model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)
```

```python
model.fit(

    X,

    y,

    epochs=300

)
```
