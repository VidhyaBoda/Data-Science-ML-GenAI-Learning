# Matplotlib Scatter Plot Tasks


TASK 1

```python
#  Create 20 random numbers for X and Y.
#  Draw a red scatter plot.
#  Set:
# o X label → "X Values"
# o Y label → "Y Values"
# o Title → "My First Scatter"


import matplotlib.pyplot as plt

import numpy as np

x = np.random.randint(50,100,20)
y = np.random.randint(50,100,20)

plt.scatter(x,y, c = 'r')

plt.title("My First Scatter")

plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```python
#  Create 30 random values.
#  Draw a scatter plot with:
# o color = green
# o size = 100
#  Show the plot.

import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100,200,30)
y = np.random.randint(100,200,30)

plt.scatter(x,y, s = 100, c= 'g')
plt.plot(x,y)
plt.show()
```

TASK 3

```python
#  Create two scatter plots on the same graph:
# o First → red
# o Second → blue
#  Use same number of points.
#  Add a legend.

import matplotlib.pyplot as plt
import numpy as np

# Generate same number of points
x = np.arange(1, 11)

x1 = np.random.randint(10,50,10)
x2 = np.random.randint(20,60,10)

plt.scatter(x,x1, c = 'r', label = 'Dataset 1')
plt.scatter(x,x2, c = 'b', label = 'Dataset 2')

plt.legend()

plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.title("Two Scatter Plots Examples")

plt.show()
```

```python
#  Generate random ages and weights.
#  Plot them using scatter.
#  Label:
# o X → Age
# o Y → Weight

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

ages = np.random.randint(100,200,10)
weight = np.random.randint(100,200,10)

plt.scatter(ages,weight)

plt.xlabel("Age")
plt.ylabel("Weight")

plt.title("Age vs Weight Scatter Plot")

plt.show()
```

```python
#  Create a scatter plot with:
# o color = orange
# o size = 80
# o alpha = 0.5
#  Observe the transparency effect.

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

x = np.random.randint(1,50,40)
y = np.random.randint(1,50,40)

plt.scatter(x,y, c = 'orange', s = 80, alpha = 0.5)

plt.xlabel("X-Values")
plt.ylabel("Y-Values")

plt.title("Scatter plot with transparency")

plt.show()
```

```python
# Plot 50 random points.
#  Change:
# o color
# o marker size
#  Display the graph.

import matplotlib.pyplot as plt

import numpy as np

np.random.seed(1)

xpoint = np.random.randint(0,100,50)
ypoint = np.random.randint(0,100,50)

plt.scatter(xpoint,xpoint, c = 'g', marker = 'D', s = 20)

plt.xlabel("X-Values")
plt.ylabel("Y-Values")

plt.title("Scatter Plot - 50 Random Points")

plt.show()
```

TASK 8

```python
#  Create two groups of random points.
#  Plot both using scatter.
#  Add:
# o Title
# o Legend

import matplotlib.pyplot as plt
import numpy as np

x1 = np.random.randint(10,100,25)
x2 = np.random.randint(10,100,25)

y1 = np.random.randint(10,50,25)
y2 = np.random.randint(10,50,25)

plt.scatter(x1,x2, c='purple',label="Group 1")
plt.scatter(y1,y2, c='orange',label="Group 2")

plt.xlabel("X-Values")
plt.ylabel("Y-Values")

plt.title("Two Groups of random points")

plt.legend()

plt.show()
```
