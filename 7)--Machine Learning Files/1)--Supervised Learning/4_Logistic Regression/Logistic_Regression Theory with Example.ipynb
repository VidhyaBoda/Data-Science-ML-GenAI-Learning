# What is Logistic Regression?

Logistic Regression is a classification algorithm, not a regression algorithm (even though its name says "regression").

It is used when the output (target) has only two possible values like:

Yes / No

0 / 1

Male / Female

Spam / Not Spam

✅ Why Can’t We Use Linear Regression for Classification?

Because:

Linear Regression predicts continuous values, while classification needs a probability between 0 and 1.

Linear Regression can give outputs like –5 or 20, but probabilities must be between 0 and 1.

✅ How Logistic Regression Works?
Step 1: Create a Linear Equation

Just like Linear Regression:



z=mX+c

Step 2: Pass it through the Sigmoid Function

The sigmoid function converts any value into 0 to 1.

Sigmoid(z)=
![Screenshot 2025-11-06 102435.png](<attachment:Screenshot 2025-11-06 102435.png>)


This gives us a probability like:

0.85 → 85% chance it belongs to class 1

0.20 → 20% chance → class 0

✅ How Do We Decide Class?

We use a threshold = 0.5 by default:

| Probability | Output Class |
| ----------- | ------------ |
| ≥ 0.5       | Class 1      |
| < 0.5       | Class 0      |


✔ Email Spam Detector (Spam = 1, Not Spam = 0)
✔ Tumor Detection (Malignant = 1, Benign = 0)
✔ Loan Approval (Approve = 1, Reject = 0)

```python
import pandas as pd

df=pd.read_csv('titanic.csv')

# df.isnull().sum()

df['Age']=df['Age'].fillna(df['Age'].mean())
df.head()

df['Sex']=df['Sex'].map({'male':0,'female':1})
df.head()


x=df[['Pclass','Sex','Age','Fare']]
y=df['Survived']

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)


from sklearn.linear_model import LogisticRegression

model=LogisticRegression(max_iter=5000)
model.fit(x_train,y_train)

y_pred=model.predict(x_test)

from sklearn.metrics import accuracy_score

print("Accuracy",accuracy_score(y_test,y_pred))
```
