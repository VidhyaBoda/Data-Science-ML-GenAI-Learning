<h1>TASK 1</h1>
<h3>Objective:</h3>
<h4>For the givencsv, write a python program satisfying following tasks to demonstrate application of machine learning through 
multiple linear regression as follows – </h4>
<h3>Dataset:</h3>
<h4>RealEstate.csv</h4>
<h3>Steps to Complete the Task:</h3>
<h4>
1. Import required libraries.<br>
2. Load RealEstate dataset, create a dataframe and check datatypes of its attributes using appropriate method.<br>
3. Remove ‘No’ column from the dataframe.<br>
4. Check for any null values in features using appropriate method.<br>
5. Create feature variables x and y as given above.<br> 
6. Create training and testing sets of feature variables with 70% of data for training and with random state of 110.<br>
7. Create and fit regression model using appropriate method.<br>
8. Use testing set created in step 6 to find and print the prediction of the outcome.<br>
9. Find and print coefficient and mean squared error of the regression model.</h4>

```python
# Step - 1 (Import Reguired Libraries)
import numpy as np
import pandas as pd
# Step - 2 (Read the csv and check the values. Also if data is categorical then create dummies and remove unnecessary columns)
df = pd.read_csv("RealEstate.csv")
print(df.head())
print(df.info())
df.drop("No",axis=1,inplace=True)
print(df.head())
print(df.info())
print(df.isna().sum())
# Step - 3 (Extract feature and target value)
x = df.drop("Y house price of unit area",axis=1)
y = df["Y house price of unit area"]
print(x.shape)
print(y.shape)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=110)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
# Step - 4 (Train the model. Also check for coefficient snd intercept value)
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
coefficient = model.coef_
intercept = model.intercept_
print("Slope:",coefficient)
print("Intercept:",intercept)
# Step - 5 (Predict the value for testing part)
y_pred = model.predict(x_test)
# Step - 6 (Compare the testing value to the predicted values)
compare = pd.DataFrame({"Actual":y_test,"Predicted":y_pred})
print(compare)
# Step - 7 (Check for the accuracy parameters)
from sklearn import metrics
print("Accuracy: ",metrics.r2_score(y_test,y_pred))
print("MAE: ",metrics.mean_absolute_error(y_test,y_pred))
print("MSE: ",metrics.mean_squared_error(y_test,y_pred))
```

```python
# Step - 1 (Import Reguired Libraries)
import numpy as np
import pandas as pd
# Step - 2 (Read the csv and check the values. Also if data is categorical then create dummies and remove unnecessary columns)
df = pd.read_csv("winequality.csv")
print(df.head())
print(df.info())
df.drop("Id",axis=1,inplace=True)
print(df.head())
print(df.info())
print(df.isna().sum())
# Step - 3 (Extract feature and target value)
x = df.drop("quality",axis=1)
y = df["quality"]
print(x.shape)
print(y.shape)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
# Step - 4 (Train the model. Also check for coefficient snd intercept value)
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
coefficients = model.coef_
intercept = model.intercept_
print(coefficients)
print(intercept)
# Step - 5 (Predict the value for testing part)
y_pred = model.predict(x_test)
# Step - 6 (Compare the testing value to the predicted values)
compare = pd.DataFrame({"Actual":y_test,"Predicted":y_pred})
print(compare)
x_ans = [[8,0.4,0.40,15,0.048,40,150,0.99,3,0.45,10.5]]
y_ans = model.predict(x_ans)
# Step - 7 (Check for the accuracy parameters)
from sklearn import metrics
print("Accuracy: ",metrics.r2_score(y_test,y_pred))
print("MAE: ",metrics.mean_absolute_error(y_test,y_pred))
print("MSE: ",metrics.mean_squared_error(y_test,y_pred))
print("Predicted value for the given values in question is: ",y_ans)
```

<h1>TASK 3</h1>
<h3>Objective:</h3>
<h4>Consider variables x and y created from a pandas dataframe  “car.csv” .</h4>
<h3>Dataset:</h3>
<h4>car.csv</h4>
<h3>Steps to Complete the Task:</h3>
<h4>
Create new column named “Age_car” (Age_car=2023-year) <br>
For multiple linear regression problem, x contains the independent variables ( Age_car , Driven_kms , Fuel_Type , Selling_type , Transmission )  and y contains the dependent (Selling_Price) variable which is to be predicted.<br>
Write a Python program to spilt x and y into training and testing datasets with a 20% split. Then create a multiple linear regression model using the training data and print its coefficients ,intercept and mean squared error</h4>

```python
# Step - 1 (Import Reguired Libraries)
import numpy as np
import pandas as pd
# Step - 2 (Read the csv and check the values. Also if data is categorical then create dummies and remove unnecessary columns)
df = pd.read_csv("car.csv")
print(df.head())
print(df.info())
print(df.isna().sum())
df.drop("Car_Name",axis=1,inplace=True)
df.drop("Owner",axis=1,inplace=True)
df["Age"] = 2025 - df["Year"]
df.drop("Year",axis=1,inplace=True)
df = pd.get_dummies(df,drop_first=True)
print(df.head())
# Step - 3 (Extract feature and target value)
x = df.drop("Selling_Price",axis=1)
y = df["Selling_Price"]
print(x.shape)
print(y.shape)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
# Step - 4 (Train the model. Also check for coefficient snd intercept value)
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
coefficients = model.coef_
intercept = model.intercept_
print(coefficients)
print(intercept)
# Step - 5 (Predict the value for testing part)
y_pred = model.predict(x_test)
# Step - 6 (Compare the testing value to the predicted values)
compare = pd.DataFrame({"Actual":y_test,"Predicted":y_pred})
print(compare)
# Step - 7 (Check for the accuracy parameters)
from sklearn import metrics
print("Accuracy: ",metrics.r2_score(y_test,y_pred))
print("MAE: ",metrics.mean_absolute_error(y_test,y_pred))
print("MSE: ",metrics.mean_squared_error(y_test,y_pred))
```
