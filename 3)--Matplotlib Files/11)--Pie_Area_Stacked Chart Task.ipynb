# Task 1 (Very Easy)

```python
# Create a pie chart showing favorite fruits of students:
#  Labels: Apple, Banana, Mango, Orange
#  Values: 20, 30, 25, 25
#  Show percentage
#  Add legend at upper right

import matplotlib.pyplot as plt

Fruits = ['Apple', 'Banana', 'Mango', 'Orange']

Values = [20, 30, 25, 25]

plt.pie(Values,labels = Fruits,autopct = '%0.1f%%')

plt.legend(loc = 'upper right')

plt.show()
```

```python
# Create a pie chart of monthly expenses:
#  Rent, Food, Travel, Savings
#  Any values of your choice
#  Explode Savings
#  Start angle = 180
#  Add shadow

import matplotlib.pyplot as plt

expenses = ['Rent','Food','Travel','Savings']

values = [100,150,200,250]

e = [0, 0, 0, 0.2]

plt.figure(figsize=(5,5))

plt.pie(values, labels = expenses, explode = e, shadow = True, startangle = 180)

plt.title("Monthly Expenses Distribution")

plt.show()
```

```python
# Create an area chart for daily sales:
#  X-axis: Days (1 to 7)
#  Y-axis: Any sales values
#  Set transparency (alpha)
#  Add title and labels

import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
sales = [1200, 1500, 1700, 1300, 1800, 2000, 1600]

plt.plot(days, sales, color = 'black')

plt.fill_between(days, sales, alpha=0.5)

plt.title("Daily Sales Report")

plt.xlabel("Days")
plt.ylabel("Sales Amount")

plt.show()
```

# Task 9 (Thinking)


```python
# Create a stacked area chart where:
#  Legend is at upper left
#  Labels are meaningful
#  Chart shows total contribution clearly

import matplotlib.pyplot as plt

import numpy as np

days = [1,2,3,4,5]

math_hours =    [2, 3, 4, 3, 5]
science_hours = [1, 2, 2, 3, 2]
english_hours = [0, 1, 2, 2, 3]

total_hours = np.array(math_hours) + np.array(science_hours) + np.array(english_hours)

plt.stackplot(days, math_hours, science_hours, english_hours, labels = ['Math study hours','Science study hours','English study hours'],alpha = 0.7)

plt.plot(days,total_hours,linestyle='--',linewidth = 2)

plt.legend(loc='upper left')

plt.title("Total Daily Study Contribution by Subject")

plt.xlabel("Day Number")
plt.ylabel("Total Study hours")

plt.grid(alpha=0.3)

plt.show()
```

```python
# Create all three charts using:
#  One dataset
#  Different titles
#  Proper labels
#  Neat legend placement
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Common dataset
days = [1, 2, 3, 4, 5]
product_A = [100, 120, 140, 130, 150]
product_B = [80, 90, 100, 95, 110]
product_C = [60, 70, 75, 85, 95]

# Total sales per product (for pie chart)
total_A = sum(product_A)
total_B = sum(product_B)
total_C = sum(product_C)

# Total per day (for area chart)
total_daily = np.array(product_A) + np.array(product_B) + np.array(product_C)
```

```python
plt.figure()

plt.pie([total_A, total_B, total_C],
        labels=['Product A', 'Product B', 'Product C'],
        autopct='%1.1f%%')

plt.title("Overall Sales Contribution by Product")
plt.show()
```

```python
plt.figure()

plt.fill_between(days, total_daily, alpha=0.6)
plt.plot(days, total_daily, linestyle='--')

plt.title("Total Sales Trend Over Days")
plt.xlabel("Days")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure()

plt.stackplot(days,
              product_A,
              product_B,
              product_C,
              labels=['Product A', 'Product B', 'Product C'],
              alpha=0.7)

plt.legend(loc='upper left')

plt.title("Daily Sales Contribution by Product")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
```
