# 1.Mean()->Average

formula:

Mean=sum of all values/Number of values

```python
import pandas as pd

data={
    'Math':[70,80,90,60,50,50],
    'Science':[88,92,95,65,80,80]
}

df=pd.DataFrame(data)

print("Math Mean:",df['Math'].mean())
print("Science Mean:",df['Science'].mean())
```

# 2. Median()->Middle Value

sorted data ke beech wala element.

odd count->beech vala ek value.

Even count->do beech ke average.

```python
print("Math Median:",df['Math'].median())
print("Science Median:",df['Science'].median())
```

# 3.mode()->Most Frequent Value

sabse jyada repeat hone vala number.

```python
print('Math Mode:',df['Math'].mode())
print("Science Mode:",df['Science'].mode())
```

```python
# 1. Create a DataFrame with columns:English,Math,Science.Find mean,median,and mode for each.

import pandas as pd

data = {
        'English' : [20,20,30,35,40,50],
        'Math' : [50,55,60,65,50,75],
        'Science' : [80,85,80,90,95,100]
    }

df = pd.DataFrame(data)
print(df)

print("----------------------------------")
print('English Mean:',df['Math'].mean())
print('English Median:',df['Math'].median())
print('English Mode:',df['Math'].mode())

print("----------------------------------")

print('Math Mean:',df['Math'].mean())
print('Math Median:',df['Math'].median())
print('Math Mode:',df['Math'].mode())

print("----------------------------------")

print('Science Mean:',df['Math'].mean())
print('Science Median:',df['Math'].median())
print('Science Mode:',df['Math'].mode())
```

```python
# 2. Create a DataFrame with salaries of 10 employees.Find the average,median,and most common salary.

import pandas as pd

Salaries = [35000, 42000, 50000, 42000, 60000, 
            50000, 45000, 50000, 70000, 42000]

df = pd.DataFrame({
    "Salary": Salaries}
)
print(df)

print("-------------")

average_salary = df['Salary'].mean()
print('Avarage Salary :', average_salary)

median_salary = df['Salary'].median()
print('Median Salary:', median_salary)

mode_salary = df['Salary'].mode()
print('Most common salary:', mode_salary)
```

```python
# 3. Generate marks of students in 5 subjects randomly(use pd.DataFrame).Calculate the subject-wise mean,median,and mode.

import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    "Math": np.random.randint(40, 100, 10),
    "Science": np.random.randint(40, 100, 10),
    "English": np.random.randint(40, 100, 10),
    "Computer": np.random.randint(40, 100, 10),
    "History": np.random.randint(40, 100, 10)
}

df = pd.DataFrame(data)

print(df)

print("----------------")

mean_marks = df.mean()
print("Subject-wise Mean:\n", mean_marks)

print("----------------")

median_marks = df.median()
print("Subject-wise Median:\n", median_marks)

print("----------------")

mode_marks = df.mode()
print("Subject-wise Mode:\n", mode_marks)
```

```python
# 4. Create a DataFrame of house prices,find the mean price,median price and mode price.

import pandas as pd

prices = [45, 60, 75, 60, 90, 
          120, 75, 60, 150, 200]

df = pd.DataFrame({
    "House_Price_Lakhs": prices
})

print(df)

print("---------------")

# Mean Price
mean_price = df["House_Price_Lakhs"].mean()

# Median Price
median_price = df["House_Price_Lakhs"].median()

# Mode Price
mode_price = df["House_Price_Lakhs"].mode()

print("Mean Price:", mean_price)
print("Median Price:", median_price)
print("Mode Price:", mode_price)0
```

```python
# 1: Student Marks
data={ 
    "Name":["Aman","Riya","Kabir","Sneha","Vikram"], 
    "Math":[78,85,92,66,74], 
    "Science":[81,79,95,72,88], 
    "English":[69,91,84,77,80] 
}

df = pd.DataFrame(data)
print(df)

print("---------------")

# Q-1: Subject-wise minimum marks(column-wise).

subject_min = df[["Math", "Science", "English"]].min()
print("Subject-wise Minimum Marks:\n", subject_min)

print("---------------")

# Q-2: Each student ka maximum marks(row-wise).

student_max = df[["Math", "Science", "English"]].max(axis=1)
print("Each Student Maximum Marks:\n", student_max)
```

```python
# 2: Employee Salary

import pandas as pd
data={ 
      "EmpID":[101,102,103,104,105,106], 
      "Name":["Raj","Simran","Amit","Neha","Vivek","Anjali"], 
      "Department":["HR","IT","Finance","HR","IT","Finance"], 
      "Salary":[35000,50000,42000,37000,60000,39000] }

df = pd.DataFrame(data)

print(df)


print("----------------")
# Q-1 Department-wise minimum salary.

dept_min_salary = df.groupby("Department")["Salary"].min()

print("Department-wise Minimum Salary:\n")

print(dept_min_salary)

print("----------------")
# Q-2 Department-wise maximum salary.

dept_max_salary = df.groupby("Department")["Salary"].max()

print("Department-wise Maximum salary:\n")

print(dept_max_salary)
```

```python
# 3: Daily Temperatures

import pandas as pd

data={ 
        "Day":["Mon","Tue","Wed","Thu","Fri","Sat","Sun"], 
        "Temperature":[32,35,30,28,33,31,29] 
    }

df = pd.DataFrame(data)

print(df)

print("-----------------")
# Q-1 Find coldest day(minimum temerature).

min_temp = df["Temperature"].min()
print("Minimum Temperature:",min_temp)

print("-----------------")

coldest_day = df.loc[df["Temperature"].idxmin()]
print("Coldest Day Details:\n", coldest_day)


print("-----------------")
# Q-2 Find hottest day(maximum temperature).

max_temp = df["Temperature"].max()
print("Minimum Temperature:",max_temp)

print("-----------------")

hottest_day = df.loc[df["Temperature"].idxmax()]
print("Hottest Day Details:\n", hottest_day)
```

# std()->Standard Deviation

yeh ek statistical operation hai jo batata hai ki tumhare data ke values mean se kitane spread (door-pass)hai.

agr std() chhota hai->data values mean ke aas-paas hai.

agr std() bada hai-> data values mean se jyada door door hai.

```python
import pandas as pd

data={
    "Student":['A','B','C','D','E'],
    "Math":[85,90,78,92,88],
    "Science":[80,85,82,95,89]
}

df=pd.DataFrame(data)

print("Math STD:",df['Math'].std())
print("Science STD:",df['Science'].std())

# iska matlab->Maths ke marks mean ke as pas jyada clustered hai.science ke marks thoda jyada spread hai.
```

# Variance(var())

variance batata hai ki tumhre data ke values mean se kitane spread hai.

->ye standard deviation ka square hota hai.

->Agar variance 0 ke close hai->saare values lagbhag same hai.

->Agar variance bada hai -> values mean se door door bikhra hue hai.

formula=
Variance=(std)**2
std=root(Variance)

```python
import pandas as pd

data={
    "Student":['A','B','C','D','E'],
    "Math":[85,90,78,92,88],
    "Science":[80,85,82,95,89]
}

df=pd.DataFrame(data)

print("Math Variance:",df['Math'].var())
print("Science Variance",df['Science'].var())
```

# corr()->Correlation

Correlation ek statistical measure hai jo dikhata hai ki do columns ke beech realtionship(association)kitna strong hai aur kaisa hai.

->Value hamesha -1 se +1 ke beech ati hai.

+1->Strong Positive Relation
-1->Strong Negative Relation
0->No relation.

```python
import pandas as pd

data={
    "Height":[150,160,170,180,190],
    "Weight":[50,55,65,75,85],
    "Age":[10,15,20,25,30]
}

df=pd.DataFrame(data)

# correlation matrix

print(df.corr())

# matlab height or weight ke beech strong positive correlation hai(0.99)
```

# Q-1:

```python
data={ 
        "Maths":[78,85,92,88,76,95,89], 
        "Science":[82,79,91,85,80,96,90], 
        "English":[70,75,80,72,78,85,77], 
        "Age":[14,15,16,14,15,16,17] 
    }

df = pd.DataFrame(data)

print(df)
```

```python
# Q-1 Find out the standard deviation of each subject(Maths,Science,English). kis subject ke marks sabse jyada spread(variability) mein hai.

std_values = df[["Maths","Science","English"]].std()
print("Standard Deviation:\n", std_values)
# 👉 Maths has highest standard deviation.
```

```python
# Q-2 Calculate the variance of Science subject. Agar variance jyada hai to iska kya matlab hai?

science_variance = df["Science"].var()
print("Science Variance:", science_variance)
# If variance is high:
# ->Scores are widely spread from the average.
# ->Some students are scoring very high, some much lower.
# ->Indicates uneven performance.
```

```python
# Q-3 (A)Check the correlation between Maths and Science. Agar value+1 ke karib hai to iska kya matlab hoga?

correlation = df["Maths"].corr(df["Science"])
print("Correlation (Maths vs Science):", correlation)
# If correlation ≈ +1:
# ->When Maths increases → Science also increases
# ->Strong positive relationship
# ->Students good in Maths are usually good in Science
```

```python
# Q-4 
# (A) Ek naya column banao "Result" jisme likha ho: 
# ->"Pass" agar student ke Maths marks>=80 hai.
# ->"Fail" agar Maths marks<80 hai

df["Result"] = df["Maths"].apply(lambda x: "Pass" if x >= 80 else "Fail")

# (B) Ek naya column "AgeGroup" banao:
# ->"Teen" agar Age<16
# ->"Young" agar Age>=16

df["AgeGroup"] = df["Age"].apply(lambda x: "Teen" if x < 16 else "Young")

# (C) apply() use karke ek naya column "Total" banao jisme Maths+Science+English ka sum ho.

df["Total"] = df.apply(
    lambda row: row["Maths"] + row["Science"] + row["English"],
    axis=1
)

print(df)
```

# Q-2:

```python
import pandas as pd

data={ 
        "Height":[150,160,165,170,180,175,155], 
        "Weight":[50,55,60,65,80,75,52], 
        "Marks":[70,85,90,60,75,95,88], 
        "Age":[14,15,16,15,17,16,14] 
    }

df = pd.DataFrame(data)

print(df)
```

```python
# 1.
# (A) Find the std of Height,Weight and Marks.

std_values = df[["Height","Weight","Marks"]].std()
print("Standard Deviation:\n", std_values)


# (B) Which column has the highest variability(spread)?

# 👉 Marks has highest standard deviation (~12.31).
```

```python
# 2. Calculate the variance of Marks.

marks_variance = df["Marks"].var()
print("Variance of Marks:", marks_variance)
```

```python
# 3.(A) Find the correlation between Height and Weight.

height_weight_corr = df["Height"].corr(df["Weight"])
print("Correlation (Height vs Weight):", height_weight_corr)

# ->Do taller student genrally have higher weight?
# ✔ Yes — strongly.
# Correlation ≈ 0.96, which is very close to +1.

print("---------------")

# (B) Find the correlation between Marks and Age.

marks_age_corr = df["Marks"].corr(df["Age"])
print("Correlation (Marks vs Age):", marks_age_corr)

# ->Do older students score higher?
# ⚠ Not strongly.
# Correlation ≈ 0.44
```

```python
# 4.(A) Create a new column "BMI" using the formula:
# BMI=Weigth/(Height/100)**2

df["BMI"] = df["Weight"] / (df["Height"]/100)**2
print(df)

print("----------------")

# (B) Create a new column "Performace":
# ->"Excellent" if Marks>=85
# ->"Good" if Marks between 70 and 84
# ->"Average" if Marks<70

import numpy as np
conditions = [
    df["Marks"] >= 85,
    (df["Marks"] >= 70) & (df["Marks"] <= 84),
    df["Marks"] < 70
]

choices = ["Excellent", "Good", "Average"]

df["Performance"] = np.select(conditions, choices)

print(df)

print("----------------")

# (C) Create a new column "Category":
# ->"Minor" if Age<16
# ->"Adult" if Age>=16

df["Category"] = np.where(df["Age"] < 16, "Minor", "Adult")

print(df)
```
