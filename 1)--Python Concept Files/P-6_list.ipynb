# 🧩 1. What is a List?

👉 Definition:
A list is a collection of ordered, changeable (mutable) items in Python.
It can store different data types — like numbers, strings, or even other lists.


✅ Lists are written in square brackets [ ]

✅ Lists are mutable → You can change, add, or remove items.

```python
fruits=["Apple","Mango","Bannana","Orange"]
numbers=[1,2,3,4,5]
mixed=["Sumit",22,"Java",12,True]

print(fruits)
print(numbers)
print(mixed)
```

# 2.Accessing List Elements

Lists are index-based (start from 0).

```python
names=["Shivansh","Suraj","Mohit"]
print(names[2])
print(names[0])
```

# ✂️ 4. Slicing Lists

You can extract part of a list using list[start:end:step].

```python
nums=[10,20,30,40,50,60]
print(nums[1:4:2])
print(nums[:3])
print(nums[::-1])
```

# 🔁 5. Changing List Elements (Mutable Property)

```python
colors=["red","green","blue"]
colors[1]="yellow"
print(colors)
```

# ➕ 6. Adding Elements

Method	                 Description	                 Example
append()	             Adds at the end 	     fruits.append("orange")
insert(index, item)	 Adds at specific position	fruits.insert(1, "kiwi")
extend()              Add multiple items	         fruits.extend()


```python
fruits = ["apple", "banana"]
# fruits.append("cherry")
# fruits.insert(1, "kiwi")
fruits.extend(["mango", "grapes"])
print(fruits)

fruits.remove("apple")
print(fruits)
```

# 🗑️ 7. Removing Elements

Method	                 Description	              Example

remove(value)	  Removes first occurrence of value 	fruits.remove("apple")
pop(index)	Removes by index (default last)	fruits.pop(2)
del list[index]	Delete by index	del fruits[1]
clear()	Remove all elements	fruits.clear()

# 8. Searching / Membership

```python
fruits = ["apple", "banana", "mango"]
print("banana" in fruits)   # True
fruits.remove("banana")
print("apple" not in fruits) # False
print("banana" in fruits)   # True

list1=[10,20,5]
print(max(list1))
```

```python
# Q-12 Remove duplicate numbers. 

nums = [1, 2, 2, 3, 4, 4, 5]

unique_nums = []
for n in nums:
    if n not in unique_nums:
        unique_nums.append(n)

print(unique_nums)
```
